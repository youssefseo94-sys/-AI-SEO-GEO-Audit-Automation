```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
SEO Audit (Technical + On-Page) from sitemap.xml  |  فحص SEO تقني وفحص صفحات من خريطة الموقع

- Reads robots.txt and sitemap.xml (including sitemap index files)
- Checks pages one by one, discovers extra pages via internal links, finds broken links and orphan pages
- Optional Google PageSpeed test (--psi N) and a warning for JavaScript-rendered pages
- Gives a score out of 100 (higher = better)
- Exports an Excel report: issues, how to fix them, and where to fix them
- Report language: Modern Standard Arabic (--lang ar) or English (--lang en)

Install:
    pip install requests beautifulsoup4 pandas openpyxl

Usage:
    python seo_audit_v3.py https://example.com
    python seo_audit_v3.py https://example.com --lang en --max 200 --out report.xlsx
    python seo_audit_v3.py https://example.com --max 0 --crawl 200   (all sitemap pages + 200 discovered pages)
    python seo_audit_v3.py https://example.com --psi 5                (Google PageSpeed on 5 pages; default 1 = homepage; 0 = off)
"""
import argparse
import gzip
import re
import sys
import time
from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from urllib.parse import urldefrag, urljoin, urlparse
from urllib.robotparser import RobotFileParser

import pandas as pd
import requests
from bs4 import BeautifulSoup

# ======================================================================
# 👇👇👇  هنا كتبدل رابط الموقع اللي بغيتي تفحصو (الموقع الحالي: odysseedelaterre.fr)
#         CHANGE THE WEBSITE TO AUDIT HERE (the only line you need to edit)
# ======================================================================
SITE_URL = "https://seo-geo-insights.blogspot.com/"

# 👇 لغة التقرير: "ar" = عربية فصحى  |  "en" = English
#    REPORT LANGUAGE: "en" = Arabic  |  "en" = English
REPORT_LANG = "ar"

# 👇 (اختياري) مفتاح Google PageSpeed. خليه فارغ "" إلا ما عندكش
PSI_API_KEY = "zaSyD0XbcuK74WPFo0umfMJTEHas5ppSDm3nkAI"
# ======================================================================
# (إلا كتبتي رابط فالأمر %run هو اللي كيتطبق، وإلا ماكتبتيش كيتطبق SITE_URL)

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/124.0 Safari/537.36 SEOAuditBot/1.0"
}

# Points deducted from a page's score for each issue, by severity
WEIGHTS = {"Critical": 25, "High": 12, "Medium": 5, "Low": 2, "Info": 0}
SEV_ORDER = {"Critical": 0, "High": 1, "Medium": 2, "Low": 3, "Info": 4}
SEVLBL = {
    "ar": {"Critical": "حرج", "High": "عالٍ", "Medium": "متوسط", "Low": "منخفض", "Info": "ملاحظة"},
    "en": {"Critical": "Critical", "High": "High", "Medium": "Medium", "Low": "Low", "Info": "Info"},
}

LANG = "ar"

# ------------------------------------------------------------------ النصوص / UI strings
S = {
    "ar": {
        "chars": "{n} حرفًا", "words": "{n} كلمة", "sec": "{n} ثانية",
        "robots_dis": "ممنوعة بقاعدة Disallow في robots.txt",
        "chain": "{n} عمليات إعادة توجيه ← {u}",
        "redir": "يعيد التوجيه إلى {u}",
        "img": "{a} من أصل {b} صورة بلا نص بديل",
        "mixed": "{n} ملفًا يُحمَّل عبر http",
        "yes": "نعم", "no": "لا",
        "g90": "ممتاز", "g75": "جيد", "g50": "متوسط (يحتاج إلى تحسينات)", "g0": "ضعيف (مشكلات كبيرة)",
        "sheets": ["الملخص", "أهم الإصلاحات", "كل المشكلات", "الصفحات", "الروابط المكسورة", "السرعة"],
        "status_n": "رمز الاستجابة {n}",
        "page_cols": {"URL": "الرابط", "Status": "الحالة", "Final URL": "الرابط النهائي", "Redirects": "عدد إعادات التوجيه",
                      "Response (s)": "زمن الاستجابة (ثانية)", "Size (KB)": "الحجم (كيلوبايت)", "Title": "عنوان الصفحة",
                      "Title len": "طول العنوان", "Meta description": "وصف الميتا", "Desc len": "طول الوصف",
                      "H1": "العنوان الرئيسي H1", "H1 count": "عدد H1", "Canonical": "Canonical", "Noindex": "Noindex",
                      "Images": "عدد الصور", "Images no alt": "صور بلا نص بديل", "Internal links": "روابط داخلية",
                      "External links": "روابط خارجية", "Words": "عدد الكلمات", "Score": "التقييم (من 100)",
                      "Issues": "عدد المشكلات"},
        "bl_cols": ["الصفحة المصدر", "الرابط المكسور", "الحالة", "النوع", "نص الرابط"],
        "int": "داخلي", "ext": "خارجي",
        "speed_cols": ["الرابط", "الجهاز", "الأداء (من 100)", "FCP (ثانية)", "LCP (ثانية)", "CLS", "TBT (مللي ثانية)",
                       "TTFB (مللي ثانية)", "تقييم المستخدمين الحقيقيين (28 يومًا)", "INP للمستخدمين (مللي ثانية)"],
        "broken_n": "{n} رابط مكسور",
        "js_d": "نص قليل مع عناصر JavaScript كثيرة",
        "psi_d": "الأداء {s}/100",
        "crawl_found": "صفحات أُضيفت للفحص عبر الروابط الداخلية (خارج الخريطة)",
        "links_checked": "عدد الروابط التي جرى فحصها",
        "broken_total": "عدد الروابط المكسورة",
        "orphan_skip": "تم تخطي كشف الصفحات اليتيمة لأن الفحص لم يشمل كل الصفحات. استخدم --max 0 مع --crawl أكبر.",
        "p_crawl": "🕸️ تم اكتشاف {n} صفحة إضافية عبر الروابط الداخلية، جارٍ فحصها ...",
        "p_links": "🔗 جارٍ فحص {n} رابط بحثًا عن الروابط المكسورة ...",
        "p_psi": "⚡ جارٍ فحص السرعة عبر Google PageSpeed ({n} صفحة، على الجوال والكمبيوتر، قد يستغرق عدة دقائق) ...",
        "dev_mobile": "جوال", "dev_desktop": "كمبيوتر",
        "p_psi_err": "⚠️ تعذّر فحص PageSpeed ({d}) للرابط {u} ({e}). إن كان الخطأ 429 فالحصة المجانية استُنفدت: جرّب لاحقًا أو استعمل --psi-key، أو افحص الرابط يدويًا في pagespeed.web.dev",
        "sum_cols": ["البند", "القيمة"],
        "top_cols": ["الخطورة", "المشكلة", "عدد الصفحات المتأثرة", "% من الصفحات", "طريقة الإصلاح", "مكان الإصلاح"],
        "iss_cols": ["الرابط", "الخطورة", "المشكلة", "التفاصيل", "طريقة الإصلاح", "مكان الإصلاح", "الرمز"],
        "site": "الموقع", "sitemap": "خريطة الموقع (Sitemap)", "sm_total": "عدد الروابط في الخريطة",
        "checked": "عدد الصفحات المفحوصة", "score": "التقييم العام (من 100)", "verdict": "الحكم",
        "issues_of": "عدد المشكلات ({s})",
        "note_l": "ملاحظة",
        "note": "التقييم من 100: كلما ارتفع الرقم كان الموقع أفضل. هذا تقييم تقديري من هذه الأداة وليس مقياسًا رسميًا من Google.",
        "site_issue": "مشكلة على مستوى الموقع",
        "no_robots": "لا يوجد ملف robots.txt",
        "no_robots_fix": "أنشئ ملف robots.txt في جذر الموقع وأضف إليه رابط خريطة الموقع.",
        "no_robots_where": "جذر الموقع: /robots.txt",
        "p_enter": "🔗 أدخل رابط الموقع (مثال: https://example.com): ",
        "p_search": "⏳ جارٍ البحث عن خريطة الموقع لـ {b} ...",
        "p_nosm": "❌ لم يتم العثور على خريطة موقع (robots.txt أو /sitemap.xml أو /sitemap_index.xml أو /wp-sitemap.xml).",
        "p_found": "✅ خريطة الموقع: {sm} | {n} رابط",
        "p_sample": "ℹ️ سيتم فحص {m} صفحة (عينة موزعة على الموقع). استخدم --max 0 لفحص الكل.",
        "p_prog": "   تم فحص {n}/{m} صفحة",
        "p_score": "🏁 التقييم العام: {s}/100 ← {g}",
        "p_top": "🔧 أهم المشكلات التي يجب إصلاحها:",
        "p_top_row": "   [{sev}] {issue} ({n} صفحة)",
        "p_file": "📄 التقرير الكامل: {f}",
        "rtl": True,
    },
    "en": {
        "chars": "{n} chars", "words": "{n} words", "sec": "{n} s",
        "robots_dis": "Disallowed by robots.txt",
        "chain": "{n} redirects -> {u}",
        "redir": "Redirects to {u}",
        "img": "{a} of {b} images have no alt text",
        "mixed": "{n} resources loaded over http",
        "yes": "Yes", "no": "No",
        "g90": "Excellent", "g75": "Good", "g50": "Average (needs work)", "g0": "Poor (major issues)",
        "sheets": ["Summary", "Top Fixes", "All Issues", "Pages", "Broken Links", "Speed"],
        "status_n": "Status {n}",
        "page_cols": {},
        "bl_cols": ["Source page", "Broken link", "Status", "Type", "Anchor text"],
        "int": "Internal", "ext": "External",
        "speed_cols": ["URL", "Device", "Performance (/100)", "FCP (s)", "LCP (s)", "CLS", "TBT (ms)",
                       "TTFB (ms)", "Real-user rating (28 days)", "Real-user INP (ms)"],
        "broken_n": "{n} broken links",
        "js_d": "Very little text but many JavaScript elements",
        "psi_d": "Performance {s}/100",
        "crawl_found": "Pages added through internal links (not in sitemap)",
        "links_checked": "Links checked",
        "broken_total": "Broken links found",
        "orphan_skip": "Orphan-page detection skipped because not all pages were audited. Use --max 0 with a larger --crawl.",
        "p_crawl": "🕸️ Found {n} extra pages through internal links, auditing them ...",
        "p_links": "🔗 Checking {n} links for broken ones ...",
        "p_psi": "⚡ Running Google PageSpeed ({n} pages, mobile and desktop, may take several minutes) ...",
        "dev_mobile": "Mobile", "dev_desktop": "Desktop",
        "p_psi_err": "⚠️ PageSpeed failed ({d}) for {u} ({e}). If it is 429 the free quota is used up: try later or use --psi-key, or test the URL manually at pagespeed.web.dev",
        "sum_cols": ["Item", "Value"],
        "top_cols": ["Severity", "Issue", "Pages affected", "% of pages", "How to fix", "Where to fix"],
        "iss_cols": ["URL", "Severity", "Issue", "Detail", "How to fix", "Where to fix", "Code"],
        "site": "Website", "sitemap": "Sitemap", "sm_total": "URLs in sitemap",
        "checked": "Pages audited", "score": "Overall score (out of 100)", "verdict": "Verdict",
        "issues_of": "Issues ({s})",
        "note_l": "Note",
        "note": "Score is out of 100: the higher the better. It is this tool's own estimate, not an official Google metric.",
        "site_issue": "Site-wide issue",
        "no_robots": "No robots.txt file",
        "no_robots_fix": "Create a robots.txt file in the site root and add the sitemap URL to it.",
        "no_robots_where": "Site root: /robots.txt",
        "p_enter": "🔗 Enter the website URL (e.g. https://example.com): ",
        "p_search": "⏳ Looking for the sitemap of {b} ...",
        "p_nosm": "❌ No sitemap found (robots.txt, /sitemap.xml, /sitemap_index.xml, /wp-sitemap.xml).",
        "p_found": "✅ Sitemap: {sm} | {n} URLs",
        "p_sample": "ℹ️ Auditing {m} pages (an evenly spread sample). Use --max 0 to audit all.",
        "p_prog": "   {n}/{m} pages audited",
        "p_score": "🏁 Overall score: {s}/100 -> {g}",
        "p_top": "🔧 Top issues to fix:",
        "p_top_row": "   [{sev}] {issue} ({n} pages)",
        "p_file": "📄 Full report: {f}",
        "rtl": False,
    },
}


def tr(key, **kw):
    return S[LANG][key].format(**kw)


# ------------------------------------------------------------------ كتالوج المشكلات / Issue catalog
# code: (severity, (desc_ar, desc_en), (fix_ar, fix_en), (where_ar, where_en))
CATALOG = {
    "FETCH_FAIL": ("Critical",
        ("تعذّر تحميل الصفحة (انتهاء المهلة أو خطأ في الاتصال)", "Page failed to load (timeout or connection error)"),
        ("تأكد من أن الخادم يعمل بسرعة جيدة، وأن جدار الحماية أو Cloudflare لا يحجب الزوار.",
         "Make sure the server is up and fast, and that a firewall or Cloudflare is not blocking visitors."),
        ("الاستضافة / الخادم / إعدادات جدار الحماية", "Hosting / server / firewall settings")),
    "HTTP_4XX": ("Critical",
        ("الصفحة تعرض خطأ 4xx (مثل 404: الصفحة غير موجودة)", "Page returns a 4xx error (e.g. 404 Not Found)"),
        ("أزل الرابط من خريطة الموقع، أو أنشئ إعادة توجيه 301 إلى الصفحة البديلة، أو أعد نشر الصفحة.",
         "Remove the URL from the sitemap, add a 301 redirect to the replacement page, or restore the page."),
        ("ملف sitemap.xml وإعدادات إعادة التوجيه (.htaccess / Cloudflare / إضافة Redirection)",
         "sitemap.xml and redirect settings (.htaccess / Cloudflare / Redirection plugin)")),
    "HTTP_5XX": ("Critical",
        ("خطأ في الخادم (5xx)", "Server error (5xx)"),
        ("راجع سجلات الخادم (logs)؛ غالبًا تكون المشكلة في PHP أو قاعدة البيانات أو إضافة. تواصل مع الاستضافة إن لم تجد السبب.",
         "Check the server logs; it is usually a PHP, database or plugin issue. Contact your host if you cannot find the cause."),
        ("الخادم / الاستضافة / السجلات", "Server / hosting / logs")),
    "NOINDEX": ("Critical",
        ("الصفحة تحتوي على noindex (تمنع Google من فهرستها) رغم وجودها في خريطة الموقع",
         "Page has noindex (blocks Google from indexing it) yet is listed in the sitemap"),
        ("إن أردت ظهورها في Google فأزل noindex، وإن لم ترد ذلك فأزلها من خريطة الموقع.",
         "Remove noindex if the page should appear in Google; otherwise remove it from the sitemap."),
        ("وسم robots في <head> / ترويسة X-Robots-Tag / إعدادات Yoast أو RankMath",
         "robots meta tag in <head> / X-Robots-Tag header / Yoast or RankMath settings")),
    "ROBOTS_BLOCKED": ("Critical",
        ("الصفحة محظورة في ملف robots.txt", "Page is blocked by robots.txt"),
        ("عدّل قاعدة Disallow التي تحجب هذا الرابط.", "Edit the Disallow rule that blocks this URL."),
        ("ملف robots.txt في جذر الموقع", "robots.txt in the site root")),
    "NOT_HTTPS": ("High",
        ("الصفحة لا تعمل عبر HTTPS", "Page is not served over HTTPS"),
        ("فعّل شهادة SSL وأضف إعادة توجيه 301 من http إلى https لكامل الموقع.",
         "Install an SSL certificate and 301-redirect http to https site-wide."),
        ("الاستضافة (SSL) + .htaccess / Cloudflare / إعدادات الموقع", "Hosting (SSL) + .htaccess / Cloudflare / site settings")),
    "MIXED_CONTENT": ("High",
        ("صفحة HTTPS تحمّل ملفات (صور أو سكربتات) عبر http", "HTTPS page loads resources (images/scripts) over http"),
        ("استبدل كل روابط http:// بـ https:// في الصور وملفات CSS وJS.",
         "Replace all http:// links with https:// for images, CSS and JS."),
        ("كود القالب / المحتوى / إضافة مثل Really Simple SSL", "Theme code / content / a plugin such as Really Simple SSL")),
    "REDIRECT": ("Medium",
        ("الرابط في خريطة الموقع يعيد التوجيه إلى رابط آخر", "URL in the sitemap redirects elsewhere"),
        ("ضع الرابط النهائي (الذي يعطي حالة 200) في خريطة الموقع بدل الرابط القديم.",
         "Put the final URL (status 200) in the sitemap instead of the old one."),
        ("ملف sitemap.xml", "sitemap.xml")),
    "REDIRECT_CHAIN": ("High",
        ("سلسلة إعادة توجيه (أكثر من إعادة توجيه واحدة)", "Redirect chain (more than one redirect)"),
        ("اجعلها إعادة توجيه واحدة مباشرة من الرابط القديم إلى الرابط النهائي.",
         "Use a single direct redirect from the old URL to the final URL."),
        ("إعدادات إعادة التوجيه (.htaccess / Cloudflare / إضافة)", "Redirect settings (.htaccess / Cloudflare / plugin)")),
    "SLOW": ("Medium",
        ("الصفحة بطيئة (زمن الاستجابة أكثر من 1.5 ثانية)", "Slow page (response time over 1.5 seconds)"),
        ("فعّل التخزين المؤقت (Cache) وضغط الصور وشبكة CDN، وحسّن الاستضافة. قِس الأداء عبر PageSpeed Insights.",
         "Enable caching, compress images, use a CDN and upgrade hosting. Measure with PageSpeed Insights."),
        ("الاستضافة + إضافة التخزين المؤقت (WP Rocket / LiteSpeed) + الصور",
         "Hosting + cache plugin (WP Rocket / LiteSpeed) + images")),
    "BIG_PAGE": ("Low",
        ("حجم الصفحة كبير (أكثر من 1.5 ميغابايت من HTML)", "Large page (over 1.5 MB of HTML)"),
        ("قلّل الكود الزائد وفعّل ضغط gzip/brotli.", "Reduce unnecessary markup and enable gzip/brotli compression."),
        ("القالب / إعدادات الخادم", "Theme / server settings")),
    "TITLE_MISSING": ("High",
        ("لا يوجد عنوان للصفحة (Title)", "Missing title tag"),
        ("أضف وسم <title> فريدًا يتضمن الكلمة المفتاحية (30-60 حرفًا).",
         "Add a unique <title> containing the target keyword (30-60 characters)."),
        ("وسم <head> للصفحة / حقل عنوان SEO في Yoast أو RankMath", "Page <head> / SEO Title field in Yoast or RankMath")),
    "TITLE_SHORT": ("Medium",
        ("عنوان الصفحة قصير (أقل من 30 حرفًا)", "Title too short (under 30 characters)"),
        ("أطِل العنوان وأضف الكلمة المفتاحية وما يميّز الصفحة (30-60 حرفًا).",
         "Lengthen the title and include the keyword and a clear benefit (30-60 characters)."),
        ("حقل عنوان SEO في Yoast / RankMath أو وسم <title>", "SEO Title field in Yoast / RankMath or the <title> tag")),
    "TITLE_LONG": ("Medium",
        ("عنوان الصفحة طويل (أكثر من 60 حرفًا) وسيُقتطع في نتائج Google",
         "Title too long (over 60 characters) and will be truncated in Google"),
        ("اختصر العنوان وضع الكلمة المهمة في بدايته.", "Shorten the title and put the important keyword first."),
        ("حقل عنوان SEO في Yoast / RankMath أو وسم <title>", "SEO Title field in Yoast / RankMath or the <title> tag")),
    "TITLE_DUP": ("Medium",
        ("العنوان نفسه مستخدم في صفحات أخرى", "Same title used on other pages"),
        ("يجب أن يكون لكل صفحة عنوان خاص بها.", "Give every page its own unique title."),
        ("حقل عنوان SEO في كل صفحة", "SEO Title field on each page")),
    "DESC_MISSING": ("Medium",
        ("لا يوجد وصف ميتا (Meta Description)", "Missing meta description"),
        ("أضف وصفًا جذابًا (70-160 حرفًا) يتضمن الكلمة المفتاحية.",
         "Add a compelling description (70-160 characters) with the target keyword."),
        ("حقل الوصف في Yoast / RankMath أو وسم <meta name='description'> في <head>",
         "Description field in Yoast / RankMath or the <meta name='description'> tag in <head>")),
    "DESC_SHORT": ("Low",
        ("وصف الميتا قصير (أقل من 70 حرفًا)", "Meta description too short (under 70 characters)"),
        ("أطِل الوصف قليلًا (70-160 حرفًا).", "Lengthen it a little (70-160 characters)."),
        ("حقل وصف الميتا", "Meta description field")),
    "DESC_LONG": ("Low",
        ("وصف الميتا طويل (أكثر من 160 حرفًا)", "Meta description too long (over 160 characters)"),
        ("اختصر الوصف حتى لا يُقتطع.", "Shorten it so it is not truncated."),
        ("حقل وصف الميتا", "Meta description field")),
    "DESC_DUP": ("Medium",
        ("وصف الميتا نفسه مستخدم في صفحات أخرى", "Same meta description used on other pages"),
        ("اكتب وصفًا خاصًا بكل صفحة.", "Write a unique description for each page."),
        ("حقل وصف الميتا في كل صفحة", "Meta description field on each page")),
    "H1_MISSING": ("High",
        ("لا يوجد عنوان رئيسي H1", "Missing H1 heading"),
        ("أضف H1 واحدًا في الصفحة يتضمن الكلمة المفتاحية الرئيسية.", "Add a single H1 containing the main keyword."),
        ("محتوى الصفحة / القالب (عنوان الصفحة يجب أن يكون H1)", "Page content / theme (the page title should be an H1)")),
    "H1_MULTI": ("Medium",
        ("أكثر من H1 في الصفحة نفسها", "More than one H1 on the page"),
        ("اترك H1 واحدًا فقط وحوّل البقية إلى H2 أو H3.", "Keep one H1 and change the rest to H2/H3."),
        ("محتوى الصفحة / القالب / أداة بناء الصفحات", "Page content / theme / page builder")),
    "H1_DUP": ("Low",
        ("العنوان الرئيسي H1 نفسه مستخدم في صفحات أخرى", "Same H1 used on other pages"),
        ("اجعل H1 فريدًا لكل صفحة.", "Make each page's H1 unique."),
        ("محتوى الصفحة", "Page content")),
    "CANON_MISSING": ("Medium",
        ("لا يوجد رابط Canonical", "Missing canonical tag"),
        ("أضف <link rel='canonical'> يشير إلى الرابط الأصلي للصفحة.",
         "Add <link rel='canonical'> pointing to the page's preferred URL."),
        ("<head> / Yoast / RankMath (متقدم ← Canonical URL)", "<head> / Yoast / RankMath (Advanced > Canonical URL)")),
    "CANON_OTHER": ("Medium",
        ("رابط Canonical يشير إلى صفحة أخرى", "Canonical points to a different URL"),
        ("تأكد من أن ذلك مقصود، وإلا فاجعل الـ canonical يشير إلى الصفحة نفسها.",
         "Make sure this is intentional; otherwise point the canonical to the page itself."),
        ("<head> / Yoast / RankMath (متقدم ← Canonical URL)", "<head> / Yoast / RankMath (Advanced > Canonical URL)")),
    "IMG_ALT": ("Medium",
        ("صور بلا نص بديل (alt)", "Images missing alt text"),
        ("أضف نصًا بديلًا وصفيًا لكل صورة (يفيد SEO وإمكانية الوصول).",
         "Add descriptive alt text to every image (helps SEO and accessibility)."),
        ("مكتبة الوسائط أو الوسم <img alt=''>", "Media Library or the <img alt=''> attribute")),
    "THIN": ("Medium",
        ("محتوى قليل (أقل من 300 كلمة)", "Thin content (under 300 words)"),
        ("أضف محتوى مفيدًا وأصليًا يجيب عن نية الباحث، وتجنب الصفحات شبه الفارغة.",
         "Add useful, original content that answers search intent; avoid near-empty pages."),
        ("محتوى الصفحة", "Page content")),
    "NO_VIEWPORT": ("High",
        ("لا يوجد وسم viewport (الموقع غير مهيأ للجوال)", "Missing viewport tag (not mobile-friendly)"),
        ("أضف <meta name='viewport' content='width=device-width, initial-scale=1'>.",
         "Add <meta name='viewport' content='width=device-width, initial-scale=1'>."),
        ("<head> / القالب (header.php)", "<head> / theme (header.php)")),
    "NO_LANG": ("Low",
        ("لا يوجد وسم lang في <html>", "Missing lang attribute on <html>"),
        ("أضف lang مثل <html lang='ar'> أو 'fr' أو 'en'.", "Add lang, e.g. <html lang='en'>."),
        ("القالب (header.php) / إعدادات اللغة", "Theme (header.php) / language settings")),
    "NO_OG": ("Low",
        ("لا توجد وسوم Open Graph (معاينة المشاركة في فيسبوك وواتساب)",
         "Missing Open Graph tags (share previews on Facebook/WhatsApp)"),
        ("أضف og:title وog:description وog:image.", "Add og:title, og:description and og:image."),
        ("Yoast / RankMath (التواصل الاجتماعي) أو <head>", "Yoast / RankMath (Social) or <head>")),
    "NO_SCHEMA": ("Low",
        ("لا توجد بيانات منظمة (Schema)", "No structured data (Schema)"),
        ("أضف Schema بصيغة JSON-LD مناسبة (Article / Product / LocalBusiness / FAQ ...).",
         "Add suitable JSON-LD Schema (Article / Product / LocalBusiness / FAQ ...)."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    "URL_LONG": ("Low",
        ("الرابط طويل (أكثر من 115 حرفًا)", "URL too long (over 115 characters)"),
        ("اختصر الرابط واكتفِ بكلمات مفتاحية قليلة.", "Shorten the URL and keep only a few keywords."),
        ("الرابط الدائم (Slug) للصفحة", "Page slug / permalink")),
    "URL_STYLE": ("Low",
        ("الرابط يحتوي على أحرف كبيرة أو شرطة سفلية أو معاملات (parameters)",
         "URL contains uppercase letters, underscores or parameters"),
        ("استخدم أحرفًا صغيرة وشرطة (-) بدل الشرطة السفلية، وتجنب المعاملات.",
         "Use lowercase and hyphens instead of underscores; avoid parameters."),
        ("الرابط الدائم / إعدادات الروابط", "Slug / permalink settings")),
    "NO_INTERNAL_LINKS": ("Medium",
        ("لا توجد روابط داخلية في الصفحة", "No internal links on the page"),
        ("أضف روابط داخلية إلى صفحات مهمة أخرى في الموقع.", "Add internal links to other important pages."),
        ("محتوى الصفحة / القائمة", "Page content / menu")),
    "BROKEN_INTERNAL": ("High",
        ("روابط داخلية مكسورة (تؤدي إلى صفحات غير موجودة أو بها خطأ)", "Broken internal links (point to missing or erroring pages)"),
        ("عدّل الرابط ليشير إلى صفحة صحيحة، أو أنشئ إعادة توجيه 301، أو أزل الرابط. التفاصيل في ورقة «الروابط المكسورة».",
         "Point the link to a valid page, add a 301 redirect, or remove the link. See the 'Broken Links' sheet for details."),
        ("محتوى الصفحة / القائمة / القالب", "Page content / menu / theme")),
    "BROKEN_EXTERNAL": ("Medium",
        ("روابط خارجية مكسورة", "Broken external links"),
        ("استبدل الرابط بمصدر يعمل أو أزله. التفاصيل في ورقة «الروابط المكسورة».",
         "Replace the link with a working source or remove it. See the 'Broken Links' sheet for details."),
        ("محتوى الصفحة", "Page content")),
    "NOT_IN_SITEMAP": ("Low",
        ("صفحة موجودة ويمكن الوصول إليها لكنها غير مدرجة في خريطة الموقع", "Live page that is not listed in the sitemap"),
        ("أضفها إلى خريطة الموقع إن كانت صفحة مهمة، أو أضف لها noindex إن لم تكن كذلك.",
         "Add it to the sitemap if it is important, or noindex it if it is not."),
        ("sitemap.xml / إعدادات الخرائط في Yoast أو RankMath", "sitemap.xml / Sitemap settings in Yoast or RankMath")),
    "ORPHAN": ("Medium",
        ("صفحة يتيمة: لا توجد أي صفحة مفحوصة تربط إليها داخليًا", "Orphan page: no audited page links to it"),
        ("أضف روابط داخلية إليها من صفحات ذات صلة أو من القائمة الرئيسية.",
         "Add internal links to it from related pages or the main menu."),
        ("محتوى الصفحات / القائمة", "Page content / menu")),
    "HEADING_SKIP": ("Low",
        ("ترتيب العناوين غير منطقي (قفز مثلًا من H1 إلى H3)", "Illogical heading order (e.g. jumping from H1 to H3)"),
        ("رتّب العناوين H1 ثم H2 ثم H3 دون قفز بين المستويات.", "Order headings H1, H2, H3 without skipping levels."),
        ("محتوى الصفحة / القالب", "Page content / theme")),
    "JS_SUSPECT": ("Info",
        ("الصفحة تبدو معتمدة على JavaScript لعرض المحتوى، وقد تكون بعض نتائج الفحص غير دقيقة",
         "Page seems to rely on JavaScript to render content, so some results may be inaccurate"),
        ("تحقق من الصفحة يدويًا في المتصفح (قارن View Source مع Inspect) قبل الاعتماد على نتائجها.",
         "Check the page manually in the browser (compare View Source with Inspect) before trusting its results."),
        ("-", "-")),
    "PSI_POOR": ("Medium",
        ("أداء ضعيف على الجوال حسب Google PageSpeed (أقل من 50)", "Poor mobile performance per Google PageSpeed (below 50)"),
        ("اضغط الصور، فعّل التخزين المؤقت وCDN، قلّل ملفات JS وCSS غير الضرورية، وراجع تقرير PageSpeed للتفاصيل.",
         "Compress images, enable caching and a CDN, cut unneeded JS/CSS, and review the PageSpeed report for details."),
        ("الصور / القالب / الإضافات / الاستضافة", "Images / theme / plugins / hosting")),
    "PSI_AVG": ("Low",
        ("أداء متوسط على الجوال حسب Google PageSpeed (50-89)", "Average mobile performance per Google PageSpeed (50-89)"),
        ("حسّن الصور والتخزين المؤقت وقلّل الملفات الثقيلة.", "Improve images and caching, and reduce heavy files."),
        ("الصور / القالب / الإضافات / الاستضافة", "Images / theme / plugins / hosting")),
}


def cat(code):
    sev, d, f, w = CATALOG[code]
    i = 0 if LANG == "ar" else 1
    return sev, d[i], f[i], w[i]


# ------------------------------------------------------------------ الشبكة / Network
def make_session():
    s = requests.Session()
    s.headers.update(HEADERS)
    return s


def fetch(session, url, timeout, retries=2):
    last = None
    for i in range(retries + 1):
        try:
            return session.get(url, timeout=timeout, allow_redirects=True)
        except Exception as e:  # noqa
            last = e
            time.sleep(1 + i)
    raise last


# ------------------------------------------------------------------ Sitemap
def read_locs(content, url):
    if url.endswith(".gz") or content[:2] == b"\x1f\x8b":
        content = gzip.decompress(content)
    text = content.decode("utf-8", errors="ignore")
    is_index = "<sitemapindex" in text
    locs = [re.sub(r"<!\[CDATA\[|\]\]>", "", m).strip()
            for m in re.findall(r"<loc>\s*(.*?)\s*</loc>", text, re.S)]
    return is_index, locs


def expand_sitemap(session, url, timeout, depth=0, seen=None):
    seen = seen if seen is not None else set()
    if url in seen or depth > 3:
        return []
    seen.add(url)
    try:
        r = fetch(session, url, timeout, retries=1)
        if r.status_code != 200:
            return []
        is_index, locs = read_locs(r.content, url)
    except Exception:
        return []
    if not is_index:
        return locs
    urls = []
    for child in locs[:50]:
        urls += expand_sitemap(session, child, timeout, depth + 1, seen)
    return urls


def discover(session, base, timeout):
    robots_txt = ""
    candidates = []
    try:
        r = session.get(base + "/robots.txt", timeout=timeout)
        if r.status_code == 200 and "<html" not in r.text[:500].lower():
            robots_txt = r.text
            for line in robots_txt.splitlines():
                if line.lower().startswith("sitemap:"):
                    candidates.append(line.split(":", 1)[1].strip())
    except Exception:
        pass
    for p in ("/sitemap.xml", "/sitemap_index.xml", "/wp-sitemap.xml"):
        candidates.append(base + p)
    for c in dict.fromkeys(candidates):
        urls = expand_sitemap(session, c, timeout)
        if urls:
            return c, list(dict.fromkeys(urls)), robots_txt
    return None, [], robots_txt


# ------------------------------------------------------------------ فحص صفحة / Page analysis
def norm(u):
    u = u.split("#")[0].strip().lower()
    return u[:-1] if u.endswith("/") else u


ASSET_RE = re.compile(r"\.(jpe?g|png|gif|webp|svg|ico|pdf|zip|rar|css|js|mp4|mp3|docx?|xlsx?|pptx?)(\?.*)?$", re.I)


def looks_like_site(s):
    """True for 'https://x', 'http://x' or a bare domain like 'example.com' (not a file path)."""
    s = (s or "").strip()
    if s.lower().startswith(("http://", "https://")):
        return True
    return bool(re.match(r"^([a-z0-9-]+\.)+[a-z]{2,}(:\d+)?(/.*)?$", s, re.I)) and not s.lower().endswith(".json")


def same_host(a, b):
    ha = urlparse(a).netloc.lower().replace("www.", "")
    hb = urlparse(b).netloc.lower().replace("www.", "")
    return ha == hb


def check_link(session, url, timeout=10):
    """Returns the HTTP status of a link (or 'ERROR')."""
    try:
        r = session.head(url, timeout=timeout, allow_redirects=True)
        if r.status_code >= 400:
            r = session.get(url, timeout=timeout, allow_redirects=True, stream=True)
            r.close()
        return r.status_code
    except Exception:
        return "ERROR"


def is_broken(st):
    # 401/403/429 are usually bot protection, not a truly broken link
    return st == "ERROR" or st in (404, 410) or (isinstance(st, int) and 500 <= st < 600)


PSI_CAT = {
    "ar": {"FAST": "سريع", "AVERAGE": "متوسط", "SLOW": "بطيء"},
    "en": {"FAST": "Fast", "AVERAGE": "Average", "SLOW": "Slow"},
}


def parse_psi(data):
    """Extract [score, FCP s, LCP s, CLS, TBT ms, TTFB ms, real-user rating, real-user INP ms] from a PSI response."""
    lh = data["lighthouseResult"]
    a = lh["audits"]

    def num(key, scale=1, nd=None):
        try:
            v = a[key]["numericValue"] / scale
            return round(v, nd) if nd is not None else round(v)
        except Exception:
            return "n/a"

    score = round(lh["categories"]["performance"]["score"] * 100)
    field = data.get("loadingExperience") or {}
    cat = PSI_CAT[LANG].get(field.get("overall_category"), "n/a")
    inp = (field.get("metrics", {}).get("INTERACTION_TO_NEXT_PAINT") or {}).get("percentile", "n/a")
    return [score,
            num("first-contentful-paint", 1000, 2),
            num("largest-contentful-paint", 1000, 2),
            num("cumulative-layout-shift", 1, 3),
            num("total-blocking-time"),
            num("server-response-time"),
            cat, inp]


def pagespeed(session, url, key="", strategy="mobile", timeout=90):
    """Google PageSpeed Insights API (mobile or desktop). Returns (values, error)."""
    params = {"url": url, "strategy": strategy, "category": "performance"}
    if key:
        params["key"] = key
    try:
        r = session.get("https://www.googleapis.com/pagespeedonline/v5/runPagespeed",
                        params=params, timeout=timeout)
        if r.status_code != 200:
            return None, f"HTTP {r.status_code}"
        return parse_psi(r.json()), ""
    except Exception as e:  # noqa
        return None, str(e)[:80]


def analyze(url, session, timeout, delay, rp):
    issues = []
    row = {"URL": url}
    try:
        r = fetch(session, url, timeout)
    except Exception as e:  # noqa
        row["Status"] = "ERROR"
        issues.append(("FETCH_FAIL", str(e)[:150]))
        return row, issues
    time.sleep(delay)

    final = r.url
    status = r.status_code
    row.update({
        "Status": status,
        "Final URL": final,
        "Redirects": len(r.history),
        "Response (s)": round(r.elapsed.total_seconds(), 2),
        "Size (KB)": round(len(r.content) / 1024),
    })

    if rp is not None and not rp.can_fetch("*", url):
        issues.append(("ROBOTS_BLOCKED", tr("robots_dis")))
    if len(r.history) >= 2:
        issues.append(("REDIRECT_CHAIN", tr("chain", n=len(r.history), u=final)))
    elif len(r.history) == 1:
        issues.append(("REDIRECT", tr("redir", u=final)))
    if 400 <= status < 500:
        issues.append(("HTTP_4XX", tr("status_n", n=status)))
        return row, issues
    if status >= 500:
        issues.append(("HTTP_5XX", tr("status_n", n=status)))
        return row, issues

    if not final.startswith("https://"):
        issues.append(("NOT_HTTPS", final))
    if row["Response (s)"] > 1.5:
        issues.append(("SLOW", tr("sec", n=row["Response (s)"])))
    if len(r.content) > 1_500_000:
        issues.append(("BIG_PAGE", f"{row['Size (KB)']} KB"))

    ctype = r.headers.get("Content-Type", "")
    if "html" not in ctype.lower():
        return row, issues

    soup = BeautifulSoup(r.content, "html.parser")

    # URL
    path = urlparse(final)
    if len(final) > 115:
        issues.append(("URL_LONG", tr("chars", n=len(final))))
    if re.search(r"[A-Z]", path.path) or "_" in path.path or path.query:
        issues.append(("URL_STYLE", final))

    # Title
    t = soup.find("title")
    title = t.get_text(strip=True) if t else ""
    row["Title"] = title
    row["Title len"] = len(title)
    if not title:
        issues.append(("TITLE_MISSING", ""))
    elif len(title) < 30:
        issues.append(("TITLE_SHORT", f"{tr('chars', n=len(title))}: {title}"))
    elif len(title) > 60:
        issues.append(("TITLE_LONG", f"{tr('chars', n=len(title))}: {title}"))

    # Meta description
    m = soup.find("meta", attrs={"name": re.compile(r"^description$", re.I)})
    desc = (m.get("content") or "").strip() if m else ""
    row["Meta description"] = desc
    row["Desc len"] = len(desc)
    if not desc:
        issues.append(("DESC_MISSING", ""))
    elif len(desc) < 70:
        issues.append(("DESC_SHORT", tr("chars", n=len(desc))))
    elif len(desc) > 160:
        issues.append(("DESC_LONG", tr("chars", n=len(desc))))

    # H1
    h1s = [h.get_text(" ", strip=True) for h in soup.find_all("h1")]
    h1s = [h for h in h1s if h]
    row["H1"] = h1s[0] if h1s else ""
    row["H1 count"] = len(h1s)
    if not h1s:
        issues.append(("H1_MISSING", ""))
    elif len(h1s) > 1:
        issues.append(("H1_MULTI", f"{len(h1s)} H1"))

    # Canonical
    can = soup.find("link", rel=lambda v: v and "canonical" in
                    ([x.lower() for x in v] if isinstance(v, list) else [v.lower()]))
    canonical = (can.get("href") or "").strip() if can else ""
    row["Canonical"] = canonical
    if not canonical:
        issues.append(("CANON_MISSING", ""))
    elif norm(urljoin(final, canonical)) != norm(final):
        issues.append(("CANON_OTHER", f"canonical = {canonical}"))

    # Noindex
    rb = soup.find("meta", attrs={"name": re.compile(r"^robots$", re.I)})
    robots_meta = (rb.get("content") or "").lower() if rb else ""
    xrt = r.headers.get("X-Robots-Tag", "").lower()
    noindex = "noindex" in robots_meta or "noindex" in xrt
    row["Noindex"] = tr("yes") if noindex else tr("no")
    if noindex:
        issues.append(("NOINDEX", robots_meta or xrt))

    # Mobile / lang / social / schema
    if not soup.find("meta", attrs={"name": re.compile(r"^viewport$", re.I)}):
        issues.append(("NO_VIEWPORT", ""))
    if not (soup.html and soup.html.get("lang")):
        issues.append(("NO_LANG", ""))
    if not soup.find("meta", property="og:title"):
        issues.append(("NO_OG", ""))
    if not (soup.find("script", type="application/ld+json") or soup.find(attrs={"itemtype": True})):
        issues.append(("NO_SCHEMA", ""))

    # Images
    imgs = soup.find_all("img")
    no_alt = [i for i in imgs if i.get("alt") is None]
    row["Images"] = len(imgs)
    row["Images no alt"] = len(no_alt)
    if no_alt:
        issues.append(("IMG_ALT", tr("img", a=len(no_alt), b=len(imgs))))

    # Mixed content
    if final.startswith("https://"):
        bad = 0
        for tag in soup.find_all(["img", "script", "iframe", "source", "link"]):
            u = tag.get("src") or (tag.get("href") if tag.name == "link" else None)
            if u and u.startswith("http://"):
                bad += 1
        if bad:
            issues.append(("MIXED_CONTENT", tr("mixed", n=bad)))

    # Links (internal + external), kept for the crawl / broken-link phases
    links, seen_links = [], set()
    for a in soup.find_all("a", href=True):
        href = a["href"].strip()
        if not href or href.startswith(("#", "mailto:", "tel:", "javascript:", "sms:", "whatsapp:")):
            continue
        absu = urldefrag(urljoin(final, href))[0]
        if not absu.startswith(("http://", "https://")) or absu in seen_links:
            continue
        seen_links.add(absu)
        links.append((absu, same_host(absu, final), a.get_text(" ", strip=True)[:60]))
    row["_links"] = links
    internal = sum(1 for _, i, _t in links if i)
    row["Internal links"] = internal
    row["External links"] = len(links) - internal
    if internal == 0:
        issues.append(("NO_INTERNAL_LINKS", ""))

    # Heading order (e.g. H1 -> H3 without H2)
    levels = [int(h.name[1]) for h in soup.find_all(re.compile(r"^h[1-6]$"))]
    if any(b - a > 1 for a, b in zip(levels, levels[1:])):
        issues.append(("HEADING_SKIP", "H" + " > H".join(str(x) for x in levels[:12])))

    # Word count
    n_scripts = len(soup.find_all("script"))
    js_root = soup.find(id=re.compile(r"^(root|app|__next|__nuxt)$"))
    for tag in soup(["script", "style", "noscript"]):
        tag.decompose()
    words = len(soup.get_text(" ", strip=True).split())
    row["Words"] = words
    if words < 300:
        issues.append(("THIN", tr("words", n=words)))
    if words < 50 and (js_root is not None or n_scripts >= 8):
        issues.append(("JS_SUSPECT", tr("js_d")))

    return row, issues


# ------------------------------------------------------------------ التقييم والتقرير / Scoring & report
def grade(score):
    if score >= 90:
        return tr("g90")
    if score >= 75:
        return tr("g75")
    if score >= 50:
        return tr("g50")
    return tr("g0")


def style_sheet(ws, widths, sev_col=None):
    from openpyxl.styles import Alignment, Font, PatternFill
    head = PatternFill("solid", fgColor="1F4E78")
    for c in ws[1]:
        c.font = Font(bold=True, color="FFFFFF")
        c.fill = head
        c.alignment = Alignment(vertical="center", wrap_text=True)
    for col, w in widths.items():
        ws.column_dimensions[col].width = w
    ws.freeze_panes = "A2"
    ws.auto_filter.ref = ws.dimensions
    ws.sheet_view.rightToLeft = S[LANG]["rtl"]
    raw = {"Critical": "F8CBAD", "High": "FFE699", "Medium": "FFF2CC", "Low": "E2EFDA", "Info": "DDEBF7"}
    colors = {SEVLBL[LANG][k]: v for k, v in raw.items()}
    for row in ws.iter_rows(min_row=2):
        for c in row:
            c.alignment = Alignment(vertical="top", wrap_text=True)
        if sev_col is not None:
            v = row[sev_col].value
            if v in colors:
                row[sev_col].fill = PatternFill("solid", fgColor=colors[v])


def main():
    global LANG
    ap = argparse.ArgumentParser(description="SEO Audit (Technical + On-Page)")
    ap.add_argument("url", nargs="?", help="website URL")
    ap.add_argument("--lang", choices=["ar", "en"], default=REPORT_LANG, help="report language: ar (Arabic) or en (English)")
    ap.add_argument("--max", type=int, default=100, help="number of sitemap pages to audit (0 = all)")
    ap.add_argument("--crawl", type=int, default=50,
                    help="extra pages to discover through internal links (0 = off)")
    ap.add_argument("--links-max", type=int, default=300, help="max links to check for broken ones (0 = off)")
    ap.add_argument("--psi", type=int, default=1, help="pages to test with Google PageSpeed (default 1 = homepage, 0 = off)")
    ap.add_argument("--psi-key", default=PSI_API_KEY, help="optional Google PageSpeed API key")
    ap.add_argument("--workers", type=int, default=5)
    ap.add_argument("--delay", type=float, default=0.3, help="delay between requests (seconds)")
    ap.add_argument("--timeout", type=int, default=15)
    ap.add_argument("--out", default="", help="Excel file name (default: seo_audit_<domain>.xlsx)")
    args, _ = ap.parse_known_args()  # works in Colab/Jupyter too
    LANG = args.lang if args.lang in ("ar", "en") else "ar"

    # In Colab/Jupyter, sys.argv contains "-f /path/kernel-xxx.json", which argparse may mistake
    # for the URL. Only accept an argument that really looks like a website.
    arg_url = args.url if looks_like_site(args.url) else ""
    url = (arg_url or SITE_URL or input(tr("p_enter"))).strip()
    if not url.startswith("http"):
        url = "https://" + url
    p = urlparse(url)
    if not p.netloc:
        print("❌ Invalid website URL / رابط الموقع غير صالح:", repr(url))
        sys.exit(1)
    base = f"{p.scheme}://{p.netloc}"
    if not args.out:  # one file per website, so reports never overwrite each other
        args.out = "seo_audit_" + re.sub(r"[^a-z0-9]+", "_", p.netloc.lower()).strip("_") + ".xlsx"

    session = make_session()
    print(tr("p_search", b=base))
    sm_url, urls, robots_txt = discover(session, base, args.timeout)
    if not urls:
        print(tr("p_nosm"))
        sys.exit(1)
    total = len(urls)
    print(tr("p_found", sm=sm_url, n=total))
    sitemap_all = {norm(u) for u in urls}

    sampled = False
    if args.max and total > args.max:
        step = total / args.max
        urls = [urls[int(i * step)] for i in range(args.max)]  # evenly spread sample
        sampled = True
        print(tr("p_sample", m=args.max))
    sitemap_pages = set(urls)

    rp = None
    if robots_txt:
        rp = RobotFileParser()
        rp.parse(robots_txt.splitlines())

    def run(url_list):
        out = []
        with ThreadPoolExecutor(max_workers=args.workers) as ex:
            futs = {ex.submit(analyze, u, session, args.timeout, args.delay, rp): u for u in url_list}
            for n, f in enumerate(as_completed(futs), 1):
                try:
                    out.append(f.result())
                except Exception as e:  # noqa
                    out.append(({"URL": futs[f], "Status": "ERROR"}, [("FETCH_FAIL", str(e)[:150])]))
                if n % 10 == 0 or n == len(futs):
                    print(tr("p_prog", n=n, m=len(futs)))
        return out

    def known(res):
        s = set()
        for row, _ in res:
            s.add(norm(row["URL"]))
            if row.get("Final URL"):
                s.add(norm(row["Final URL"]))
        return s

    # ---- Phase 1: sitemap pages
    results = run(urls)
    analyzed = known(results)

    # ---- Phase 2: pages discovered through internal links (not in the sitemap)
    crawl_added = 0
    crawl_truncated = False
    if args.crawl:
        cand = {}
        for row, _ in results:
            for lnk, internal, _t in row.get("_links", []):
                n = norm(lnk)
                if (internal and n not in analyzed and n not in sitemap_all and n not in cand
                        and not ASSET_RE.search(lnk) and (rp is None or rp.can_fetch("*", lnk))):
                    cand[n] = lnk
        cand_urls = list(cand.values())
        crawl_truncated = len(cand_urls) > args.crawl
        cand_urls = cand_urls[:args.crawl]
        if cand_urls:
            print(tr("p_crawl", n=len(cand_urls)))
            res2 = run(cand_urls)
            for row, issues in res2:
                if row.get("Status") == 200 and "Title" in row and not row.get("Redirects"):
                    issues.append(("NOT_IN_SITEMAP", ""))
            results += res2
            analyzed |= known(res2)
            crawl_added = len(res2)

    # ---- Phase 3: broken links
    status_map = {}
    for row, _ in results:
        for key in (row["URL"], row.get("Final URL")):
            if key:
                status_map[norm(key)] = row.get("Status")
    targets = {}
    for row, _ in results:
        for lnk, internal, _t in row.get("_links", []):
            n = norm(lnk)
            if n not in status_map and n not in targets:
                targets[n] = (lnk, internal)
    ordered = sorted(targets.values(), key=lambda x: not x[1])  # internal links first
    ordered = ordered[:args.links_max] if args.links_max else []
    checked = {}
    if ordered:
        print(tr("p_links", n=len(ordered)))
        with ThreadPoolExecutor(max_workers=8) as ex:
            futs = {ex.submit(check_link, session, lnk): norm(lnk) for lnk, _i in ordered}
            for f in as_completed(futs):
                checked[futs[f]] = f.result()
    status_map.update(checked)

    broken_rows = []
    for row, issues in results:
        n_int = n_ext = 0
        for lnk, internal, text in row.get("_links", []):
            st = status_map.get(norm(lnk))
            if st is not None and is_broken(st):
                broken_rows.append([row["URL"], lnk, st, tr("int") if internal else tr("ext"), text])
                if internal:
                    n_int += 1
                else:
                    n_ext += 1
        if n_int:
            issues.append(("BROKEN_INTERNAL", tr("broken_n", n=n_int)))
        if n_ext:
            issues.append(("BROKEN_EXTERNAL", tr("broken_n", n=n_ext)))

    # ---- Phase 4: orphan pages (only when every sitemap page was audited and the crawl was complete)
    orphan_note = ""
    if sampled or crawl_truncated or not args.crawl:
        orphan_note = tr("orphan_skip")
    else:
        inbound = set()
        for row, _ in results:
            me = {norm(row["URL"]), norm(row.get("Final URL") or row["URL"])}
            for lnk, internal, _t in row.get("_links", []):
                n = norm(lnk)
                if internal and n not in me:
                    inbound.add(n)
        home = {norm(base)}
        for row, issues in results:
            if row["URL"] in sitemap_pages and row.get("Status") == 200 and "Title" in row:
                keys = {norm(row["URL"]), norm(row.get("Final URL") or row["URL"])}
                if not (keys & inbound) and not (keys & home):
                    issues.append(("ORPHAN", ""))

    # ---- Phase 5: Google PageSpeed (optional)
    speed_rows = []
    if args.psi:
        print(tr("p_psi", n=args.psi))
        idx = {r[0]["URL"]: r for r in results}
        psi_targets = [r[0]["URL"] for r in results if r[0].get("Status") == 200 and "Title" in r[0]]
        psi_targets.sort(key=lambda u: norm(u) != norm(base))  # homepage first
        for u in psi_targets[:args.psi]:
            for dev in ("mobile", "desktop"):
                label = tr("dev_" + dev)
                res, err = pagespeed(session, u, args.psi_key, dev)
                if res is None:
                    speed_rows.append([u, label] + ["n/a"] * 8)
                    print(tr("p_psi_err", d=label, u=u, e=err))
                else:
                    speed_rows.append([u, label, *res])
                    if dev == "mobile":  # score deductions use the mobile result
                        if res[0] < 50:
                            idx[u][1].append(("PSI_POOR", tr("psi_d", s=res[0])))
                        elif res[0] < 90:
                            idx[u][1].append(("PSI_AVG", tr("psi_d", s=res[0])))
                time.sleep(1)

    # ---- Duplicate Title / Description / H1
    def dups(key):
        c = Counter(r[0].get(key) for r in results if r[0].get(key))
        return {k for k, v in c.items() if v > 1}

    d_title, d_desc, d_h1 = dups("Title"), dups("Meta description"), dups("H1")
    for row, issues in results:
        if row.get("Title") in d_title:
            issues.append(("TITLE_DUP", row["Title"]))
        if row.get("Meta description") in d_desc:
            issues.append(("DESC_DUP", row["Meta description"][:100]))
        if row.get("H1") in d_h1:
            issues.append(("H1_DUP", row["H1"]))

    # ---- Scoring
    issue_rows, page_rows = [], []
    for row, issues in results:
        row.pop("_links", None)
        codes = {}
        for code, detail in issues:
            codes.setdefault(code, detail)
        penalty = sum(WEIGHTS[CATALOG[c][0]] for c in codes)
        row["Score"] = max(0, 100 - penalty)
        row["Issues"] = len(codes)
        page_rows.append(row)
        for code, detail in codes.items():
            sev, desc, fix, where = cat(code)
            issue_rows.append((SEV_ORDER[sev], code,
                               [row["URL"], SEVLBL[LANG][sev], desc, detail, fix, where, code]))
    issue_rows.sort(key=lambda x: (x[0], x[1]))
    issue_list = [x[2] for x in issue_rows]

    site_penalty = 0
    site_issue_rows = []
    if not robots_txt:
        site_penalty += 5
        site_issue_rows.append((tr("site_issue"),
                                f"{tr('no_robots')} -> {tr('no_robots_fix')} ({tr('no_robots_where')})"))
    avg = sum(r["Score"] for r in page_rows) / len(page_rows)
    site_score = max(0, round(avg - site_penalty, 1))

    # Aggregated top issues
    agg = Counter(x[1] for x in issue_rows)
    top_rows = []
    for code, cnt in agg.items():
        sev, desc, fix, where = cat(code)
        top_rows.append((SEV_ORDER[sev], -cnt,
                         [SEVLBL[LANG][sev], desc, cnt, round(cnt * 100 / len(page_rows), 1), fix, where]))
    top_rows.sort(key=lambda x: (x[0], x[1]))
    top_list = [x[2] for x in top_rows]
    page_rows.sort(key=lambda x: x["Score"])

    sev_count = Counter(CATALOG[x[1]][0] for x in issue_rows)
    summary = [
        (tr("site"), base),
        (tr("sitemap"), sm_url),
        (tr("sm_total"), total),
        (tr("checked"), len(page_rows)),
        (tr("crawl_found"), crawl_added),
        (tr("links_checked"), len(checked)),
        (tr("broken_total"), len(broken_rows)),
        (tr("score"), site_score),
        (tr("verdict"), grade(site_score)),
    ]
    for sev in ("Critical", "High", "Medium", "Low"):
        summary.append((tr("issues_of", s=SEVLBL[LANG][sev]), sev_count.get(sev, 0)))
    summary.append((tr("note_l"), tr("note")))
    if orphan_note:
        summary.append((tr("note_l"), orphan_note))
    summary += site_issue_rows

    sh = S[LANG]["sheets"]
    with pd.ExcelWriter(args.out, engine="openpyxl") as w:
        pd.DataFrame(summary, columns=S[LANG]["sum_cols"]).to_excel(w, sheet_name=sh[0], index=False)
        pd.DataFrame(top_list, columns=S[LANG]["top_cols"]).to_excel(w, sheet_name=sh[1], index=False)
        pd.DataFrame(issue_list, columns=S[LANG]["iss_cols"]).to_excel(w, sheet_name=sh[2], index=False)
        pd.DataFrame(page_rows).rename(columns=S[LANG]["page_cols"]).to_excel(w, sheet_name=sh[3], index=False)
        pd.DataFrame(broken_rows, columns=S[LANG]["bl_cols"]).to_excel(w, sheet_name=sh[4], index=False)
        if speed_rows:
            pd.DataFrame(speed_rows, columns=S[LANG]["speed_cols"]).to_excel(w, sheet_name=sh[5], index=False)
        style_sheet(w.sheets[sh[0]], {"A": 36, "B": 90})
        style_sheet(w.sheets[sh[1]], {"A": 12, "B": 48, "C": 14, "D": 12, "E": 60, "F": 55}, sev_col=0)
        style_sheet(w.sheets[sh[2]], {"A": 50, "B": 12, "C": 45, "D": 40, "E": 55, "F": 50, "G": 16}, sev_col=1)
        style_sheet(w.sheets[sh[3]], {"A": 50, "B": 9, "C": 40})
        style_sheet(w.sheets[sh[4]], {"A": 50, "B": 50, "C": 10, "D": 12, "E": 30})
        if speed_rows:
            style_sheet(w.sheets[sh[5]], {"A": 50, "B": 12, "C": 20, "D": 12, "E": 12, "F": 10, "G": 16, "H": 16, "I": 28, "J": 22})

    print("\n" + "=" * 50)
    print(tr("p_score", s=site_score, g=grade(site_score)))
    print("   " + " | ".join(f"{SEVLBL[LANG][s]}: {sev_count.get(s, 0)}"
                              for s in ("Critical", "High", "Medium", "Low")))
    print(f"   {tr('broken_total')}: {len(broken_rows)}")
    print(tr("p_top"))
    for x in top_list[:5]:
        print(tr("p_top_row", sev=x[0], issue=x[1], n=x[2]))
    print(tr("p_file", f=args.out))

    try:  # In Google Colab the file downloads automatically
        from google.colab import files  # type: ignore
        files.download(args.out)
    except Exception:
        pass


if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
SEO Audit v4 (Technical + On-Page) from sitemap.xml  |  فحص SEO تقني وفحص صفحات من خريطة الموقع

- Reads robots.txt and sitemap.xml (including sitemap index files)
- Checks pages one by one, discovers extra pages via internal links, finds broken links and orphan pages
- Optional Google PageSpeed test (--psi N) and a warning for JavaScript-rendered pages
- Gives a score out of 100 (higher = better)
- Exports an Excel report: issues, how to fix them, and where to fix them
- NEW: detects the site platform (Blogger / WordPress / Shopify / Wix / other) and gives platform-specific "where to fix" tips
- NEW: also exports a colourful professional PDF summary report
- Report language: Modern Standard Arabic (--lang ar) or English (--lang en)

Install:
    pip install requests beautifulsoup4 pandas openpyxl

Usage:
    python seo_audit_v3.py https://example.com
    python seo_audit_v3.py https://example.com --lang en --max 200 --out report.xlsx
    python seo_audit_v3.py https://example.com --max 0 --crawl 200   (all sitemap pages + 200 discovered pages)
    python seo_audit_v3.py https://example.com --psi 5                (Google PageSpeed on 5 pages; default 1 = homepage; 0 = off)
"""
import argparse
import datetime
import gzip
import os
import re
import subprocess
import sys
import time
from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from urllib.parse import urldefrag, urljoin, urlparse
from urllib.robotparser import RobotFileParser

import pandas as pd
import requests
from bs4 import BeautifulSoup

# ======================================================================
# 👇👇👇  هنا كتبدل رابط الموقع اللي بغيتي تفحصو (الموقع الحالي: odysseedelaterre.fr)
#         CHANGE THE WEBSITE TO AUDIT HERE (the only line you need to edit)
# ======================================================================
SITE_URL = "https://odysseedelaterre.fr"

# 👇 لغة التقرير: "aren" = عربية فصحى  |  "en" = English
#    REPORT LANGUAGE: "en" = Arabic  |  "en" = English
REPORT_LANG = "en"

# 👇 (اختياري) مفتاح Google PageSpeed. خليه فارغ "" إلا ما عندكش
PSI_API_KEY = "AIzaSyD0XbcuK74WPFo0umfMJTEHas5ppSDm3nk"

# 👇 الاسم الذي يظهر في تقرير PDF (غيّره لاسمك أو اسم وكالتك)
#    Name shown on the PDF report
AUTHOR_NAME = "Youssef Ouhmou"
# ======================================================================
# (إلا كتبتي رابط فالأمر %run هو اللي كيتطبق، وإلا ماكتبتيش كيتطبق SITE_URL)

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/124.0 Safari/537.36 SEOAuditBot/1.0"
}

# Points deducted from a page's score for each issue, by severity
WEIGHTS = {"Critical": 25, "High": 12, "Medium": 5, "Low": 2, "Info": 0}
SEV_ORDER = {"Critical": 0, "High": 1, "Medium": 2, "Low": 3, "Info": 4}
SEVLBL = {
    "ar": {"Critical": "حرج", "High": "عالٍ", "Medium": "متوسط", "Low": "منخفض", "Info": "ملاحظة"},
    "en": {"Critical": "Critical", "High": "High", "Medium": "Medium", "Low": "Low", "Info": "Info"},
}

LANG = "ar"
PLATFORM = "other"   # blogger / wordpress / shopify / wix / squarespace / webflow / other (detected in main)

# ------------------------------------------------------------------ النصوص / UI strings
S = {
    "ar": {
        "chars": "{n} حرفًا", "words": "{n} كلمة", "sec": "{n} ثانية",
        "robots_dis": "ممنوعة بقاعدة Disallow في robots.txt",
        "chain": "{n} عمليات إعادة توجيه ← {u}",
        "redir": "يعيد التوجيه إلى {u}",
        "img": "{a} من أصل {b} صورة بلا نص بديل",
        "mixed": "{n} ملفًا يُحمَّل عبر http",
        "yes": "نعم", "no": "لا",
        "g90": "ممتاز", "g75": "جيد", "g50": "متوسط (يحتاج إلى تحسينات)", "g0": "ضعيف (مشكلات كبيرة)",
        "sheets": ["الملخص", "أهم الإصلاحات", "كل المشكلات", "الصفحات", "الروابط المكسورة", "السرعة"],
        "status_n": "رمز الاستجابة {n}",
        "page_cols": {"URL": "الرابط", "Status": "الحالة", "Final URL": "الرابط النهائي", "Redirects": "عدد إعادات التوجيه",
                      "Response (s)": "زمن الاستجابة (ثانية)", "Size (KB)": "الحجم (كيلوبايت)", "Title": "عنوان الصفحة",
                      "Title len": "طول العنوان", "Meta description": "وصف الميتا", "Desc len": "طول الوصف",
                      "H1": "العنوان الرئيسي H1", "H1 count": "عدد H1", "Canonical": "Canonical", "Noindex": "Noindex",
                      "Images": "عدد الصور", "Images no alt": "صور بلا نص بديل", "Internal links": "روابط داخلية",
                      "External links": "روابط خارجية", "Words": "عدد الكلمات", "Score": "التقييم (من 100)",
                      "Issues": "عدد المشكلات"},
        "bl_cols": ["الصفحة المصدر", "الرابط المكسور", "الحالة", "النوع", "نص الرابط"],
        "int": "داخلي", "ext": "خارجي",
        "speed_cols": ["الرابط", "الجهاز", "الأداء (من 100)", "FCP (ثانية)", "LCP (ثانية)", "CLS", "TBT (مللي ثانية)",
                       "TTFB (مللي ثانية)", "تقييم المستخدمين الحقيقيين (28 يومًا)", "INP للمستخدمين (مللي ثانية)"],
        "broken_n": "{n} رابط مكسور",
        "js_d": "نص قليل مع عناصر JavaScript كثيرة",
        "psi_d": "الأداء {s}/100",
        "crawl_found": "صفحات أُضيفت للفحص عبر الروابط الداخلية (خارج الخريطة)",
        "links_checked": "عدد الروابط التي جرى فحصها",
        "broken_total": "عدد الروابط المكسورة",
        "orphan_skip": "تم تخطي كشف الصفحات اليتيمة لأن الفحص لم يشمل كل الصفحات. استخدم --max 0 مع --crawl أكبر.",
        "p_crawl": "🕸️ تم اكتشاف {n} صفحة إضافية عبر الروابط الداخلية، جارٍ فحصها ...",
        "p_links": "🔗 جارٍ فحص {n} رابط بحثًا عن الروابط المكسورة ...",
        "p_psi": "⚡ جارٍ فحص السرعة عبر Google PageSpeed ({n} صفحة، على الجوال والكمبيوتر، قد يستغرق عدة دقائق) ...",
        "dev_mobile": "جوال", "dev_desktop": "كمبيوتر",
        "p_psi_err": "⚠️ تعذّر فحص PageSpeed ({d}) للرابط {u} ({e}). إن كان الخطأ 429 فالحصة المجانية استُنفدت: جرّب لاحقًا أو استعمل --psi-key، أو افحص الرابط يدويًا في pagespeed.web.dev",
        "sum_cols": ["البند", "القيمة"],
        "top_cols": ["الخطورة", "المشكلة", "عدد الصفحات المتأثرة", "% من الصفحات", "طريقة الإصلاح", "مكان الإصلاح"],
        "iss_cols": ["الرابط", "الخطورة", "المشكلة", "التفاصيل", "طريقة الإصلاح", "مكان الإصلاح", "الرمز"],
        "site": "الموقع", "sitemap": "خريطة الموقع (Sitemap)", "sm_total": "عدد الروابط في الخريطة",
        "checked": "عدد الصفحات المفحوصة", "score": "التقييم العام (من 100)", "verdict": "الحكم",
        "issues_of": "عدد المشكلات ({s})",
        "note_l": "ملاحظة",
        "note": "التقييم من 100: كلما ارتفع الرقم كان الموقع أفضل. هذا تقييم تقديري من هذه الأداة وليس مقياسًا رسميًا من Google.",
        "site_issue": "مشكلة على مستوى الموقع",
        "no_robots": "لا يوجد ملف robots.txt",
        "no_robots_fix": "أنشئ ملف robots.txt في جذر الموقع وأضف إليه رابط خريطة الموقع.",
        "no_robots_where": "جذر الموقع: /robots.txt",
        "p_enter": "🔗 أدخل رابط الموقع (مثال: https://example.com): ",
        "p_search": "⏳ جارٍ البحث عن خريطة الموقع لـ {b} ...",
        "p_nosm": "❌ لم يتم العثور على خريطة موقع (robots.txt أو /sitemap.xml أو /sitemap_index.xml أو /wp-sitemap.xml).",
        "p_found": "✅ خريطة الموقع: {sm} | {n} رابط",
        "p_sample": "ℹ️ سيتم فحص {m} صفحة (عينة موزعة على الموقع). استخدم --max 0 لفحص الكل.",
        "p_prog": "   تم فحص {n}/{m} صفحة",
        "p_score": "🏁 التقييم العام: {s}/100 ← {g}",
        "p_top": "🔧 أهم المشكلات التي يجب إصلاحها:",
        "p_top_row": "   [{sev}] {issue} ({n} صفحة)",
        "p_file": "📄 التقرير الكامل: {f}",
        "platform": "منصة الموقع", "plat_unknown": "غير محدد",
        "plat_note": "لم يتم التعرف على منصة الموقع، لذلك مكان الإصلاح عام. نفّذ الحلول من لوحة تحكم منصتك أو اطلب ذلك من المطوّر.",
        "menu_note": "أسماء القوائم والأزرار قد تختلف قليلًا حسب تحديثات المنصة.",
        "p_platform": "🧭 منصة الموقع: {p}",
        "pdf_title": "تقرير فحص SEO", "pdf_prepared": "إعداد", "pdf_date": "التاريخ", "pdf_platform": "المنصة",
        "pdf_pages": "الصفحات المفحوصة", "pdf_broken": "روابط مكسورة", "pdf_crit": "مشكلات حرجة", "pdf_high": "مشكلات عالية",
        "pdf_sev": "توزيع المشكلات حسب الخطورة", "pdf_speed": "سرعة الجوال (Google PageSpeed)",
        "pdf_exec": "الملخص التنفيذي", "pdf_top": "أهم الإصلاحات", "pdf_pg": "صفحة",
        "pdf_how": "طريقة الإصلاح", "pdf_where": "مكان الإصلاح",
        "pdf_none": "لم تظهر مشكلات واضحة ضمن الفحوصات الأساسية. تابع الأداء في Search Console واستمر في تحسين المحتوى.",
        "pdf_file": "📕 تقرير PDF: {f}",
        "pdf_nofont": "⚠️ تعذّر العثور على خط يدعم العربية، لذلك لم يُنشأ PDF. جرّب --lang en.",
        "exec_tpl": "تم فحص {pages} صفحة من الموقع {site} (المنصة: {platform}). التقييم العام {score} من 100 ({grade}). وُجدت {n_issues} مشكلة، منها {n_crit} حرجة و{n_high} عالية الخطورة. نوصي بالبدء بما يلي: {top}. ثم تنفيذ بقية الإصلاحات بالترتيب الموضح في ملف Excel المرفق.",
        "rtl": True,
    },
    "en": {
        "chars": "{n} chars", "words": "{n} words", "sec": "{n} s",
        "robots_dis": "Disallowed by robots.txt",
        "chain": "{n} redirects -> {u}",
        "redir": "Redirects to {u}",
        "img": "{a} of {b} images have no alt text",
        "mixed": "{n} resources loaded over http",
        "yes": "Yes", "no": "No",
        "g90": "Excellent", "g75": "Good", "g50": "Average (needs work)", "g0": "Poor (major issues)",
        "sheets": ["Summary", "Top Fixes", "All Issues", "Pages", "Broken Links", "Speed"],
        "status_n": "Status {n}",
        "page_cols": {},
        "bl_cols": ["Source page", "Broken link", "Status", "Type", "Anchor text"],
        "int": "Internal", "ext": "External",
        "speed_cols": ["URL", "Device", "Performance (/100)", "FCP (s)", "LCP (s)", "CLS", "TBT (ms)",
                       "TTFB (ms)", "Real-user rating (28 days)", "Real-user INP (ms)"],
        "broken_n": "{n} broken links",
        "js_d": "Very little text but many JavaScript elements",
        "psi_d": "Performance {s}/100",
        "crawl_found": "Pages added through internal links (not in sitemap)",
        "links_checked": "Links checked",
        "broken_total": "Broken links found",
        "orphan_skip": "Orphan-page detection skipped because not all pages were audited. Use --max 0 with a larger --crawl.",
        "p_crawl": "🕸️ Found {n} extra pages through internal links, auditing them ...",
        "p_links": "🔗 Checking {n} links for broken ones ...",
        "p_psi": "⚡ Running Google PageSpeed ({n} pages, mobile and desktop, may take several minutes) ...",
        "dev_mobile": "Mobile", "dev_desktop": "Desktop",
        "p_psi_err": "⚠️ PageSpeed failed ({d}) for {u} ({e}). If it is 429 the free quota is used up: try later or use --psi-key, or test the URL manually at pagespeed.web.dev",
        "sum_cols": ["Item", "Value"],
        "top_cols": ["Severity", "Issue", "Pages affected", "% of pages", "How to fix", "Where to fix"],
        "iss_cols": ["URL", "Severity", "Issue", "Detail", "How to fix", "Where to fix", "Code"],
        "site": "Website", "sitemap": "Sitemap", "sm_total": "URLs in sitemap",
        "checked": "Pages audited", "score": "Overall score (out of 100)", "verdict": "Verdict",
        "issues_of": "Issues ({s})",
        "note_l": "Note",
        "note": "Score is out of 100: the higher the better. It is this tool's own estimate, not an official Google metric.",
        "site_issue": "Site-wide issue",
        "no_robots": "No robots.txt file",
        "no_robots_fix": "Create a robots.txt file in the site root and add the sitemap URL to it.",
        "no_robots_where": "Site root: /robots.txt",
        "p_enter": "🔗 Enter the website URL (e.g. https://example.com): ",
        "p_search": "⏳ Looking for the sitemap of {b} ...",
        "p_nosm": "❌ No sitemap found (robots.txt, /sitemap.xml, /sitemap_index.xml, /wp-sitemap.xml).",
        "p_found": "✅ Sitemap: {sm} | {n} URLs",
        "p_sample": "ℹ️ Auditing {m} pages (an evenly spread sample). Use --max 0 to audit all.",
        "p_prog": "   {n}/{m} pages audited",
        "p_score": "🏁 Overall score: {s}/100 -> {g}",
        "p_top": "🔧 Top issues to fix:",
        "p_top_row": "   [{sev}] {issue} ({n} pages)",
        "p_file": "📄 Full report: {f}",
        "platform": "Site platform", "plat_unknown": "Not detected",
        "plat_note": "The site platform was not recognised, so the 'where to fix' tips are generic. Apply the fixes from your platform's admin panel or ask your developer.",
        "menu_note": "Menu and button names may differ slightly as platforms update.",
        "p_platform": "🧭 Site platform: {p}",
        "pdf_title": "SEO Audit Report", "pdf_prepared": "Prepared by", "pdf_date": "Date", "pdf_platform": "Platform",
        "pdf_pages": "Pages audited", "pdf_broken": "Broken links", "pdf_crit": "Critical issues", "pdf_high": "High issues",
        "pdf_sev": "Issues by severity", "pdf_speed": "Mobile speed (Google PageSpeed)",
        "pdf_exec": "Executive summary", "pdf_top": "Top fixes", "pdf_pg": "pages",
        "pdf_how": "How to fix", "pdf_where": "Where to fix",
        "pdf_none": "No clear issues were found in the basic checks. Keep tracking performance in Search Console and keep improving content.",
        "pdf_file": "📕 PDF report: {f}",
        "pdf_nofont": "⚠️ No Arabic-capable font found, so the PDF was not created. Try --lang en.",
        "exec_tpl": "We audited {pages} pages of {site} (platform: {platform}). The overall score is {score} out of 100 ({grade}). We found {n_issues} issues, {n_crit} critical and {n_high} high severity. We recommend starting with: {top}. The remaining fixes follow, in order, in the attached Excel report.",
        "rtl": False,
    },
}


def tr(key, **kw):
    return S[LANG][key].format(**kw)


# ------------------------------------------------------------------ كتالوج المشكلات / Issue catalog
# code: (severity, (desc_ar, desc_en), (fix_ar, fix_en), (where_ar, where_en))
CATALOG = {
    "FETCH_FAIL": ("Critical",
        ("تعذّر تحميل الصفحة (انتهاء المهلة أو خطأ في الاتصال)", "Page failed to load (timeout or connection error)"),
        ("تأكد من أن الخادم يعمل بسرعة جيدة، وأن جدار الحماية أو Cloudflare لا يحجب الزوار.",
         "Make sure the server is up and fast, and that a firewall or Cloudflare is not blocking visitors."),
        ("الاستضافة / الخادم / إعدادات جدار الحماية", "Hosting / server / firewall settings")),
    "HTTP_4XX": ("Critical",
        ("الصفحة تعرض خطأ 4xx (مثل 404: الصفحة غير موجودة)", "Page returns a 4xx error (e.g. 404 Not Found)"),
        ("أزل الرابط من خريطة الموقع، أو أنشئ إعادة توجيه 301 إلى الصفحة البديلة، أو أعد نشر الصفحة.",
         "Remove the URL from the sitemap, add a 301 redirect to the replacement page, or restore the page."),
        ("ملف sitemap.xml وإعدادات إعادة التوجيه (.htaccess / Cloudflare / إضافة Redirection)",
         "sitemap.xml and redirect settings (.htaccess / Cloudflare / Redirection plugin)")),
    "HTTP_5XX": ("Critical",
        ("خطأ في الخادم (5xx)", "Server error (5xx)"),
        ("راجع سجلات الخادم (logs)؛ غالبًا تكون المشكلة في PHP أو قاعدة البيانات أو إضافة. تواصل مع الاستضافة إن لم تجد السبب.",
         "Check the server logs; it is usually a PHP, database or plugin issue. Contact your host if you cannot find the cause."),
        ("الخادم / الاستضافة / السجلات", "Server / hosting / logs")),
    "NOINDEX": ("Critical",
        ("الصفحة تحتوي على noindex (تمنع Google من فهرستها) رغم وجودها في خريطة الموقع",
         "Page has noindex (blocks Google from indexing it) yet is listed in the sitemap"),
        ("إن أردت ظهورها في Google فأزل noindex، وإن لم ترد ذلك فأزلها من خريطة الموقع.",
         "Remove noindex if the page should appear in Google; otherwise remove it from the sitemap."),
        ("وسم robots في <head> / ترويسة X-Robots-Tag / إعدادات Yoast أو RankMath",
         "robots meta tag in <head> / X-Robots-Tag header / Yoast or RankMath settings")),
    "ROBOTS_BLOCKED": ("Critical",
        ("الصفحة محظورة في ملف robots.txt", "Page is blocked by robots.txt"),
        ("عدّل قاعدة Disallow التي تحجب هذا الرابط.", "Edit the Disallow rule that blocks this URL."),
        ("ملف robots.txt في جذر الموقع", "robots.txt in the site root")),
    "NOT_HTTPS": ("High",
        ("الصفحة لا تعمل عبر HTTPS", "Page is not served over HTTPS"),
        ("فعّل شهادة SSL وأضف إعادة توجيه 301 من http إلى https لكامل الموقع.",
         "Install an SSL certificate and 301-redirect http to https site-wide."),
        ("الاستضافة (SSL) + .htaccess / Cloudflare / إعدادات الموقع", "Hosting (SSL) + .htaccess / Cloudflare / site settings")),
    "MIXED_CONTENT": ("High",
        ("صفحة HTTPS تحمّل ملفات (صور أو سكربتات) عبر http", "HTTPS page loads resources (images/scripts) over http"),
        ("استبدل كل روابط http:// بـ https:// في الصور وملفات CSS وJS.",
         "Replace all http:// links with https:// for images, CSS and JS."),
        ("كود القالب / المحتوى / إضافة مثل Really Simple SSL", "Theme code / content / a plugin such as Really Simple SSL")),
    "REDIRECT": ("Medium",
        ("الرابط في خريطة الموقع يعيد التوجيه إلى رابط آخر", "URL in the sitemap redirects elsewhere"),
        ("ضع الرابط النهائي (الذي يعطي حالة 200) في خريطة الموقع بدل الرابط القديم.",
         "Put the final URL (status 200) in the sitemap instead of the old one."),
        ("ملف sitemap.xml", "sitemap.xml")),
    "REDIRECT_CHAIN": ("High",
        ("سلسلة إعادة توجيه (أكثر من إعادة توجيه واحدة)", "Redirect chain (more than one redirect)"),
        ("اجعلها إعادة توجيه واحدة مباشرة من الرابط القديم إلى الرابط النهائي.",
         "Use a single direct redirect from the old URL to the final URL."),
        ("إعدادات إعادة التوجيه (.htaccess / Cloudflare / إضافة)", "Redirect settings (.htaccess / Cloudflare / plugin)")),
    "SLOW": ("Medium",
        ("الصفحة بطيئة (زمن الاستجابة أكثر من 1.5 ثانية)", "Slow page (response time over 1.5 seconds)"),
        ("فعّل التخزين المؤقت (Cache) وضغط الصور وشبكة CDN، وحسّن الاستضافة. قِس الأداء عبر PageSpeed Insights.",
         "Enable caching, compress images, use a CDN and upgrade hosting. Measure with PageSpeed Insights."),
        ("الاستضافة + إضافة التخزين المؤقت (WP Rocket / LiteSpeed) + الصور",
         "Hosting + cache plugin (WP Rocket / LiteSpeed) + images")),
    "BIG_PAGE": ("Low",
        ("حجم الصفحة كبير (أكثر من 1.5 ميغابايت من HTML)", "Large page (over 1.5 MB of HTML)"),
        ("قلّل الكود الزائد وفعّل ضغط gzip/brotli.", "Reduce unnecessary markup and enable gzip/brotli compression."),
        ("القالب / إعدادات الخادم", "Theme / server settings")),
    "TITLE_MISSING": ("High",
        ("لا يوجد عنوان للصفحة (Title)", "Missing title tag"),
        ("أضف وسم <title> فريدًا يتضمن الكلمة المفتاحية (30-60 حرفًا).",
         "Add a unique <title> containing the target keyword (30-60 characters)."),
        ("وسم <head> للصفحة / حقل عنوان SEO في Yoast أو RankMath", "Page <head> / SEO Title field in Yoast or RankMath")),
    "TITLE_SHORT": ("Medium",
        ("عنوان الصفحة قصير (أقل من 30 حرفًا)", "Title too short (under 30 characters)"),
        ("أطِل العنوان وأضف الكلمة المفتاحية وما يميّز الصفحة (30-60 حرفًا).",
         "Lengthen the title and include the keyword and a clear benefit (30-60 characters)."),
        ("حقل عنوان SEO في Yoast / RankMath أو وسم <title>", "SEO Title field in Yoast / RankMath or the <title> tag")),
    "TITLE_LONG": ("Medium",
        ("عنوان الصفحة طويل (أكثر من 60 حرفًا) وسيُقتطع في نتائج Google",
         "Title too long (over 60 characters) and will be truncated in Google"),
        ("اختصر العنوان وضع الكلمة المهمة في بدايته.", "Shorten the title and put the important keyword first."),
        ("حقل عنوان SEO في Yoast / RankMath أو وسم <title>", "SEO Title field in Yoast / RankMath or the <title> tag")),
    "TITLE_DUP": ("Medium",
        ("العنوان نفسه مستخدم في صفحات أخرى", "Same title used on other pages"),
        ("يجب أن يكون لكل صفحة عنوان خاص بها.", "Give every page its own unique title."),
        ("حقل عنوان SEO في كل صفحة", "SEO Title field on each page")),
    "DESC_MISSING": ("Medium",
        ("لا يوجد وصف ميتا (Meta Description)", "Missing meta description"),
        ("أضف وصفًا جذابًا (70-160 حرفًا) يتضمن الكلمة المفتاحية.",
         "Add a compelling description (70-160 characters) with the target keyword."),
        ("حقل الوصف في Yoast / RankMath أو وسم <meta name='description'> في <head>",
         "Description field in Yoast / RankMath or the <meta name='description'> tag in <head>")),
    "DESC_SHORT": ("Low",
        ("وصف الميتا قصير (أقل من 70 حرفًا)", "Meta description too short (under 70 characters)"),
        ("أطِل الوصف قليلًا (70-160 حرفًا).", "Lengthen it a little (70-160 characters)."),
        ("حقل وصف الميتا", "Meta description field")),
    "DESC_LONG": ("Low",
        ("وصف الميتا طويل (أكثر من 160 حرفًا)", "Meta description too long (over 160 characters)"),
        ("اختصر الوصف حتى لا يُقتطع.", "Shorten it so it is not truncated."),
        ("حقل وصف الميتا", "Meta description field")),
    "DESC_DUP": ("Medium",
        ("وصف الميتا نفسه مستخدم في صفحات أخرى", "Same meta description used on other pages"),
        ("اكتب وصفًا خاصًا بكل صفحة.", "Write a unique description for each page."),
        ("حقل وصف الميتا في كل صفحة", "Meta description field on each page")),
    "H1_MISSING": ("High",
        ("لا يوجد عنوان رئيسي H1", "Missing H1 heading"),
        ("أضف H1 واحدًا في الصفحة يتضمن الكلمة المفتاحية الرئيسية.", "Add a single H1 containing the main keyword."),
        ("محتوى الصفحة / القالب (عنوان الصفحة يجب أن يكون H1)", "Page content / theme (the page title should be an H1)")),
    "H1_MULTI": ("Medium",
        ("أكثر من H1 في الصفحة نفسها", "More than one H1 on the page"),
        ("اترك H1 واحدًا فقط وحوّل البقية إلى H2 أو H3.", "Keep one H1 and change the rest to H2/H3."),
        ("محتوى الصفحة / القالب / أداة بناء الصفحات", "Page content / theme / page builder")),
    "H1_DUP": ("Low",
        ("العنوان الرئيسي H1 نفسه مستخدم في صفحات أخرى", "Same H1 used on other pages"),
        ("اجعل H1 فريدًا لكل صفحة.", "Make each page's H1 unique."),
        ("محتوى الصفحة", "Page content")),
    "CANON_MISSING": ("Medium",
        ("لا يوجد رابط Canonical", "Missing canonical tag"),
        ("أضف <link rel='canonical'> يشير إلى الرابط الأصلي للصفحة.",
         "Add <link rel='canonical'> pointing to the page's preferred URL."),
        ("<head> / Yoast / RankMath (متقدم ← Canonical URL)", "<head> / Yoast / RankMath (Advanced > Canonical URL)")),
    "CANON_OTHER": ("Medium",
        ("رابط Canonical يشير إلى صفحة أخرى", "Canonical points to a different URL"),
        ("تأكد من أن ذلك مقصود، وإلا فاجعل الـ canonical يشير إلى الصفحة نفسها.",
         "Make sure this is intentional; otherwise point the canonical to the page itself."),
        ("<head> / Yoast / RankMath (متقدم ← Canonical URL)", "<head> / Yoast / RankMath (Advanced > Canonical URL)")),
    "IMG_ALT": ("Medium",
        ("صور بلا نص بديل (alt)", "Images missing alt text"),
        ("أضف نصًا بديلًا وصفيًا لكل صورة (يفيد SEO وإمكانية الوصول).",
         "Add descriptive alt text to every image (helps SEO and accessibility)."),
        ("مكتبة الوسائط أو الوسم <img alt=''>", "Media Library or the <img alt=''> attribute")),
    "THIN": ("Medium",
        ("محتوى قليل (أقل من 300 كلمة)", "Thin content (under 300 words)"),
        ("أضف محتوى مفيدًا وأصليًا يجيب عن نية الباحث، وتجنب الصفحات شبه الفارغة.",
         "Add useful, original content that answers search intent; avoid near-empty pages."),
        ("محتوى الصفحة", "Page content")),
    "NO_VIEWPORT": ("High",
        ("لا يوجد وسم viewport (الموقع غير مهيأ للجوال)", "Missing viewport tag (not mobile-friendly)"),
        ("أضف <meta name='viewport' content='width=device-width, initial-scale=1'>.",
         "Add <meta name='viewport' content='width=device-width, initial-scale=1'>."),
        ("<head> / القالب (header.php)", "<head> / theme (header.php)")),
    "NO_LANG": ("Low",
        ("لا يوجد وسم lang في <html>", "Missing lang attribute on <html>"),
        ("أضف lang مثل <html lang='ar'> أو 'fr' أو 'en'.", "Add lang, e.g. <html lang='en'>."),
        ("القالب (header.php) / إعدادات اللغة", "Theme (header.php) / language settings")),
    "NO_OG": ("Low",
        ("لا توجد وسوم Open Graph (معاينة المشاركة في فيسبوك وواتساب)",
         "Missing Open Graph tags (share previews on Facebook/WhatsApp)"),
        ("أضف og:title وog:description وog:image.", "Add og:title, og:description and og:image."),
        ("Yoast / RankMath (التواصل الاجتماعي) أو <head>", "Yoast / RankMath (Social) or <head>")),
    "NO_SCHEMA": ("Low",
        ("لا توجد بيانات منظمة (Schema)", "No structured data (Schema)"),
        ("أضف Schema بصيغة JSON-LD مناسبة (Article / Product / LocalBusiness / FAQ ...).",
         "Add suitable JSON-LD Schema (Article / Product / LocalBusiness / FAQ ...)."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    "URL_LONG": ("Low",
        ("الرابط طويل (أكثر من 115 حرفًا)", "URL too long (over 115 characters)"),
        ("اختصر الرابط واكتفِ بكلمات مفتاحية قليلة.", "Shorten the URL and keep only a few keywords."),
        ("الرابط الدائم (Slug) للصفحة", "Page slug / permalink")),
    "URL_STYLE": ("Low",
        ("الرابط يحتوي على أحرف كبيرة أو شرطة سفلية أو معاملات (parameters)",
         "URL contains uppercase letters, underscores or parameters"),
        ("استخدم أحرفًا صغيرة وشرطة (-) بدل الشرطة السفلية، وتجنب المعاملات.",
         "Use lowercase and hyphens instead of underscores; avoid parameters."),
        ("الرابط الدائم / إعدادات الروابط", "Slug / permalink settings")),
    "NO_INTERNAL_LINKS": ("Medium",
        ("لا توجد روابط داخلية في الصفحة", "No internal links on the page"),
        ("أضف روابط داخلية إلى صفحات مهمة أخرى في الموقع.", "Add internal links to other important pages."),
        ("محتوى الصفحة / القائمة", "Page content / menu")),
    "BROKEN_INTERNAL": ("High",
        ("روابط داخلية مكسورة (تؤدي إلى صفحات غير موجودة أو بها خطأ)", "Broken internal links (point to missing or erroring pages)"),
        ("عدّل الرابط ليشير إلى صفحة صحيحة، أو أنشئ إعادة توجيه 301، أو أزل الرابط. التفاصيل في ورقة «الروابط المكسورة».",
         "Point the link to a valid page, add a 301 redirect, or remove the link. See the 'Broken Links' sheet for details."),
        ("محتوى الصفحة / القائمة / القالب", "Page content / menu / theme")),
    "BROKEN_EXTERNAL": ("Medium",
        ("روابط خارجية مكسورة", "Broken external links"),
        ("استبدل الرابط بمصدر يعمل أو أزله. التفاصيل في ورقة «الروابط المكسورة».",
         "Replace the link with a working source or remove it. See the 'Broken Links' sheet for details."),
        ("محتوى الصفحة", "Page content")),
    "NOT_IN_SITEMAP": ("Low",
        ("صفحة موجودة ويمكن الوصول إليها لكنها غير مدرجة في خريطة الموقع", "Live page that is not listed in the sitemap"),
        ("أضفها إلى خريطة الموقع إن كانت صفحة مهمة، أو أضف لها noindex إن لم تكن كذلك.",
         "Add it to the sitemap if it is important, or noindex it if it is not."),
        ("sitemap.xml / إعدادات الخرائط في Yoast أو RankMath", "sitemap.xml / Sitemap settings in Yoast or RankMath")),
    "ORPHAN": ("Medium",
        ("صفحة يتيمة: لا توجد أي صفحة مفحوصة تربط إليها داخليًا", "Orphan page: no audited page links to it"),
        ("أضف روابط داخلية إليها من صفحات ذات صلة أو من القائمة الرئيسية.",
         "Add internal links to it from related pages or the main menu."),
        ("محتوى الصفحات / القائمة", "Page content / menu")),
    "HEADING_SKIP": ("Low",
        ("ترتيب العناوين غير منطقي (قفز مثلًا من H1 إلى H3)", "Illogical heading order (e.g. jumping from H1 to H3)"),
        ("رتّب العناوين H1 ثم H2 ثم H3 دون قفز بين المستويات.", "Order headings H1, H2, H3 without skipping levels."),
        ("محتوى الصفحة / القالب", "Page content / theme")),
    "JS_SUSPECT": ("Info",
        ("الصفحة تبدو معتمدة على JavaScript لعرض المحتوى، وقد تكون بعض نتائج الفحص غير دقيقة",
         "Page seems to rely on JavaScript to render content, so some results may be inaccurate"),
        ("تحقق من الصفحة يدويًا في المتصفح (قارن View Source مع Inspect) قبل الاعتماد على نتائجها.",
         "Check the page manually in the browser (compare View Source with Inspect) before trusting its results."),
        ("-", "-")),
    "PSI_POOR": ("Medium",
        ("أداء ضعيف على الجوال حسب Google PageSpeed (أقل من 50)", "Poor mobile performance per Google PageSpeed (below 50)"),
        ("اضغط الصور، فعّل التخزين المؤقت وCDN، قلّل ملفات JS وCSS غير الضرورية، وراجع تقرير PageSpeed للتفاصيل.",
         "Compress images, enable caching and a CDN, cut unneeded JS/CSS, and review the PageSpeed report for details."),
        ("الصور / القالب / الإضافات / الاستضافة", "Images / theme / plugins / hosting")),
    "PSI_AVG": ("Low",
        ("أداء متوسط على الجوال حسب Google PageSpeed (50-89)", "Average mobile performance per Google PageSpeed (50-89)"),
        ("حسّن الصور والتخزين المؤقت وقلّل الملفات الثقيلة.", "Improve images and caching, and reduce heavy files."),
        ("الصور / القالب / الإضافات / الاستضافة", "Images / theme / plugins / hosting")),
}


# ------------------------------------------------------------------ منصة الموقع / Platform-specific tips
# كل مشكلة لها "مكان إصلاح" يختلف حسب المنصة. WordPress يبقى كما في الكتالوج، والمنصات الأخرى لها نصوصها.
CODE_GROUP = {
    "HTTP_4XX": "redirect", "REDIRECT_CHAIN": "redirect", "NOINDEX": "noindex", "NOT_HTTPS": "https",
    "MIXED_CONTENT": "mixed", "SLOW": "slow", "TITLE_MISSING": "title", "TITLE_SHORT": "title",
    "TITLE_LONG": "title", "TITLE_DUP": "title", "DESC_MISSING": "desc", "DESC_DUP": "desc",
    "CANON_MISSING": "canon", "CANON_OTHER": "canon", "NO_OG": "og", "NO_SCHEMA": "schema",
    "NOT_IN_SITEMAP": "sitemap", "NO_VIEWPORT": "viewport", "NO_LANG": "lang", "IMG_ALT": "alt",
    "URL_LONG": "slug", "URL_STYLE": "slug", "ROBOTS_BLOCKED": "robots",
}
# group: (arabic, english)
PW = {
    "blogger": {
        "title": ("عنوان المقال نفسه في المحرر. وعنوان الصفحة الرئيسية: Settings ← Title", "The post title itself in the editor. Homepage title: Settings > Title"),
        "desc": ("Settings ← Search preferences ← Meta description للمدونة، وللمقال: Post settings ← Search description", "Settings > Search preferences > Meta description for the blog; for a post: Post settings > Search description"),
        "canon": ("Blogger يضيفه تلقائيًا في أغلب القوالب؛ راجع القالب: Theme ← Edit HTML", "Blogger adds it automatically in most themes; check the theme: Theme > Edit HTML"),
        "redirect": ("Settings ← Search preferences ← Custom redirects (وإزالة الرابط من خريطة الموقع إن لزم)", "Settings > Search preferences > Custom redirects (and remove the URL from the sitemap if needed)"),
        "noindex": ("Settings ← Search preferences ← Custom robots header tags، وللمقال: Post settings ← Custom robots tags", "Settings > Search preferences > Custom robots header tags; for a post: Post settings > Custom robots tags"),
        "https": ("Settings ← HTTPS ← فعّل HTTPS redirect", "Settings > HTTPS > enable HTTPS redirect"),
        "mixed": ("Theme ← Edit HTML ومحتوى المقالات: استبدل http:// بـ https://", "Theme > Edit HTML and post content: replace http:// with https://"),
        "slow": ("Layout (احذف الأدوات غير اللازمة) + ضغط الصور + قالب خفيف", "Layout (remove unneeded gadgets) + compress images + a lightweight theme"),
        "og": ("Theme ← Edit HTML: أضف وسوم og داخل <head> أو اختر قالبًا يتضمنها", "Theme > Edit HTML: add og tags inside <head> or choose a theme that includes them"),
        "schema": ("Theme ← Edit HTML: أضف JSON-LD داخل <head>", "Theme > Edit HTML: add JSON-LD inside <head>"),
        "sitemap": ("Blogger ينشئ خريطة الموقع تلقائيًا (/sitemap.xml)، والصفحات الثابتة غالبًا في /sitemap-pages.xml", "Blogger creates the sitemap automatically (/sitemap.xml); static pages are usually in /sitemap-pages.xml"),
        "viewport": ("Theme ← Edit HTML: وسم viewport في <head>، أو بدّل القالب بقالب متجاوب", "Theme > Edit HTML: viewport tag in <head>, or switch to a responsive theme"),
        "lang": ("Theme ← Edit HTML: خاصية lang في وسم <html>", "Theme > Edit HTML: the lang attribute on the <html> tag"),
        "alt": ("محرر Blogger: اضغط على الصورة ← Image properties ← Alt text", "Blogger editor: click the image > Image properties > Alt text"),
        "slug": ("Post settings ← Permalink (رابط مخصص)", "Post settings > Permalink (custom permalink)"),
        "robots": ("Settings ← Search preferences ← Custom robots.txt", "Settings > Search preferences > Custom robots.txt"),
    },
    "shopify": {
        "title": ("الصفحة أو المنتج ← Search engine listing ← Edit. وللرئيسية: Online Store ← Preferences", "Page or product > Search engine listing > Edit. Homepage: Online Store > Preferences"),
        "desc": ("الصفحة أو المنتج ← Search engine listing ← Edit. وللرئيسية: Online Store ← Preferences", "Page or product > Search engine listing > Edit. Homepage: Online Store > Preferences"),
        "canon": ("Shopify يضيفه تلقائيًا في القوالب الرسمية؛ راجع theme.liquid", "Shopify adds it automatically in official themes; check theme.liquid"),
        "redirect": ("Online Store ← Navigation ← URL redirects", "Online Store > Navigation > URL redirects"),
        "noindex": ("Search engine listing للصفحة أو ملف theme.liquid", "The page's Search engine listing or theme.liquid"),
        "https": ("Shopify يفعّله تلقائيًا؛ راجع Settings ← Domains", "Shopify enables it automatically; check Settings > Domains"),
        "mixed": ("كود القالب (Online Store ← Themes ← Edit code) أو المحتوى", "Theme code (Online Store > Themes > Edit code) or content"),
        "slow": ("قلّل التطبيقات (Apps) غير اللازمة + ضغط الصور", "Remove unneeded apps + compress images"),
        "og": ("theme.liquid أو تطبيق SEO", "theme.liquid or an SEO app"),
        "schema": ("theme.liquid (JSON-LD) أو تطبيق SEO", "theme.liquid (JSON-LD) or an SEO app"),
        "sitemap": ("Shopify ينشئ /sitemap.xml تلقائيًا", "Shopify creates /sitemap.xml automatically"),
        "viewport": ("theme.liquid ← وسم viewport في <head>", "theme.liquid > viewport tag in <head>"),
        "lang": ("theme.liquid ← lang في وسم <html>", "theme.liquid > lang on the <html> tag"),
        "alt": ("اضغط على صورة المنتج أو الصفحة ← Add alt text", "Click the product or page image > Add alt text"),
        "slug": ("URL handle ضمن Search engine listing", "URL handle inside Search engine listing"),
        "robots": ("robots.txt.liquid في القالب", "robots.txt.liquid in the theme"),
    },
    "wix": {
        "title": ("Pages ← النقاط الثلاث ← SEO basics", "Pages > three dots > SEO basics"),
        "desc": ("Pages ← النقاط الثلاث ← SEO basics", "Pages > three dots > SEO basics"),
        "canon": ("Wix يضيفه تلقائيًا؛ ويمكن تعديله من Page SEO ← Advanced SEO", "Wix adds it automatically; it can be edited in Page SEO > Advanced SEO"),
        "redirect": ("SEO Tools ← URL Redirect Manager", "SEO Tools > URL Redirect Manager"),
        "noindex": ("Pages ← SEO basics ← خيار السماح لمحركات البحث بفهرسة الصفحة", "Pages > SEO basics > the option that lets search engines index the page"),
        "https": ("Wix يفعّله تلقائيًا؛ راجع إعدادات الدومين", "Wix enables it automatically; check the domain settings"),
        "mixed": ("محتوى الصفحة: استبدل http:// بـ https://", "Page content: replace http:// with https://"),
        "slow": ("قلّل الوسائط والتطبيقات (Apps) وضغط الصور", "Reduce media and apps, compress images"),
        "og": ("Pages ← SEO basics ← Social share", "Pages > SEO basics > Social share"),
        "schema": ("Page SEO ← Advanced SEO ← Structured data markup", "Page SEO > Advanced SEO > Structured data markup"),
        "sitemap": ("Wix ينشئ خريطة الموقع تلقائيًا", "Wix creates the sitemap automatically"),
        "viewport": ("يتحكم به Wix تلقائيًا، وقد تكون النتيجة غير دقيقة", "Controlled by Wix automatically, so this result may be inaccurate"),
        "lang": ("يتحكم به Wix حسب لغة الموقع في الإعدادات", "Controlled by Wix from the site language in settings"),
        "alt": ("اضغط على الصورة ← Settings ← What's in the image? (Alt text)", "Click the image > Settings > What's in the image? (Alt text)"),
        "slug": ("Pages ← SEO basics ← URL slug", "Pages > SEO basics > URL slug"),
        "robots": ("SEO Tools ← Robots.txt Editor", "SEO Tools > Robots.txt Editor"),
    },
    "other": {   # منصة غير معروفة: نص عام بلا أسماء إضافات
        "title": ("حقل عنوان SEO للصفحة في لوحة تحكم موقعك أو وسم <title> في <head>", "The page's SEO title field in your admin panel, or the <title> tag in <head>"),
        "desc": ("حقل وصف SEO للصفحة في لوحة تحكم موقعك أو وسم <meta name='description'>", "The page's SEO description field in your admin panel, or the <meta name='description'> tag"),
        "canon": ("إعدادات SEO للصفحة أو وسم <link rel='canonical'> في <head>", "The page's SEO settings, or the <link rel='canonical'> tag in <head>"),
        "redirect": ("إعدادات إعادة التوجيه في منصتك أو الاستضافة", "Redirect settings in your platform or hosting"),
        "noindex": ("إعدادات فهرسة الصفحة في منصتك أو وسم robots في <head>", "The page's indexing setting in your platform, or the robots meta tag in <head>"),
        "https": ("الاستضافة (SSL) وإعدادات الدومين في منصتك", "Hosting (SSL) and domain settings in your platform"),
        "mixed": ("كود القالب أو محتوى الصفحة", "Theme code or page content"),
        "slow": ("الاستضافة + ضغط الصور + تقليل الإضافات أو التطبيقات", "Hosting + compress images + reduce plugins or apps"),
        "og": ("إعدادات المشاركة الاجتماعية في منصتك أو وسوم og في <head>", "Social sharing settings in your platform, or og tags in <head>"),
        "schema": ("إعدادات SEO أو سكربت JSON-LD في <head>", "SEO settings or a JSON-LD script in <head>"),
        "sitemap": ("إعدادات خريطة الموقع في منصتك أو ملف sitemap.xml", "Sitemap settings in your platform, or sitemap.xml"),
        "viewport": ("كود القالب: وسم viewport في <head>", "Theme code: the viewport tag in <head>"),
        "lang": ("كود القالب: خاصية lang في وسم <html>", "Theme code: the lang attribute on the <html> tag"),
        "alt": ("إعدادات الصورة (النص البديل) في منصتك أو الوسم <img alt=''>", "The image's alt text setting in your platform, or the <img alt=''> attribute"),
        "slug": ("الرابط الدائم (Slug) للصفحة في منصتك", "The page slug / permalink in your platform"),
        "robots": ("ملف robots.txt في جذر الموقع أو إعدادات منصتك", "robots.txt in the site root, or your platform's settings"),
    },
}
PW["squarespace"] = PW["other"]
PW["webflow"] = PW["other"]
# بعض الحلول (وليس المكان فقط) تتغير حسب المنصة
PWF = {
    "blogger": {"HTTP_5XX": ("هذا غالبًا خطأ مؤقت من خوادم Blogger. أعد الفحص لاحقًا، وإن استمر فراجع حالة الخدمة.", "This is usually a temporary error on Blogger's servers. Re-run the audit later, and check the service status if it persists.")},
    "shopify": {"HTTP_5XX": ("راجع حالة Shopify وتطبيقات المتجر والقالب؛ وتواصل مع دعم Shopify إن استمر الخطأ.", "Check Shopify status, the store apps and the theme; contact Shopify support if it persists.")},
    "wix": {"HTTP_5XX": ("راجع حالة Wix وتواصل مع دعم Wix إن استمر الخطأ.", "Check Wix status and contact Wix support if the error persists.")},
}
PLATFORM_NAMES = {"blogger": "Blogger", "wordpress": "WordPress", "shopify": "Shopify", "wix": "Wix",
                  "squarespace": "Squarespace", "webflow": "Webflow"}


def platform_label():
    return PLATFORM_NAMES.get(PLATFORM) or tr("plat_unknown")


def detect_platform(html, url):
    h = html.lower()
    host = urlparse(url).netloc.lower()
    gen = ""
    try:
        g = BeautifulSoup(html, "html.parser").find("meta", attrs={"name": re.compile(r"^generator$", re.I)})
        gen = (g.get("content") or "").lower() if g else ""
    except Exception:
        pass
    if "blogger" in gen or "blogspot." in host:
        return "blogger"
    if "wordpress" in gen or "/wp-content/" in h or "/wp-includes/" in h:
        return "wordpress"
    if "shopify" in gen or "cdn.shopify.com" in h:
        return "shopify"
    if "wix.com" in gen or "wixstatic.com" in h:
        return "wix"
    if "squarespace" in h:
        return "squarespace"
    if "webflow" in gen or "webflow.com" in h:
        return "webflow"
    return "other"


def cat(code):
    sev, d, f, w = CATALOG[code]
    i = 0 if LANG == "ar" else 1
    fix, where = f[i], w[i]
    if PLATFORM != "wordpress":
        fx = PWF.get(PLATFORM, {}).get(code)
        if fx:
            fix = fx[i]
        grp = CODE_GROUP.get(code)
        pw = PW.get(PLATFORM, PW["other"]).get(grp) if grp else None
        if pw:
            where = pw[i]
    return sev, d[i], fix, where


# ------------------------------------------------------------------ الشبكة / Network
def make_session():
    s = requests.Session()
    s.headers.update(HEADERS)
    return s


def fetch(session, url, timeout, retries=2):
    last = None
    for i in range(retries + 1):
        try:
            return session.get(url, timeout=timeout, allow_redirects=True)
        except Exception as e:  # noqa
            last = e
            time.sleep(1 + i)
    raise last


# ------------------------------------------------------------------ Sitemap
def read_locs(content, url):
    if url.endswith(".gz") or content[:2] == b"\x1f\x8b":
        content = gzip.decompress(content)
    text = content.decode("utf-8", errors="ignore")
    is_index = "<sitemapindex" in text
    locs = [re.sub(r"<!\[CDATA\[|\]\]>", "", m).strip()
            for m in re.findall(r"<loc>\s*(.*?)\s*</loc>", text, re.S)]
    return is_index, locs


def expand_sitemap(session, url, timeout, depth=0, seen=None):
    seen = seen if seen is not None else set()
    if url in seen or depth > 3:
        return []
    seen.add(url)
    try:
        r = fetch(session, url, timeout, retries=1)
        if r.status_code != 200:
            return []
        is_index, locs = read_locs(r.content, url)
    except Exception:
        return []
    if not is_index:
        return locs
    urls = []
    for child in locs[:50]:
        urls += expand_sitemap(session, child, timeout, depth + 1, seen)
    return urls


def discover(session, base, timeout):
    robots_txt = ""
    candidates = []
    try:
        r = session.get(base + "/robots.txt", timeout=timeout)
        if r.status_code == 200 and "<html" not in r.text[:500].lower():
            robots_txt = r.text
            for line in robots_txt.splitlines():
                if line.lower().startswith("sitemap:"):
                    candidates.append(line.split(":", 1)[1].strip())
    except Exception:
        pass
    for p in ("/sitemap.xml", "/sitemap_index.xml", "/wp-sitemap.xml"):
        candidates.append(base + p)
    for c in dict.fromkeys(candidates):
        urls = expand_sitemap(session, c, timeout)
        if urls:
            return c, list(dict.fromkeys(urls)), robots_txt
    return None, [], robots_txt


# ------------------------------------------------------------------ فحص صفحة / Page analysis
def norm(u):
    u = u.split("#")[0].strip().lower()
    return u[:-1] if u.endswith("/") else u


ASSET_RE = re.compile(r"\.(jpe?g|png|gif|webp|svg|ico|pdf|zip|rar|css|js|mp4|mp3|docx?|xlsx?|pptx?)(\?.*)?$", re.I)


def looks_like_site(s):
    """True for 'https://x', 'http://x' or a bare domain like 'example.com' (not a file path)."""
    s = (s or "").strip()
    if s.lower().startswith(("http://", "https://")):
        return True
    return bool(re.match(r"^([a-z0-9-]+\.)+[a-z]{2,}(:\d+)?(/.*)?$", s, re.I)) and not s.lower().endswith(".json")


def same_host(a, b):
    ha = urlparse(a).netloc.lower().replace("www.", "")
    hb = urlparse(b).netloc.lower().replace("www.", "")
    return ha == hb


def check_link(session, url, timeout=10):
    """Returns the HTTP status of a link (or 'ERROR')."""
    try:
        r = session.head(url, timeout=timeout, allow_redirects=True)
        if r.status_code >= 400:
            r = session.get(url, timeout=timeout, allow_redirects=True, stream=True)
            r.close()
        return r.status_code
    except Exception:
        return "ERROR"


def is_broken(st):
    # 401/403/429 are usually bot protection, not a truly broken link
    return st == "ERROR" or st in (404, 410) or (isinstance(st, int) and 500 <= st < 600)


PSI_CAT = {
    "ar": {"FAST": "سريع", "AVERAGE": "متوسط", "SLOW": "بطيء"},
    "en": {"FAST": "Fast", "AVERAGE": "Average", "SLOW": "Slow"},
}


def parse_psi(data):
    """Extract [score, FCP s, LCP s, CLS, TBT ms, TTFB ms, real-user rating, real-user INP ms] from a PSI response."""
    lh = data["lighthouseResult"]
    a = lh["audits"]

    def num(key, scale=1, nd=None):
        try:
            v = a[key]["numericValue"] / scale
            return round(v, nd) if nd is not None else round(v)
        except Exception:
            return "n/a"

    score = round(lh["categories"]["performance"]["score"] * 100)
    field = data.get("loadingExperience") or {}
    cat = PSI_CAT[LANG].get(field.get("overall_category"), "n/a")
    inp = (field.get("metrics", {}).get("INTERACTION_TO_NEXT_PAINT") or {}).get("percentile", "n/a")
    return [score,
            num("first-contentful-paint", 1000, 2),
            num("largest-contentful-paint", 1000, 2),
            num("cumulative-layout-shift", 1, 3),
            num("total-blocking-time"),
            num("server-response-time"),
            cat, inp]


def pagespeed(session, url, key="", strategy="mobile", timeout=90):
    """Google PageSpeed Insights API (mobile or desktop). Returns (values, error)."""
    params = {"url": url, "strategy": strategy, "category": "performance"}
    if key:
        params["key"] = key
    try:
        r = session.get("https://www.googleapis.com/pagespeedonline/v5/runPagespeed",
                        params=params, timeout=timeout)
        if r.status_code != 200:
            return None, f"HTTP {r.status_code}"
        return parse_psi(r.json()), ""
    except Exception as e:  # noqa
        return None, str(e)[:80]


def analyze(url, session, timeout, delay, rp):
    issues = []
    row = {"URL": url}
    try:
        r = fetch(session, url, timeout)
    except Exception as e:  # noqa
        row["Status"] = "ERROR"
        issues.append(("FETCH_FAIL", str(e)[:150]))
        return row, issues
    time.sleep(delay)

    final = r.url
    status = r.status_code
    row.update({
        "Status": status,
        "Final URL": final,
        "Redirects": len(r.history),
        "Response (s)": round(r.elapsed.total_seconds(), 2),
        "Size (KB)": round(len(r.content) / 1024),
    })

    if rp is not None and not rp.can_fetch("*", url):
        issues.append(("ROBOTS_BLOCKED", tr("robots_dis")))
    if len(r.history) >= 2:
        issues.append(("REDIRECT_CHAIN", tr("chain", n=len(r.history), u=final)))
    elif len(r.history) == 1:
        issues.append(("REDIRECT", tr("redir", u=final)))
    if 400 <= status < 500:
        issues.append(("HTTP_4XX", tr("status_n", n=status)))
        return row, issues
    if status >= 500:
        issues.append(("HTTP_5XX", tr("status_n", n=status)))
        return row, issues

    if not final.startswith("https://"):
        issues.append(("NOT_HTTPS", final))
    if row["Response (s)"] > 1.5:
        issues.append(("SLOW", tr("sec", n=row["Response (s)"])))
    if len(r.content) > 1_500_000:
        issues.append(("BIG_PAGE", f"{row['Size (KB)']} KB"))

    ctype = r.headers.get("Content-Type", "")
    if "html" not in ctype.lower():
        return row, issues

    soup = BeautifulSoup(r.content, "html.parser")

    # URL
    path = urlparse(final)
    if len(final) > 115:
        issues.append(("URL_LONG", tr("chars", n=len(final))))
    if re.search(r"[A-Z]", path.path) or "_" in path.path or path.query:
        issues.append(("URL_STYLE", final))

    # Title
    t = soup.find("title")
    title = t.get_text(strip=True) if t else ""
    row["Title"] = title
    row["Title len"] = len(title)
    if not title:
        issues.append(("TITLE_MISSING", ""))
    elif len(title) < 30:
        issues.append(("TITLE_SHORT", f"{tr('chars', n=len(title))}: {title}"))
    elif len(title) > 60:
        issues.append(("TITLE_LONG", f"{tr('chars', n=len(title))}: {title}"))

    # Meta description
    m = soup.find("meta", attrs={"name": re.compile(r"^description$", re.I)})
    desc = (m.get("content") or "").strip() if m else ""
    row["Meta description"] = desc
    row["Desc len"] = len(desc)
    if not desc:
        issues.append(("DESC_MISSING", ""))
    elif len(desc) < 70:
        issues.append(("DESC_SHORT", tr("chars", n=len(desc))))
    elif len(desc) > 160:
        issues.append(("DESC_LONG", tr("chars", n=len(desc))))

    # H1
    h1s = [h.get_text(" ", strip=True) for h in soup.find_all("h1")]
    h1s = [h for h in h1s if h]
    row["H1"] = h1s[0] if h1s else ""
    row["H1 count"] = len(h1s)
    if not h1s:
        issues.append(("H1_MISSING", ""))
    elif len(h1s) > 1:
        issues.append(("H1_MULTI", f"{len(h1s)} H1"))

    # Canonical
    can = soup.find("link", rel=lambda v: v and "canonical" in
                    ([x.lower() for x in v] if isinstance(v, list) else [v.lower()]))
    canonical = (can.get("href") or "").strip() if can else ""
    row["Canonical"] = canonical
    if not canonical:
        issues.append(("CANON_MISSING", ""))
    elif norm(urljoin(final, canonical)) != norm(final):
        issues.append(("CANON_OTHER", f"canonical = {canonical}"))

    # Noindex
    rb = soup.find("meta", attrs={"name": re.compile(r"^robots$", re.I)})
    robots_meta = (rb.get("content") or "").lower() if rb else ""
    xrt = r.headers.get("X-Robots-Tag", "").lower()
    noindex = "noindex" in robots_meta or "noindex" in xrt
    row["Noindex"] = tr("yes") if noindex else tr("no")
    if noindex:
        issues.append(("NOINDEX", robots_meta or xrt))

    # Mobile / lang / social / schema
    if not soup.find("meta", attrs={"name": re.compile(r"^viewport$", re.I)}):
        issues.append(("NO_VIEWPORT", ""))
    if not (soup.html and soup.html.get("lang")):
        issues.append(("NO_LANG", ""))
    if not soup.find("meta", property="og:title"):
        issues.append(("NO_OG", ""))
    if not (soup.find("script", type="application/ld+json") or soup.find(attrs={"itemtype": True})):
        issues.append(("NO_SCHEMA", ""))

    # Images
    imgs = soup.find_all("img")
    no_alt = [i for i in imgs if i.get("alt") is None]
    row["Images"] = len(imgs)
    row["Images no alt"] = len(no_alt)
    if no_alt:
        issues.append(("IMG_ALT", tr("img", a=len(no_alt), b=len(imgs))))

    # Mixed content
    if final.startswith("https://"):
        bad = 0
        for tag in soup.find_all(["img", "script", "iframe", "source", "link"]):
            u = tag.get("src") or (tag.get("href") if tag.name == "link" else None)
            if u and u.startswith("http://"):
                bad += 1
        if bad:
            issues.append(("MIXED_CONTENT", tr("mixed", n=bad)))

    # Links (internal + external), kept for the crawl / broken-link phases
    links, seen_links = [], set()
    for a in soup.find_all("a", href=True):
        href = a["href"].strip()
        if not href or href.startswith(("#", "mailto:", "tel:", "javascript:", "sms:", "whatsapp:")):
            continue
        absu = urldefrag(urljoin(final, href))[0]
        if not absu.startswith(("http://", "https://")) or absu in seen_links:
            continue
        seen_links.add(absu)
        links.append((absu, same_host(absu, final), a.get_text(" ", strip=True)[:60]))
    row["_links"] = links
    internal = sum(1 for _, i, _t in links if i)
    row["Internal links"] = internal
    row["External links"] = len(links) - internal
    if internal == 0:
        issues.append(("NO_INTERNAL_LINKS", ""))

    # Heading order (e.g. H1 -> H3 without H2)
    levels = [int(h.name[1]) for h in soup.find_all(re.compile(r"^h[1-6]$"))]
    if any(b - a > 1 for a, b in zip(levels, levels[1:])):
        issues.append(("HEADING_SKIP", "H" + " > H".join(str(x) for x in levels[:12])))

    # Word count
    n_scripts = len(soup.find_all("script"))
    js_root = soup.find(id=re.compile(r"^(root|app|__next|__nuxt)$"))
    for tag in soup(["script", "style", "noscript"]):
        tag.decompose()
    words = len(soup.get_text(" ", strip=True).split())
    row["Words"] = words
    if words < 300:
        issues.append(("THIN", tr("words", n=words)))
    if words < 50 and (js_root is not None or n_scripts >= 8):
        issues.append(("JS_SUSPECT", tr("js_d")))

    return row, issues


# ------------------------------------------------------------------ التقييم والتقرير / Scoring & report
def robots_where():
    pw = PW.get(PLATFORM, PW["other"]).get("robots") if PLATFORM != "wordpress" else None
    return pw[0 if LANG == "ar" else 1] if pw else tr("no_robots_where")


def grade(score):
    if score >= 90:
        return tr("g90")
    if score >= 75:
        return tr("g75")
    if score >= 50:
        return tr("g50")
    return tr("g0")


def style_sheet(ws, widths, sev_col=None):
    from openpyxl.styles import Alignment, Font, PatternFill
    head = PatternFill("solid", fgColor="1F4E78")
    for c in ws[1]:
        c.font = Font(bold=True, color="FFFFFF")
        c.fill = head
        c.alignment = Alignment(vertical="center", wrap_text=True)
    for col, w in widths.items():
        ws.column_dimensions[col].width = w
    ws.freeze_panes = "A2"
    ws.auto_filter.ref = ws.dimensions
    ws.sheet_view.rightToLeft = S[LANG]["rtl"]
    raw = {"Critical": "F8CBAD", "High": "FFE699", "Medium": "FFF2CC", "Low": "E2EFDA", "Info": "DDEBF7"}
    colors = {SEVLBL[LANG][k]: v for k, v in raw.items()}
    for row in ws.iter_rows(min_row=2):
        for c in row:
            c.alignment = Alignment(vertical="top", wrap_text=True)
        if sev_col is not None:
            v = row[sev_col].value
            if v in colors:
                row[sev_col].fill = PatternFill("solid", fgColor=colors[v])


# ------------------------------------------------------------------ تقرير PDF ملوّن / Colourful PDF report
def _pip(pkg, mod):
    try:
        __import__(mod)
    except ImportError:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", pkg])


def build_pdf(path, base, site_score, n_pages, n_broken, sev_count, top_list, speed_rows, n_issues):
    _pip("reportlab", "reportlab")
    ar = LANG == "ar"
    if ar:
        _pip("arabic-reshaper", "arabic_reshaper")
        _pip("python-bidi", "bidi")
        import arabic_reshaper
        try:
            from bidi import get_display
        except ImportError:
            from bidi.algorithm import get_display
    from reportlab.lib.colors import HexColor, white
    from reportlab.lib.pagesizes import A4
    from reportlab.pdfbase import pdfmetrics
    from reportlab.pdfbase.ttfonts import TTFont
    from reportlab.pdfgen import canvas

    FONT, FONT_B = "Helvetica", "Helvetica-Bold"
    if ar:
        regs = ["/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf", "/usr/share/fonts/dejavu/DejaVuSans.ttf"]
        bolds = ["/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf", "/usr/share/fonts/dejavu/DejaVuSans-Bold.ttf"]
        reg = next((x for x in regs if os.path.exists(x)), None)
        bold = next((x for x in bolds if os.path.exists(x)), reg)
        if not reg:
            print(tr("pdf_nofont"))
            return False
        pdfmetrics.registerFont(TTFont("AR", reg))
        pdfmetrics.registerFont(TTFont("AR-B", bold))
        FONT, FONT_B = "AR", "AR-B"

        def shape(t):
            return get_display(arabic_reshaper.reshape(str(t)))
    else:
        def shape(t):
            return str(t)

    NAVY, TEAL, TXT, GREY, SOFT = HexColor("#1F3A6E"), HexColor("#16A2B8"), HexColor("#222222"), HexColor("#666666"), HexColor("#EAF1FB")
    SEV_COL = {"Critical": HexColor("#D9534F"), "High": HexColor("#F0883E"), "Medium": HexColor("#F2C94C"),
               "Low": HexColor("#2E9E5B"), "Info": HexColor("#5B9BD5")}
    lbl_to_sev = {v: k for k, v in SEVLBL[LANG].items()}
    W, H = A4
    M = 42
    SX = W - M if ar else M

    def put(c, t, y, font, size, color, x=None):
        c.setFont(font, size)
        c.setFillColor(color)
        (c.drawRightString if ar else c.drawString)(SX if x is None else x, y, shape(t))

    def wrap(t, font, size, maxw):
        lines, cur = [], ""
        for w in str(t).split(" "):
            trial = (cur + " " + w).strip()
            if not cur or pdfmetrics.stringWidth(shape(trial), font, size) <= maxw:
                cur = trial
            else:
                lines.append(cur)
                cur = w
        if cur:
            lines.append(cur)
        return lines

    c = canvas.Canvas(path, pagesize=A4)
    c.setTitle(f"{tr('pdf_title')} - {base}")
    st = {"y": H, "n": 1}

    def footer():
        c.setFont(FONT, 8)
        c.setFillColor(GREY)
        c.drawCentredString(W / 2, 22, shape(f"{tr('pdf_title')}  |  {st['n']}"))

    def new_page():
        footer()
        c.showPage()
        st["n"] += 1
        st["y"] = H - 50

    def ensure(h):
        if st["y"] - h < 50:
            new_page()

    # ---- header
    c.setFillColor(NAVY)
    c.rect(0, H - 135, W, 135, fill=1, stroke=0)
    c.setFillColor(TEAL)
    c.rect(0, H - 141, W, 6, fill=1, stroke=0)
    put(c, tr("pdf_title"), H - 56, FONT_B, 25, white)
    put(c, base, H - 82, FONT, 12, HexColor("#CFE3F3"))
    meta = (f"{tr('pdf_date')}: {datetime.date.today().isoformat()}   |   {tr('pdf_platform')}: {platform_label()}"
            f"   |   {tr('pdf_prepared')}: {AUTHOR_NAME}")
    put(c, meta, H - 112, FONT, 9, HexColor("#CFE3F3"))

    # ---- score circle
    col = (SEV_COL["Low"] if site_score >= 90 else HexColor("#7CB342") if site_score >= 75
           else SEV_COL["High"] if site_score >= 50 else SEV_COL["Critical"])
    cx, cy, r = W / 2, H - 215, 54
    c.setFillColor(SOFT)
    c.circle(cx, cy, r + 9, fill=1, stroke=0)
    c.setFillColor(col)
    c.circle(cx, cy, r, fill=1, stroke=0)
    c.setFillColor(white)
    c.setFont("Helvetica-Bold", 36)
    c.drawCentredString(cx, cy - 4, f"{site_score:g}")
    c.setFont("Helvetica", 11)
    c.drawCentredString(cx, cy - 22, "/ 100")
    verdict = grade(site_score)
    c.setFillColor(col)
    c.setFont(FONT_B, 13)
    c.drawCentredString(cx, cy - r - 26, shape(verdict))

    # ---- 4 stat cards
    cards = [(tr("pdf_pages"), n_pages, NAVY), (tr("pdf_broken"), n_broken, SEV_COL["High"]),
             (tr("pdf_crit"), sev_count.get("Critical", 0), SEV_COL["Critical"]),
             (tr("pdf_high"), sev_count.get("High", 0), SEV_COL["High"])]
    gap = 10
    bw = (W - 2 * M - 3 * gap) / 4
    by = cy - r - 112
    for i, (lab, val, ccol) in enumerate(cards):
        bx = M + (3 - i if ar else i) * (bw + gap)
        c.setFillColor(SOFT)
        c.roundRect(bx, by, bw, 58, 7, fill=1, stroke=0)
        c.setFillColor(ccol)
        c.roundRect(bx, by, 5, 58, 2, fill=1, stroke=0)
        c.setFont("Helvetica-Bold", 20)
        c.drawCentredString(bx + bw / 2, by + 30, str(val))
        c.setFillColor(GREY)
        c.setFont(FONT, 8.5)
        c.drawCentredString(bx + bw / 2, by + 12, shape(lab))
    st["y"] = by - 28

    # ---- severity bars
    put(c, tr("pdf_sev"), st["y"], FONT_B, 13, NAVY)
    st["y"] -= 20
    mx = max([sev_count.get(k, 0) for k in ("Critical", "High", "Medium", "Low")] + [1])
    barw = W - 2 * M - 120
    for k in ("Critical", "High", "Medium", "Low"):
        n = sev_count.get(k, 0)
        put(c, SEVLBL[LANG][k], st["y"], FONT, 10, TXT)
        x0 = (W - M - 70 - barw) if ar else (M + 70)
        yb = st["y"] - 2
        c.setFillColor(SOFT)
        c.roundRect(x0, yb, barw, 10, 5, fill=1, stroke=0)
        fw = max(barw * n / mx, 2) if n else 0
        if fw:
            c.setFillColor(SEV_COL[k])
            c.roundRect((x0 + barw - fw) if ar else x0, yb, fw, 10, 5, fill=1, stroke=0)
        c.setFillColor(TXT)
        c.setFont("Helvetica-Bold", 10)
        (c.drawString if ar else c.drawRightString)(M if ar else W - M, yb + 1, str(n))
        st["y"] -= 22

    # ---- PageSpeed (mobile) card, only when available
    mob = next((r for r in speed_rows if len(r) > 4 and isinstance(r[2], (int, float))), None)
    if mob:
        st["y"] -= 8
        ensure(70)
        put(c, tr("pdf_speed"), st["y"], FONT_B, 13, NAVY)
        st["y"] -= 18
        sc = mob[2]
        scol = SEV_COL["Low"] if sc >= 90 else SEV_COL["Medium"] if sc >= 50 else SEV_COL["Critical"]
        c.setFillColor(scol)
        c.roundRect(M, st["y"] - 28, W - 2 * M, 36, 7, fill=1, stroke=0)
        c.setFillColor(white)
        c.setFont("Helvetica-Bold", 12)
        c.drawCentredString(W / 2, st["y"] - 14, f"{sc}/100   |   LCP {mob[4]}s   |   CLS {mob[5]}   |   TBT {mob[6]}ms")
        st["y"] -= 46

    # ---- executive summary
    st["y"] -= 8
    ensure(120)
    put(c, tr("pdf_exec"), st["y"], FONT_B, 13, NAVY)
    st["y"] -= 18
    if top_list:
        sep = "، " if ar else ", "
        top = sep.join(x[1] for x in top_list[:3])
        text = tr("exec_tpl", pages=n_pages, site=base, platform=platform_label(), score=f"{site_score:g}", grade=verdict,
                  n_issues=n_issues, n_crit=sev_count.get("Critical", 0), n_high=sev_count.get("High", 0), top=top)
    else:
        text = tr("pdf_none")
    for ln in wrap(text, FONT, 10, W - 2 * M):
        ensure(16)
        put(c, ln, st["y"], FONT, 10, TXT)
        st["y"] -= 15

    # ---- top fixes
    st["y"] -= 10
    ensure(70)
    put(c, tr("pdf_top"), st["y"], FONT_B, 13, NAVY)
    st["y"] -= 20
    for sev_l, desc, cnt, pct, fix, where in top_list[:8]:
        sk = lbl_to_sev.get(sev_l, "Low")
        lines = [(tr("pdf_how") + ": " + fix, TXT), (tr("pdf_where") + ": " + where, GREY)]
        body = []
        for t, colr in lines:
            body += [(ln, colr) for ln in wrap(t, FONT, 8.5, W - 2 * M - 18)][:4]
        ensure(34 + 11.5 * len(body))
        dx = (W - M - 4) if ar else (M + 4)
        c.setFillColor(SEV_COL[sk])
        c.circle(dx, st["y"] + 3, 4.5, fill=1, stroke=0)
        tx = (W - M - 16) if ar else (M + 16)
        for i, ln in enumerate(wrap(f"{desc}  ({cnt} {tr('pdf_pg')})", FONT_B, 10, W - 2 * M - 18)[:2]):
            put(c, ln, st["y"], FONT_B, 10, TXT, x=tx)
            st["y"] -= 13
        for ln, colr in body:
            put(c, ln, st["y"], FONT, 8.5, colr, x=tx)
            st["y"] -= 11.5
        st["y"] -= 9

    # ---- notes
    notes = [tr("note"), tr("menu_note")] + ([tr("plat_note")] if PLATFORM in ("other", "squarespace", "webflow") else [])
    st["y"] -= 2
    for nt in notes:
        for ln in wrap(nt, FONT, 8, W - 2 * M):
            ensure(12)
            put(c, ln, st["y"], FONT, 8, GREY)
            st["y"] -= 11
    footer()
    c.save()
    return True


def main():
    global LANG, PLATFORM
    ap = argparse.ArgumentParser(description="SEO Audit v4 (Technical + On-Page)")
    ap.add_argument("url", nargs="?", help="website URL")
    ap.add_argument("--lang", choices=["ar", "en"], default=REPORT_LANG, help="report language: ar (Arabic) or en (English)")
    ap.add_argument("--max", type=int, default=100, help="number of sitemap pages to audit (0 = all)")
    ap.add_argument("--crawl", type=int, default=50,
                    help="extra pages to discover through internal links (0 = off)")
    ap.add_argument("--links-max", type=int, default=300, help="max links to check for broken ones (0 = off)")
    ap.add_argument("--psi", type=int, default=1, help="pages to test with Google PageSpeed (default 1 = homepage, 0 = off)")
    ap.add_argument("--psi-key", default=PSI_API_KEY, help="optional Google PageSpeed API key")
    ap.add_argument("--workers", type=int, default=5)
    ap.add_argument("--delay", type=float, default=0.3, help="delay between requests (seconds)")
    ap.add_argument("--timeout", type=int, default=15)
    ap.add_argument("--out", default="", help="Excel file name (default: seo_audit_<domain>.xlsx)")
    args, _ = ap.parse_known_args()  # works in Colab/Jupyter too
    LANG = args.lang if args.lang in ("ar", "en") else "ar"

    # In Colab/Jupyter, sys.argv contains "-f /path/kernel-xxx.json", which argparse may mistake
    # for the URL. Only accept an argument that really looks like a website.
    arg_url = args.url if looks_like_site(args.url) else ""
    url = (arg_url or SITE_URL or input(tr("p_enter"))).strip()
    if not url.startswith("http"):
        url = "https://" + url
    p = urlparse(url)
    if not p.netloc:
        print("❌ Invalid website URL / رابط الموقع غير صالح:", repr(url))
        sys.exit(1)
    base = f"{p.scheme}://{p.netloc}"
    if not args.out:  # one file per website, so reports never overwrite each other
        args.out = "seo_audit_" + re.sub(r"[^a-z0-9]+", "_", p.netloc.lower()).strip("_") + ".xlsx"

    session = make_session()
    try:  # detect the site platform from the homepage
        hp = fetch(session, base, args.timeout, retries=1)
        PLATFORM = detect_platform(hp.text, hp.url)
    except Exception:
        PLATFORM = "other"
    print(tr("p_platform", p=platform_label()))
    print(tr("p_search", b=base))
    sm_url, urls, robots_txt = discover(session, base, args.timeout)
    if not urls:
        print(tr("p_nosm"))
        sys.exit(1)
    total = len(urls)
    print(tr("p_found", sm=sm_url, n=total))
    sitemap_all = {norm(u) for u in urls}

    sampled = False
    if args.max and total > args.max:
        step = total / args.max
        urls = [urls[int(i * step)] for i in range(args.max)]  # evenly spread sample
        sampled = True
        print(tr("p_sample", m=args.max))
    sitemap_pages = set(urls)

    rp = None
    if robots_txt:
        rp = RobotFileParser()
        rp.parse(robots_txt.splitlines())

    def run(url_list):
        out = []
        with ThreadPoolExecutor(max_workers=args.workers) as ex:
            futs = {ex.submit(analyze, u, session, args.timeout, args.delay, rp): u for u in url_list}
            for n, f in enumerate(as_completed(futs), 1):
                try:
                    out.append(f.result())
                except Exception as e:  # noqa
                    out.append(({"URL": futs[f], "Status": "ERROR"}, [("FETCH_FAIL", str(e)[:150])]))
                if n % 10 == 0 or n == len(futs):
                    print(tr("p_prog", n=n, m=len(futs)))
        return out

    def known(res):
        s = set()
        for row, _ in res:
            s.add(norm(row["URL"]))
            if row.get("Final URL"):
                s.add(norm(row["Final URL"]))
        return s

    # ---- Phase 1: sitemap pages
    results = run(urls)
    analyzed = known(results)

    # ---- Phase 2: pages discovered through internal links (not in the sitemap)
    crawl_added = 0
    crawl_truncated = False
    if args.crawl:
        cand = {}
        for row, _ in results:
            for lnk, internal, _t in row.get("_links", []):
                n = norm(lnk)
                if (internal and n not in analyzed and n not in sitemap_all and n not in cand
                        and not ASSET_RE.search(lnk) and (rp is None or rp.can_fetch("*", lnk))):
                    cand[n] = lnk
        cand_urls = list(cand.values())
        crawl_truncated = len(cand_urls) > args.crawl
        cand_urls = cand_urls[:args.crawl]
        if cand_urls:
            print(tr("p_crawl", n=len(cand_urls)))
            res2 = run(cand_urls)
            for row, issues in res2:
                if row.get("Status") == 200 and "Title" in row and not row.get("Redirects"):
                    issues.append(("NOT_IN_SITEMAP", ""))
            results += res2
            analyzed |= known(res2)
            crawl_added = len(res2)

    # ---- Phase 3: broken links
    status_map = {}
    for row, _ in results:
        for key in (row["URL"], row.get("Final URL")):
            if key:
                status_map[norm(key)] = row.get("Status")
    targets = {}
    for row, _ in results:
        for lnk, internal, _t in row.get("_links", []):
            n = norm(lnk)
            if n not in status_map and n not in targets:
                targets[n] = (lnk, internal)
    ordered = sorted(targets.values(), key=lambda x: not x[1])  # internal links first
    ordered = ordered[:args.links_max] if args.links_max else []
    checked = {}
    if ordered:
        print(tr("p_links", n=len(ordered)))
        with ThreadPoolExecutor(max_workers=8) as ex:
            futs = {ex.submit(check_link, session, lnk): norm(lnk) for lnk, _i in ordered}
            for f in as_completed(futs):
                checked[futs[f]] = f.result()
    status_map.update(checked)

    broken_rows = []
    for row, issues in results:
        n_int = n_ext = 0
        for lnk, internal, text in row.get("_links", []):
            st = status_map.get(norm(lnk))
            if st is not None and is_broken(st):
                broken_rows.append([row["URL"], lnk, st, tr("int") if internal else tr("ext"), text])
                if internal:
                    n_int += 1
                else:
                    n_ext += 1
        if n_int:
            issues.append(("BROKEN_INTERNAL", tr("broken_n", n=n_int)))
        if n_ext:
            issues.append(("BROKEN_EXTERNAL", tr("broken_n", n=n_ext)))

    # ---- Phase 4: orphan pages (only when every sitemap page was audited and the crawl was complete)
    orphan_note = ""
    if sampled or crawl_truncated or not args.crawl:
        orphan_note = tr("orphan_skip")
    else:
        inbound = set()
        for row, _ in results:
            me = {norm(row["URL"]), norm(row.get("Final URL") or row["URL"])}
            for lnk, internal, _t in row.get("_links", []):
                n = norm(lnk)
                if internal and n not in me:
                    inbound.add(n)
        home = {norm(base)}
        for row, issues in results:
            if row["URL"] in sitemap_pages and row.get("Status") == 200 and "Title" in row:
                keys = {norm(row["URL"]), norm(row.get("Final URL") or row["URL"])}
                if not (keys & inbound) and not (keys & home):
                    issues.append(("ORPHAN", ""))

    # ---- Phase 5: Google PageSpeed (optional)
    speed_rows = []
    if args.psi:
        print(tr("p_psi", n=args.psi))
        idx = {r[0]["URL"]: r for r in results}
        psi_targets = [r[0]["URL"] for r in results if r[0].get("Status") == 200 and "Title" in r[0]]
        psi_targets.sort(key=lambda u: norm(u) != norm(base))  # homepage first
        for u in psi_targets[:args.psi]:
            for dev in ("mobile", "desktop"):
                label = tr("dev_" + dev)
                res, err = pagespeed(session, u, args.psi_key, dev)
                if res is None:
                    speed_rows.append([u, label] + ["n/a"] * 8)
                    print(tr("p_psi_err", d=label, u=u, e=err))
                else:
                    speed_rows.append([u, label, *res])
                    if dev == "mobile":  # score deductions use the mobile result
                        if res[0] < 50:
                            idx[u][1].append(("PSI_POOR", tr("psi_d", s=res[0])))
                        elif res[0] < 90:
                            idx[u][1].append(("PSI_AVG", tr("psi_d", s=res[0])))
                time.sleep(1)

    # ---- Duplicate Title / Description / H1
    def dups(key):
        c = Counter(r[0].get(key) for r in results if r[0].get(key))
        return {k for k, v in c.items() if v > 1}

    d_title, d_desc, d_h1 = dups("Title"), dups("Meta description"), dups("H1")
    for row, issues in results:
        if row.get("Title") in d_title:
            issues.append(("TITLE_DUP", row["Title"]))
        if row.get("Meta description") in d_desc:
            issues.append(("DESC_DUP", row["Meta description"][:100]))
        if row.get("H1") in d_h1:
            issues.append(("H1_DUP", row["H1"]))

    # ---- Scoring
    issue_rows, page_rows = [], []
    for row, issues in results:
        row.pop("_links", None)
        codes = {}
        for code, detail in issues:
            codes.setdefault(code, detail)
        penalty = sum(WEIGHTS[CATALOG[c][0]] for c in codes)
        row["Score"] = max(0, 100 - penalty)
        row["Issues"] = len(codes)
        page_rows.append(row)
        for code, detail in codes.items():
            sev, desc, fix, where = cat(code)
            issue_rows.append((SEV_ORDER[sev], code,
                               [row["URL"], SEVLBL[LANG][sev], desc, detail, fix, where, code]))
    issue_rows.sort(key=lambda x: (x[0], x[1]))
    issue_list = [x[2] for x in issue_rows]

    site_penalty = 0
    site_issue_rows = []
    if not robots_txt:
        site_penalty += 5
        site_issue_rows.append((tr("site_issue"),
                                f"{tr('no_robots')} -> {tr('no_robots_fix')} ({robots_where()})"))
    avg = sum(r["Score"] for r in page_rows) / len(page_rows)
    site_score = max(0, round(avg - site_penalty, 1))

    # Aggregated top issues
    agg = Counter(x[1] for x in issue_rows)
    top_rows = []
    for code, cnt in agg.items():
        sev, desc, fix, where = cat(code)
        top_rows.append((SEV_ORDER[sev], -cnt,
                         [SEVLBL[LANG][sev], desc, cnt, round(cnt * 100 / len(page_rows), 1), fix, where]))
    top_rows.sort(key=lambda x: (x[0], x[1]))
    top_list = [x[2] for x in top_rows]
    page_rows.sort(key=lambda x: x["Score"])

    sev_count = Counter(CATALOG[x[1]][0] for x in issue_rows)
    summary = [
        (tr("site"), base),
        (tr("platform"), platform_label()),
        (tr("sitemap"), sm_url),
        (tr("sm_total"), total),
        (tr("checked"), len(page_rows)),
        (tr("crawl_found"), crawl_added),
        (tr("links_checked"), len(checked)),
        (tr("broken_total"), len(broken_rows)),
        (tr("score"), site_score),
        (tr("verdict"), grade(site_score)),
    ]
    for sev in ("Critical", "High", "Medium", "Low"):
        summary.append((tr("issues_of", s=SEVLBL[LANG][sev]), sev_count.get(sev, 0)))
    summary.append((tr("note_l"), tr("note")))
    summary.append((tr("note_l"), tr("menu_note")))
    if PLATFORM in ("other", "squarespace", "webflow"):
        summary.append((tr("note_l"), tr("plat_note")))
    if orphan_note:
        summary.append((tr("note_l"), orphan_note))
    summary += site_issue_rows

    sh = S[LANG]["sheets"]
    with pd.ExcelWriter(args.out, engine="openpyxl") as w:
        pd.DataFrame(summary, columns=S[LANG]["sum_cols"]).to_excel(w, sheet_name=sh[0], index=False)
        pd.DataFrame(top_list, columns=S[LANG]["top_cols"]).to_excel(w, sheet_name=sh[1], index=False)
        pd.DataFrame(issue_list, columns=S[LANG]["iss_cols"]).to_excel(w, sheet_name=sh[2], index=False)
        pd.DataFrame(page_rows).rename(columns=S[LANG]["page_cols"]).to_excel(w, sheet_name=sh[3], index=False)
        pd.DataFrame(broken_rows, columns=S[LANG]["bl_cols"]).to_excel(w, sheet_name=sh[4], index=False)
        if speed_rows:
            pd.DataFrame(speed_rows, columns=S[LANG]["speed_cols"]).to_excel(w, sheet_name=sh[5], index=False)
        style_sheet(w.sheets[sh[0]], {"A": 36, "B": 90})
        style_sheet(w.sheets[sh[1]], {"A": 12, "B": 48, "C": 14, "D": 12, "E": 60, "F": 55}, sev_col=0)
        style_sheet(w.sheets[sh[2]], {"A": 50, "B": 12, "C": 45, "D": 40, "E": 55, "F": 50, "G": 16}, sev_col=1)
        style_sheet(w.sheets[sh[3]], {"A": 50, "B": 9, "C": 40})
        style_sheet(w.sheets[sh[4]], {"A": 50, "B": 50, "C": 10, "D": 12, "E": 30})
        if speed_rows:
            style_sheet(w.sheets[sh[5]], {"A": 50, "B": 12, "C": 20, "D": 12, "E": 12, "F": 10, "G": 16, "H": 16, "I": 28, "J": 22})

    print("\n" + "=" * 50)
    print(tr("p_score", s=site_score, g=grade(site_score)))
    print("   " + " | ".join(f"{SEVLBL[LANG][s]}: {sev_count.get(s, 0)}"
                              for s in ("Critical", "High", "Medium", "Low")))
    print(f"   {tr('broken_total')}: {len(broken_rows)}")
    print(tr("p_top"))
    for x in top_list[:5]:
        print(tr("p_top_row", sev=x[0], issue=x[1], n=x[2]))
    print(tr("p_file", f=args.out))

    pdf_path = re.sub(r"\.xlsx$", "", args.out) + ".pdf"
    pdf_ok = False
    try:
        pdf_ok = build_pdf(pdf_path, base, site_score, len(page_rows), len(broken_rows), sev_count,
                           top_list, speed_rows, len(issue_list))
        if pdf_ok:
            print(tr("pdf_file", f=pdf_path))
    except Exception as e:  # the Excel report is already saved; never lose it because of the PDF
        print("PDF error:", e)

    try:  # In Google Colab the files download automatically
        from google.colab import files  # type: ignore
        files.download(args.out)
        if pdf_ok:
            files.download(pdf_path)
    except Exception:
        pass


if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
SEO Audit v5 (Technical + On-Page) from sitemap.xml  |  فحص SEO تقني وفحص صفحات من خريطة الموقع

- Reads robots.txt and sitemap.xml (including sitemap index files)
- Checks pages one by one, discovers extra pages via internal links, finds broken links and orphan pages
- Optional Google PageSpeed test (--psi N) and a warning for JavaScript-rendered pages
- Gives a score out of 100 (higher = better)
- Exports an Excel report: issues, how to fix them, and where to fix them
- NEW: detects the site platform (Blogger / WordPress / Shopify / Wix / other) and gives platform-specific "where to fix" tips
- NEW: also exports a colourful professional PDF summary report
- v5: the score can no longer say "Good" when serious problems exist (critical issues, poor mobile speed / Core Web Vitals);
       the report explains why, flags repeated root causes (one broken menu link = many broken links), and re-checks doubtful links
- Report language: Modern Standard Arabic (--lang ar) or English (--lang en)

Install:
    pip install requests beautifulsoup4 pandas openpyxl

Usage:
    python seo_audit_v3.py https://example.com
    python seo_audit_v3.py https://example.com --lang en --max 200 --out report.xlsx
    python seo_audit_v3.py https://example.com --max 0 --crawl 200   (all sitemap pages + 200 discovered pages)
    python seo_audit_v3.py https://example.com --psi 5                (Google PageSpeed on 5 pages; default 1 = homepage; 0 = off)
"""
import argparse
import datetime
import gzip
import os
import re
import subprocess
import sys
import time
from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from urllib.parse import urldefrag, urljoin, urlparse
from urllib.robotparser import RobotFileParser

import pandas as pd
import requests
from bs4 import BeautifulSoup

# ======================================================================
# 👇👇👇  هنا كتبدل رابط الموقع اللي بغيتي تفحصو (الموقع الحالي: odysseedelaterre.fr)
#         CHANGE THE WEBSITE TO AUDIT HERE (the only line you need to edit)
# ======================================================================
SITE_URL = "https://odysseedelaterre.fr"

# 👇 لغة التقرير: "en" = عربية فصحى  |  "en" = English
#    REPORT LANGUAGE: "en" = Arabic  |  "en" = English
REPORT_LANG = "en"

# 👇 (اختياري) مفتاح Google PageSpeed. خليه فارغ "" إلا ما عندكش
PSI_API_KEY = "AIzaSyD0XbcuK74WPFo0umfMJTEHas5ppSDm3nk"

# 👇 الاسم الذي يظهر في تقرير PDF (غيّره لاسمك أو اسم وكالتك)
#    Name shown on the PDF report
AUTHOR_NAME = "Youssef.O"
# ======================================================================
# (إلا كتبتي رابط فالأمر %run هو اللي كيتطبق، وإلا ماكتبتيش كيتطبق SITE_URL)

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/124.0 Safari/537.36 SEOAuditBot/1.0"
}

# Points deducted from a page's score for each issue, by severity
WEIGHTS = {"Critical": 25, "High": 12, "Medium": 5, "Low": 2, "Info": 0}
SEV_ORDER = {"Critical": 0, "High": 1, "Medium": 2, "Low": 3, "Info": 4}
SEVLBL = {
    "ar": {"Critical": "حرج", "High": "عالٍ", "Medium": "متوسط", "Low": "منخفض", "Info": "ملاحظة"},
    "en": {"Critical": "Critical", "High": "High", "Medium": "Medium", "Low": "Low", "Info": "Info"},
}

LANG = "ar"
PLATFORM = "other"   # blogger / wordpress / shopify / wix / squarespace / webflow / other (detected in main)

# ------------------------------------------------------------------ النصوص / UI strings
S = {
    "ar": {
        "chars": "{n} حرفًا", "words": "{n} كلمة", "sec": "{n} ثانية",
        "robots_dis": "ممنوعة بقاعدة Disallow في robots.txt",
        "chain": "{n} عمليات إعادة توجيه ← {u}",
        "redir": "يعيد التوجيه إلى {u}",
        "img": "{a} من أصل {b} صورة بلا نص بديل",
        "mixed": "{n} ملفًا يُحمَّل عبر http",
        "yes": "نعم", "no": "لا",
        "g90": "ممتاز", "g75": "جيد", "g50": "متوسط (يحتاج إلى تحسينات)", "g0": "ضعيف (مشكلات كبيرة)",
        "sheets": ["الملخص", "أهم الإصلاحات", "كل المشكلات", "الصفحات", "الروابط المكسورة", "السرعة"],
        "status_n": "رمز الاستجابة {n}",
        "page_cols": {"URL": "الرابط", "Status": "الحالة", "Final URL": "الرابط النهائي", "Redirects": "عدد إعادات التوجيه",
                      "Response (s)": "زمن الاستجابة (ثانية)", "Size (KB)": "الحجم (كيلوبايت)", "Title": "عنوان الصفحة",
                      "Title len": "طول العنوان", "Meta description": "وصف الميتا", "Desc len": "طول الوصف",
                      "H1": "العنوان الرئيسي H1", "H1 count": "عدد H1", "Canonical": "Canonical", "Noindex": "Noindex",
                      "Images": "عدد الصور", "Images no alt": "صور بلا نص بديل", "Internal links": "روابط داخلية",
                      "External links": "روابط خارجية", "Words": "عدد الكلمات", "Score": "التقييم (من 100)",
                      "Issues": "عدد المشكلات"},
        "bl_cols": ["الصفحة المصدر", "الرابط المكسور", "الحالة", "النوع", "نص الرابط"],
        "int": "داخلي", "ext": "خارجي",
        "speed_cols": ["الرابط", "الجهاز", "الأداء (من 100)", "FCP (ثانية)", "LCP (ثانية)", "CLS", "TBT (مللي ثانية)",
                       "TTFB (مللي ثانية)", "تقييم المستخدمين الحقيقيين (28 يومًا)", "INP للمستخدمين (مللي ثانية)"],
        "broken_n": "{n} رابط مكسور",
        "js_d": "نص قليل مع عناصر JavaScript كثيرة",
        "psi_d": "الأداء {s}/100",
        "crawl_found": "صفحات أُضيفت للفحص عبر الروابط الداخلية (خارج الخريطة)",
        "links_checked": "عدد الروابط التي جرى فحصها",
        "broken_total": "عدد الروابط المكسورة",
        "orphan_skip": "تم تخطي كشف الصفحات اليتيمة لأن الفحص لم يشمل كل الصفحات. استخدم --max 0 مع --crawl أكبر.",
        "p_crawl": "🕸️ تم اكتشاف {n} صفحة إضافية عبر الروابط الداخلية، جارٍ فحصها ...",
        "p_links": "🔗 جارٍ فحص {n} رابط بحثًا عن الروابط المكسورة ...",
        "p_psi": "⚡ جارٍ فحص السرعة عبر Google PageSpeed ({n} صفحة، على الجوال والكمبيوتر، قد يستغرق عدة دقائق) ...",
        "dev_mobile": "جوال", "dev_desktop": "كمبيوتر",
        "p_psi_err": "⚠️ تعذّر فحص PageSpeed ({d}) للرابط {u} ({e}). إن كان الخطأ 429 فالحصة المجانية استُنفدت: جرّب لاحقًا أو استعمل --psi-key، أو افحص الرابط يدويًا في pagespeed.web.dev",
        "sum_cols": ["البند", "القيمة"],
        "top_cols": ["الخطورة", "المشكلة", "عدد الصفحات المتأثرة", "% من الصفحات", "طريقة الإصلاح", "مكان الإصلاح"],
        "iss_cols": ["الرابط", "الخطورة", "المشكلة", "التفاصيل", "طريقة الإصلاح", "مكان الإصلاح", "الرمز"],
        "site": "الموقع", "sitemap": "خريطة الموقع (Sitemap)", "sm_total": "عدد الروابط في الخريطة",
        "checked": "عدد الصفحات المفحوصة", "score": "التقييم العام (من 100)", "verdict": "الحكم",
        "issues_of": "عدد المشكلات ({s})",
        "note_l": "ملاحظة",
        "note": "التقييم من 100: كلما ارتفع الرقم كان الموقع أفضل. هذا تقييم تقديري من هذه الأداة وليس مقياسًا رسميًا من Google.",
        "site_issue": "مشكلة على مستوى الموقع",
        "no_robots": "لا يوجد ملف robots.txt",
        "no_robots_fix": "أنشئ ملف robots.txt في جذر الموقع وأضف إليه رابط خريطة الموقع.",
        "no_robots_where": "جذر الموقع: /robots.txt",
        "p_enter": "🔗 أدخل رابط الموقع (مثال: https://example.com): ",
        "p_search": "⏳ جارٍ البحث عن خريطة الموقع لـ {b} ...",
        "p_nosm": "❌ لم يتم العثور على خريطة موقع (robots.txt أو /sitemap.xml أو /sitemap_index.xml أو /wp-sitemap.xml).",
        "p_found": "✅ خريطة الموقع: {sm} | {n} رابط",
        "p_sample": "ℹ️ سيتم فحص {m} صفحة (عينة موزعة على الموقع). استخدم --max 0 لفحص الكل.",
        "p_prog": "   تم فحص {n}/{m} صفحة",
        "p_score": "🏁 التقييم العام: {s}/100 ← {g}",
        "p_top": "🔧 أهم المشكلات التي يجب إصلاحها:",
        "p_top_row": "   [{sev}] {issue} ({n} صفحة)",
        "p_file": "📄 التقرير الكامل: {f}",
        "platform": "منصة الموقع", "plat_unknown": "غير محدد",
        "plat_note": "لم يتم التعرف على منصة الموقع، لذلك مكان الإصلاح عام. نفّذ الحلول من لوحة تحكم منصتك أو اطلب ذلك من المطوّر.",
        "menu_note": "أسماء القوائم والأزرار قد تختلف قليلًا حسب تحديثات المنصة.",
        "p_platform": "🧭 منصة الموقع: {p}",
        "pdf_title": "تقرير فحص SEO", "pdf_prepared": "إعداد", "pdf_date": "التاريخ", "pdf_platform": "المنصة",
        "pdf_pages": "الصفحات المفحوصة", "pdf_broken": "روابط مكسورة", "pdf_crit": "مشكلات حرجة", "pdf_high": "مشكلات عالية",
        "pdf_sev": "توزيع المشكلات حسب الخطورة", "pdf_speed": "سرعة الجوال (Google PageSpeed)",
        "pdf_exec": "الملخص التنفيذي", "pdf_top": "أهم الإصلاحات", "pdf_pg": "صفحة",
        "pdf_how": "طريقة الإصلاح", "pdf_where": "مكان الإصلاح",
        "pdf_none": "لم تظهر مشكلات واضحة ضمن الفحوصات الأساسية. تابع الأداء في Search Console واستمر في تحسين المحتوى.",
        "pdf_file": "📕 تقرير PDF: {f}",
        "pdf_nofont": "⚠️ تعذّر العثور على خط يدعم العربية، لذلك لم يُنشأ PDF. جرّب --lang en.",
        "exec_tpl": "تم فحص {pages} صفحة من الموقع {site} (المنصة: {platform}). التقييم العام {score} من 100 ({grade}). وُجدت {n_issues} مشكلة، منها {n_crit} حرجة و{n_high} عالية الخطورة. نوصي بالبدء بما يلي: {top}. ثم تنفيذ بقية الإصلاحات بالترتيب الموضح في ملف Excel المرفق.",
        "raw_score": "التقييم قبل تطبيق السقف (متوسط درجات الصفحات)",
        "crit_pages": "الصفحات التي فيها مشكلة حرجة",
        "coverage": "تغطية خريطة الموقع (الصفحات المفحوصة / الإجمالي)",
        "cap_crit": "{p}% من الصفحات فيها مشكلة حرجة",
        "cap_psi": "أداء الجوال {s}/100 في PageSpeed",
        "cap_cwv": "Core Web Vitals على الجوال ضعيفة حسب بيانات المستخدمين الحقيقيين",
        "cap_any": "لا يمكن أن يكون التقييم «ممتازًا» مع وجود مشكلات حرجة",
        "cap_line": "تم تحديد سقف للتقييم عند {cap} لأن: {r}. المتوسط وحده كان سيُظهر الموقع أفضل مما هو عليه.",
        "root_cause": "سبب مشترك محتمل",
        "root_url": "الرابط المكسور نفسه يظهر في {n} صفحة: {u}. إصلاح واحد قد يحل المشكلة كلها.",
        "root_anchor": "نص الرابط «{a}» يتكرر في {n} رابطًا مكسورًا، وغالبًا هو رابط في القالب أو القائمة. أصلحه مرة واحدة.",
        "dup_group": "{n} صفحات لها العنوان وعدد الكلمات نفسه («{t}»، {w} كلمة): غالبًا قالب أو صفحة مكررة. راجعها كمجموعة لا كمشكلات منفصلة.",
        "verify_note": "قبل تسليم التقرير لعميل: افتح يدويًا 3 روابط 404 وصفحة noindex للتأكد، وتذكّر أن أرقام PageSpeed تتغير من فحص لآخر.",
        "rtl": True,
    },
    "en": {
        "chars": "{n} chars", "words": "{n} words", "sec": "{n} s",
        "robots_dis": "Disallowed by robots.txt",
        "chain": "{n} redirects -> {u}",
        "redir": "Redirects to {u}",
        "img": "{a} of {b} images have no alt text",
        "mixed": "{n} resources loaded over http",
        "yes": "Yes", "no": "No",
        "g90": "Excellent", "g75": "Good", "g50": "Average (needs work)", "g0": "Poor (major issues)",
        "sheets": ["Summary", "Top Fixes", "All Issues", "Pages", "Broken Links", "Speed"],
        "status_n": "Status {n}",
        "page_cols": {},
        "bl_cols": ["Source page", "Broken link", "Status", "Type", "Anchor text"],
        "int": "Internal", "ext": "External",
        "speed_cols": ["URL", "Device", "Performance (/100)", "FCP (s)", "LCP (s)", "CLS", "TBT (ms)",
                       "TTFB (ms)", "Real-user rating (28 days)", "Real-user INP (ms)"],
        "broken_n": "{n} broken links",
        "js_d": "Very little text but many JavaScript elements",
        "psi_d": "Performance {s}/100",
        "crawl_found": "Pages added through internal links (not in sitemap)",
        "links_checked": "Links checked",
        "broken_total": "Broken links found",
        "orphan_skip": "Orphan-page detection skipped because not all pages were audited. Use --max 0 with a larger --crawl.",
        "p_crawl": "🕸️ Found {n} extra pages through internal links, auditing them ...",
        "p_links": "🔗 Checking {n} links for broken ones ...",
        "p_psi": "⚡ Running Google PageSpeed ({n} pages, mobile and desktop, may take several minutes) ...",
        "dev_mobile": "Mobile", "dev_desktop": "Desktop",
        "p_psi_err": "⚠️ PageSpeed failed ({d}) for {u} ({e}). If it is 429 the free quota is used up: try later or use --psi-key, or test the URL manually at pagespeed.web.dev",
        "sum_cols": ["Item", "Value"],
        "top_cols": ["Severity", "Issue", "Pages affected", "% of pages", "How to fix", "Where to fix"],
        "iss_cols": ["URL", "Severity", "Issue", "Detail", "How to fix", "Where to fix", "Code"],
        "site": "Website", "sitemap": "Sitemap", "sm_total": "URLs in sitemap",
        "checked": "Pages audited", "score": "Overall score (out of 100)", "verdict": "Verdict",
        "issues_of": "Issues ({s})",
        "note_l": "Note",
        "note": "Score is out of 100: the higher the better. It is this tool's own estimate, not an official Google metric.",
        "site_issue": "Site-wide issue",
        "no_robots": "No robots.txt file",
        "no_robots_fix": "Create a robots.txt file in the site root and add the sitemap URL to it.",
        "no_robots_where": "Site root: /robots.txt",
        "p_enter": "🔗 Enter the website URL (e.g. https://example.com): ",
        "p_search": "⏳ Looking for the sitemap of {b} ...",
        "p_nosm": "❌ No sitemap found (robots.txt, /sitemap.xml, /sitemap_index.xml, /wp-sitemap.xml).",
        "p_found": "✅ Sitemap: {sm} | {n} URLs",
        "p_sample": "ℹ️ Auditing {m} pages (an evenly spread sample). Use --max 0 to audit all.",
        "p_prog": "   {n}/{m} pages audited",
        "p_score": "🏁 Overall score: {s}/100 -> {g}",
        "p_top": "🔧 Top issues to fix:",
        "p_top_row": "   [{sev}] {issue} ({n} pages)",
        "p_file": "📄 Full report: {f}",
        "platform": "Site platform", "plat_unknown": "Not detected",
        "plat_note": "The site platform was not recognised, so the 'where to fix' tips are generic. Apply the fixes from your platform's admin panel or ask your developer.",
        "menu_note": "Menu and button names may differ slightly as platforms update.",
        "p_platform": "🧭 Site platform: {p}",
        "pdf_title": "SEO Audit Report", "pdf_prepared": "Prepared by", "pdf_date": "Date", "pdf_platform": "Platform",
        "pdf_pages": "Pages audited", "pdf_broken": "Broken links", "pdf_crit": "Critical issues", "pdf_high": "High issues",
        "pdf_sev": "Issues by severity", "pdf_speed": "Mobile speed (Google PageSpeed)",
        "pdf_exec": "Executive summary", "pdf_top": "Top fixes", "pdf_pg": "pages",
        "pdf_how": "How to fix", "pdf_where": "Where to fix",
        "pdf_none": "No clear issues were found in the basic checks. Keep tracking performance in Search Console and keep improving content.",
        "pdf_file": "📕 PDF report: {f}",
        "pdf_nofont": "⚠️ No Arabic-capable font found, so the PDF was not created. Try --lang en.",
        "exec_tpl": "We audited {pages} pages of {site} (platform: {platform}). The overall score is {score} out of 100 ({grade}). We found {n_issues} issues, {n_crit} critical and {n_high} high severity. We recommend starting with: {top}. The remaining fixes follow, in order, in the attached Excel report.",
        "raw_score": "Score before the cap (average of page scores)",
        "crit_pages": "Pages with a critical issue",
        "coverage": "Sitemap coverage (pages audited / total)",
        "cap_crit": "{p}% of pages have a critical issue",
        "cap_psi": "Mobile performance is {s}/100 in PageSpeed",
        "cap_cwv": "Mobile Core Web Vitals are poor according to real-user data",
        "cap_any": "The score cannot be 'Excellent' while critical issues exist",
        "cap_line": "The score was capped at {cap} because: {r}. The page average alone would have made the site look better than it is.",
        "root_cause": "Likely shared cause",
        "root_url": "The same broken link appears on {n} pages: {u}. One fix may solve all of them.",
        "root_anchor": "The link text '{a}' repeats in {n} broken links, most likely a theme or menu link. Fix it once.",
        "dup_group": "{n} pages share the same title and word count ('{t}', {w} words): probably a template or duplicated page. Review them as a group, not as separate issues.",
        "verify_note": "Before handing the report to a client: manually open 3 pages marked 404 and one noindex page to confirm, and remember PageSpeed numbers change from run to run.",
        "rtl": False,
    },
}


def tr(key, **kw):
    return S[LANG][key].format(**kw)


# ------------------------------------------------------------------ كتالوج المشكلات / Issue catalog
# code: (severity, (desc_ar, desc_en), (fix_ar, fix_en), (where_ar, where_en))
CATALOG = {
    "FETCH_FAIL": ("Critical",
        ("تعذّر تحميل الصفحة (انتهاء المهلة أو خطأ في الاتصال)", "Page failed to load (timeout or connection error)"),
        ("تأكد من أن الخادم يعمل بسرعة جيدة، وأن جدار الحماية أو Cloudflare لا يحجب الزوار.",
         "Make sure the server is up and fast, and that a firewall or Cloudflare is not blocking visitors."),
        ("الاستضافة / الخادم / إعدادات جدار الحماية", "Hosting / server / firewall settings")),
    "HTTP_4XX": ("Critical",
        ("الصفحة تعرض خطأ 4xx (مثل 404: الصفحة غير موجودة)", "Page returns a 4xx error (e.g. 404 Not Found)"),
        ("أزل الرابط من خريطة الموقع، أو أنشئ إعادة توجيه 301 إلى الصفحة البديلة، أو أعد نشر الصفحة.",
         "Remove the URL from the sitemap, add a 301 redirect to the replacement page, or restore the page."),
        ("ملف sitemap.xml وإعدادات إعادة التوجيه (.htaccess / Cloudflare / إضافة Redirection)",
         "sitemap.xml and redirect settings (.htaccess / Cloudflare / Redirection plugin)")),
    "HTTP_5XX": ("Critical",
        ("خطأ في الخادم (5xx)", "Server error (5xx)"),
        ("راجع سجلات الخادم (logs)؛ غالبًا تكون المشكلة في PHP أو قاعدة البيانات أو إضافة. تواصل مع الاستضافة إن لم تجد السبب.",
         "Check the server logs; it is usually a PHP, database or plugin issue. Contact your host if you cannot find the cause."),
        ("الخادم / الاستضافة / السجلات", "Server / hosting / logs")),
    "NOINDEX": ("Critical",
        ("الصفحة تحتوي على noindex (تمنع Google من فهرستها) رغم وجودها في خريطة الموقع",
         "Page has noindex (blocks Google from indexing it) yet is listed in the sitemap"),
        ("إن أردت ظهورها في Google فأزل noindex، وإن لم ترد ذلك فأزلها من خريطة الموقع.",
         "Remove noindex if the page should appear in Google; otherwise remove it from the sitemap."),
        ("وسم robots في <head> / ترويسة X-Robots-Tag / إعدادات Yoast أو RankMath",
         "robots meta tag in <head> / X-Robots-Tag header / Yoast or RankMath settings")),
    "ROBOTS_BLOCKED": ("Critical",
        ("الصفحة محظورة في ملف robots.txt", "Page is blocked by robots.txt"),
        ("عدّل قاعدة Disallow التي تحجب هذا الرابط.", "Edit the Disallow rule that blocks this URL."),
        ("ملف robots.txt في جذر الموقع", "robots.txt in the site root")),
    "NOT_HTTPS": ("High",
        ("الصفحة لا تعمل عبر HTTPS", "Page is not served over HTTPS"),
        ("فعّل شهادة SSL وأضف إعادة توجيه 301 من http إلى https لكامل الموقع.",
         "Install an SSL certificate and 301-redirect http to https site-wide."),
        ("الاستضافة (SSL) + .htaccess / Cloudflare / إعدادات الموقع", "Hosting (SSL) + .htaccess / Cloudflare / site settings")),
    "MIXED_CONTENT": ("High",
        ("صفحة HTTPS تحمّل ملفات (صور أو سكربتات) عبر http", "HTTPS page loads resources (images/scripts) over http"),
        ("استبدل كل روابط http:// بـ https:// في الصور وملفات CSS وJS.",
         "Replace all http:// links with https:// for images, CSS and JS."),
        ("كود القالب / المحتوى / إضافة مثل Really Simple SSL", "Theme code / content / a plugin such as Really Simple SSL")),
    "REDIRECT": ("Medium",
        ("الرابط في خريطة الموقع يعيد التوجيه إلى رابط آخر", "URL in the sitemap redirects elsewhere"),
        ("ضع الرابط النهائي (الذي يعطي حالة 200) في خريطة الموقع بدل الرابط القديم.",
         "Put the final URL (status 200) in the sitemap instead of the old one."),
        ("ملف sitemap.xml", "sitemap.xml")),
    "REDIRECT_CHAIN": ("High",
        ("سلسلة إعادة توجيه (أكثر من إعادة توجيه واحدة)", "Redirect chain (more than one redirect)"),
        ("اجعلها إعادة توجيه واحدة مباشرة من الرابط القديم إلى الرابط النهائي.",
         "Use a single direct redirect from the old URL to the final URL."),
        ("إعدادات إعادة التوجيه (.htaccess / Cloudflare / إضافة)", "Redirect settings (.htaccess / Cloudflare / plugin)")),
    "SLOW": ("Medium",
        ("الصفحة بطيئة (زمن الاستجابة أكثر من 1.5 ثانية)", "Slow page (response time over 1.5 seconds)"),
        ("فعّل التخزين المؤقت (Cache) وضغط الصور وشبكة CDN، وحسّن الاستضافة. قِس الأداء عبر PageSpeed Insights.",
         "Enable caching, compress images, use a CDN and upgrade hosting. Measure with PageSpeed Insights."),
        ("الاستضافة + إضافة التخزين المؤقت (WP Rocket / LiteSpeed) + الصور",
         "Hosting + cache plugin (WP Rocket / LiteSpeed) + images")),
    "BIG_PAGE": ("Low",
        ("حجم الصفحة كبير (أكثر من 1.5 ميغابايت من HTML)", "Large page (over 1.5 MB of HTML)"),
        ("قلّل الكود الزائد وفعّل ضغط gzip/brotli.", "Reduce unnecessary markup and enable gzip/brotli compression."),
        ("القالب / إعدادات الخادم", "Theme / server settings")),
    "TITLE_MISSING": ("High",
        ("لا يوجد عنوان للصفحة (Title)", "Missing title tag"),
        ("أضف وسم <title> فريدًا يتضمن الكلمة المفتاحية (30-60 حرفًا).",
         "Add a unique <title> containing the target keyword (30-60 characters)."),
        ("وسم <head> للصفحة / حقل عنوان SEO في Yoast أو RankMath", "Page <head> / SEO Title field in Yoast or RankMath")),
    "TITLE_SHORT": ("Medium",
        ("عنوان الصفحة قصير (أقل من 30 حرفًا)", "Title too short (under 30 characters)"),
        ("أطِل العنوان وأضف الكلمة المفتاحية وما يميّز الصفحة (30-60 حرفًا).",
         "Lengthen the title and include the keyword and a clear benefit (30-60 characters)."),
        ("حقل عنوان SEO في Yoast / RankMath أو وسم <title>", "SEO Title field in Yoast / RankMath or the <title> tag")),
    "TITLE_LONG": ("Medium",
        ("عنوان الصفحة طويل (أكثر من 60 حرفًا) وسيُقتطع في نتائج Google",
         "Title too long (over 60 characters) and will be truncated in Google"),
        ("اختصر العنوان وضع الكلمة المهمة في بدايته.", "Shorten the title and put the important keyword first."),
        ("حقل عنوان SEO في Yoast / RankMath أو وسم <title>", "SEO Title field in Yoast / RankMath or the <title> tag")),
    "TITLE_DUP": ("Medium",
        ("العنوان نفسه مستخدم في صفحات أخرى", "Same title used on other pages"),
        ("يجب أن يكون لكل صفحة عنوان خاص بها.", "Give every page its own unique title."),
        ("حقل عنوان SEO في كل صفحة", "SEO Title field on each page")),
    "DESC_MISSING": ("Medium",
        ("لا يوجد وصف ميتا (Meta Description)", "Missing meta description"),
        ("أضف وصفًا جذابًا (70-160 حرفًا) يتضمن الكلمة المفتاحية.",
         "Add a compelling description (70-160 characters) with the target keyword."),
        ("حقل الوصف في Yoast / RankMath أو وسم <meta name='description'> في <head>",
         "Description field in Yoast / RankMath or the <meta name='description'> tag in <head>")),
    "DESC_SHORT": ("Low",
        ("وصف الميتا قصير (أقل من 70 حرفًا)", "Meta description too short (under 70 characters)"),
        ("أطِل الوصف قليلًا (70-160 حرفًا).", "Lengthen it a little (70-160 characters)."),
        ("حقل وصف الميتا", "Meta description field")),
    "DESC_LONG": ("Low",
        ("وصف الميتا طويل (أكثر من 160 حرفًا)", "Meta description too long (over 160 characters)"),
        ("اختصر الوصف حتى لا يُقتطع.", "Shorten it so it is not truncated."),
        ("حقل وصف الميتا", "Meta description field")),
    "DESC_DUP": ("Medium",
        ("وصف الميتا نفسه مستخدم في صفحات أخرى", "Same meta description used on other pages"),
        ("اكتب وصفًا خاصًا بكل صفحة.", "Write a unique description for each page."),
        ("حقل وصف الميتا في كل صفحة", "Meta description field on each page")),
    "H1_MISSING": ("High",
        ("لا يوجد عنوان رئيسي H1", "Missing H1 heading"),
        ("أضف H1 واحدًا في الصفحة يتضمن الكلمة المفتاحية الرئيسية.", "Add a single H1 containing the main keyword."),
        ("محتوى الصفحة / القالب (عنوان الصفحة يجب أن يكون H1)", "Page content / theme (the page title should be an H1)")),
    "H1_MULTI": ("Medium",
        ("أكثر من H1 في الصفحة نفسها", "More than one H1 on the page"),
        ("اترك H1 واحدًا فقط وحوّل البقية إلى H2 أو H3.", "Keep one H1 and change the rest to H2/H3."),
        ("محتوى الصفحة / القالب / أداة بناء الصفحات", "Page content / theme / page builder")),
    "H1_DUP": ("Low",
        ("العنوان الرئيسي H1 نفسه مستخدم في صفحات أخرى", "Same H1 used on other pages"),
        ("اجعل H1 فريدًا لكل صفحة.", "Make each page's H1 unique."),
        ("محتوى الصفحة", "Page content")),
    "CANON_MISSING": ("Medium",
        ("لا يوجد رابط Canonical", "Missing canonical tag"),
        ("أضف <link rel='canonical'> يشير إلى الرابط الأصلي للصفحة.",
         "Add <link rel='canonical'> pointing to the page's preferred URL."),
        ("<head> / Yoast / RankMath (متقدم ← Canonical URL)", "<head> / Yoast / RankMath (Advanced > Canonical URL)")),
    "CANON_OTHER": ("Medium",
        ("رابط Canonical يشير إلى صفحة أخرى", "Canonical points to a different URL"),
        ("تأكد من أن ذلك مقصود، وإلا فاجعل الـ canonical يشير إلى الصفحة نفسها.",
         "Make sure this is intentional; otherwise point the canonical to the page itself."),
        ("<head> / Yoast / RankMath (متقدم ← Canonical URL)", "<head> / Yoast / RankMath (Advanced > Canonical URL)")),
    "IMG_ALT": ("Medium",
        ("صور بلا نص بديل (alt)", "Images missing alt text"),
        ("أضف نصًا بديلًا وصفيًا لكل صورة (يفيد SEO وإمكانية الوصول).",
         "Add descriptive alt text to every image (helps SEO and accessibility)."),
        ("مكتبة الوسائط أو الوسم <img alt=''>", "Media Library or the <img alt=''> attribute")),
    "THIN": ("Medium",
        ("محتوى قليل (أقل من 300 كلمة)", "Thin content (under 300 words)"),
        ("أضف محتوى مفيدًا وأصليًا يجيب عن نية الباحث، وتجنب الصفحات شبه الفارغة.",
         "Add useful, original content that answers search intent; avoid near-empty pages."),
        ("محتوى الصفحة", "Page content")),
    "NO_VIEWPORT": ("High",
        ("لا يوجد وسم viewport (الموقع غير مهيأ للجوال)", "Missing viewport tag (not mobile-friendly)"),
        ("أضف <meta name='viewport' content='width=device-width, initial-scale=1'>.",
         "Add <meta name='viewport' content='width=device-width, initial-scale=1'>."),
        ("<head> / القالب (header.php)", "<head> / theme (header.php)")),
    "NO_LANG": ("Low",
        ("لا يوجد وسم lang في <html>", "Missing lang attribute on <html>"),
        ("أضف lang مثل <html lang='ar'> أو 'fr' أو 'en'.", "Add lang, e.g. <html lang='en'>."),
        ("القالب (header.php) / إعدادات اللغة", "Theme (header.php) / language settings")),
    "NO_OG": ("Low",
        ("لا توجد وسوم Open Graph (معاينة المشاركة في فيسبوك وواتساب)",
         "Missing Open Graph tags (share previews on Facebook/WhatsApp)"),
        ("أضف og:title وog:description وog:image.", "Add og:title, og:description and og:image."),
        ("Yoast / RankMath (التواصل الاجتماعي) أو <head>", "Yoast / RankMath (Social) or <head>")),
    "NO_SCHEMA": ("Low",
        ("لا توجد بيانات منظمة (Schema)", "No structured data (Schema)"),
        ("أضف Schema بصيغة JSON-LD مناسبة (Article / Product / LocalBusiness / FAQ ...).",
         "Add suitable JSON-LD Schema (Article / Product / LocalBusiness / FAQ ...)."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    "URL_LONG": ("Low",
        ("الرابط طويل (أكثر من 115 حرفًا)", "URL too long (over 115 characters)"),
        ("اختصر الرابط واكتفِ بكلمات مفتاحية قليلة.", "Shorten the URL and keep only a few keywords."),
        ("الرابط الدائم (Slug) للصفحة", "Page slug / permalink")),
    "URL_STYLE": ("Low",
        ("الرابط يحتوي على أحرف كبيرة أو شرطة سفلية أو معاملات (parameters)",
         "URL contains uppercase letters, underscores or parameters"),
        ("استخدم أحرفًا صغيرة وشرطة (-) بدل الشرطة السفلية، وتجنب المعاملات.",
         "Use lowercase and hyphens instead of underscores; avoid parameters."),
        ("الرابط الدائم / إعدادات الروابط", "Slug / permalink settings")),
    "NO_INTERNAL_LINKS": ("Medium",
        ("لا توجد روابط داخلية في الصفحة", "No internal links on the page"),
        ("أضف روابط داخلية إلى صفحات مهمة أخرى في الموقع.", "Add internal links to other important pages."),
        ("محتوى الصفحة / القائمة", "Page content / menu")),
    "BROKEN_INTERNAL": ("High",
        ("روابط داخلية مكسورة (تؤدي إلى صفحات غير موجودة أو بها خطأ)", "Broken internal links (point to missing or erroring pages)"),
        ("عدّل الرابط ليشير إلى صفحة صحيحة، أو أنشئ إعادة توجيه 301، أو أزل الرابط. التفاصيل في ورقة «الروابط المكسورة».",
         "Point the link to a valid page, add a 301 redirect, or remove the link. See the 'Broken Links' sheet for details."),
        ("محتوى الصفحة / القائمة / القالب", "Page content / menu / theme")),
    "BROKEN_EXTERNAL": ("Medium",
        ("روابط خارجية مكسورة", "Broken external links"),
        ("استبدل الرابط بمصدر يعمل أو أزله. التفاصيل في ورقة «الروابط المكسورة».",
         "Replace the link with a working source or remove it. See the 'Broken Links' sheet for details."),
        ("محتوى الصفحة", "Page content")),
    "NOT_IN_SITEMAP": ("Low",
        ("صفحة موجودة ويمكن الوصول إليها لكنها غير مدرجة في خريطة الموقع", "Live page that is not listed in the sitemap"),
        ("أضفها إلى خريطة الموقع إن كانت صفحة مهمة، أو أضف لها noindex إن لم تكن كذلك.",
         "Add it to the sitemap if it is important, or noindex it if it is not."),
        ("sitemap.xml / إعدادات الخرائط في Yoast أو RankMath", "sitemap.xml / Sitemap settings in Yoast or RankMath")),
    "ORPHAN": ("Medium",
        ("صفحة يتيمة: لا توجد أي صفحة مفحوصة تربط إليها داخليًا", "Orphan page: no audited page links to it"),
        ("أضف روابط داخلية إليها من صفحات ذات صلة أو من القائمة الرئيسية.",
         "Add internal links to it from related pages or the main menu."),
        ("محتوى الصفحات / القائمة", "Page content / menu")),
    "HEADING_SKIP": ("Low",
        ("ترتيب العناوين غير منطقي (قفز مثلًا من H1 إلى H3)", "Illogical heading order (e.g. jumping from H1 to H3)"),
        ("رتّب العناوين H1 ثم H2 ثم H3 دون قفز بين المستويات.", "Order headings H1, H2, H3 without skipping levels."),
        ("محتوى الصفحة / القالب", "Page content / theme")),
    "JS_SUSPECT": ("Info",
        ("الصفحة تبدو معتمدة على JavaScript لعرض المحتوى، وقد تكون بعض نتائج الفحص غير دقيقة",
         "Page seems to rely on JavaScript to render content, so some results may be inaccurate"),
        ("تحقق من الصفحة يدويًا في المتصفح (قارن View Source مع Inspect) قبل الاعتماد على نتائجها.",
         "Check the page manually in the browser (compare View Source with Inspect) before trusting its results."),
        ("-", "-")),
    "PSI_POOR": ("Medium",
        ("أداء ضعيف على الجوال حسب Google PageSpeed (أقل من 50)", "Poor mobile performance per Google PageSpeed (below 50)"),
        ("اضغط الصور، فعّل التخزين المؤقت وCDN، قلّل ملفات JS وCSS غير الضرورية، وراجع تقرير PageSpeed للتفاصيل.",
         "Compress images, enable caching and a CDN, cut unneeded JS/CSS, and review the PageSpeed report for details."),
        ("الصور / القالب / الإضافات / الاستضافة", "Images / theme / plugins / hosting")),
    "PSI_AVG": ("Low",
        ("أداء متوسط على الجوال حسب Google PageSpeed (50-89)", "Average mobile performance per Google PageSpeed (50-89)"),
        ("حسّن الصور والتخزين المؤقت وقلّل الملفات الثقيلة.", "Improve images and caching, and reduce heavy files."),
        ("الصور / القالب / الإضافات / الاستضافة", "Images / theme / plugins / hosting")),
}


# ------------------------------------------------------------------ منصة الموقع / Platform-specific tips
# كل مشكلة لها "مكان إصلاح" يختلف حسب المنصة. WordPress يبقى كما في الكتالوج، والمنصات الأخرى لها نصوصها.
CODE_GROUP = {
    "HTTP_4XX": "redirect", "REDIRECT_CHAIN": "redirect", "NOINDEX": "noindex", "NOT_HTTPS": "https",
    "MIXED_CONTENT": "mixed", "SLOW": "slow", "TITLE_MISSING": "title", "TITLE_SHORT": "title",
    "TITLE_LONG": "title", "TITLE_DUP": "title", "DESC_MISSING": "desc", "DESC_DUP": "desc",
    "CANON_MISSING": "canon", "CANON_OTHER": "canon", "NO_OG": "og", "NO_SCHEMA": "schema",
    "NOT_IN_SITEMAP": "sitemap", "NO_VIEWPORT": "viewport", "NO_LANG": "lang", "IMG_ALT": "alt",
    "URL_LONG": "slug", "URL_STYLE": "slug", "ROBOTS_BLOCKED": "robots",
}
# group: (arabic, english)
PW = {
    "blogger": {
        "title": ("عنوان المقال نفسه في المحرر. وعنوان الصفحة الرئيسية: Settings ← Title", "The post title itself in the editor. Homepage title: Settings > Title"),
        "desc": ("Settings ← Search preferences ← Meta description للمدونة، وللمقال: Post settings ← Search description", "Settings > Search preferences > Meta description for the blog; for a post: Post settings > Search description"),
        "canon": ("Blogger يضيفه تلقائيًا في أغلب القوالب؛ راجع القالب: Theme ← Edit HTML", "Blogger adds it automatically in most themes; check the theme: Theme > Edit HTML"),
        "redirect": ("Settings ← Search preferences ← Custom redirects (وإزالة الرابط من خريطة الموقع إن لزم)", "Settings > Search preferences > Custom redirects (and remove the URL from the sitemap if needed)"),
        "noindex": ("Settings ← Search preferences ← Custom robots header tags، وللمقال: Post settings ← Custom robots tags", "Settings > Search preferences > Custom robots header tags; for a post: Post settings > Custom robots tags"),
        "https": ("Settings ← HTTPS ← فعّل HTTPS redirect", "Settings > HTTPS > enable HTTPS redirect"),
        "mixed": ("Theme ← Edit HTML ومحتوى المقالات: استبدل http:// بـ https://", "Theme > Edit HTML and post content: replace http:// with https://"),
        "slow": ("Layout (احذف الأدوات غير اللازمة) + ضغط الصور + قالب خفيف", "Layout (remove unneeded gadgets) + compress images + a lightweight theme"),
        "og": ("Theme ← Edit HTML: أضف وسوم og داخل <head> أو اختر قالبًا يتضمنها", "Theme > Edit HTML: add og tags inside <head> or choose a theme that includes them"),
        "schema": ("Theme ← Edit HTML: أضف JSON-LD داخل <head>", "Theme > Edit HTML: add JSON-LD inside <head>"),
        "sitemap": ("Blogger ينشئ خريطة الموقع تلقائيًا (/sitemap.xml)، والصفحات الثابتة غالبًا في /sitemap-pages.xml", "Blogger creates the sitemap automatically (/sitemap.xml); static pages are usually in /sitemap-pages.xml"),
        "viewport": ("Theme ← Edit HTML: وسم viewport في <head>، أو بدّل القالب بقالب متجاوب", "Theme > Edit HTML: viewport tag in <head>, or switch to a responsive theme"),
        "lang": ("Theme ← Edit HTML: خاصية lang في وسم <html>", "Theme > Edit HTML: the lang attribute on the <html> tag"),
        "alt": ("محرر Blogger: اضغط على الصورة ← Image properties ← Alt text", "Blogger editor: click the image > Image properties > Alt text"),
        "slug": ("Post settings ← Permalink (رابط مخصص)", "Post settings > Permalink (custom permalink)"),
        "robots": ("Settings ← Search preferences ← Custom robots.txt", "Settings > Search preferences > Custom robots.txt"),
    },
    "shopify": {
        "title": ("الصفحة أو المنتج ← Search engine listing ← Edit. وللرئيسية: Online Store ← Preferences", "Page or product > Search engine listing > Edit. Homepage: Online Store > Preferences"),
        "desc": ("الصفحة أو المنتج ← Search engine listing ← Edit. وللرئيسية: Online Store ← Preferences", "Page or product > Search engine listing > Edit. Homepage: Online Store > Preferences"),
        "canon": ("Shopify يضيفه تلقائيًا في القوالب الرسمية؛ راجع theme.liquid", "Shopify adds it automatically in official themes; check theme.liquid"),
        "redirect": ("Online Store ← Navigation ← URL redirects", "Online Store > Navigation > URL redirects"),
        "noindex": ("Search engine listing للصفحة أو ملف theme.liquid", "The page's Search engine listing or theme.liquid"),
        "https": ("Shopify يفعّله تلقائيًا؛ راجع Settings ← Domains", "Shopify enables it automatically; check Settings > Domains"),
        "mixed": ("كود القالب (Online Store ← Themes ← Edit code) أو المحتوى", "Theme code (Online Store > Themes > Edit code) or content"),
        "slow": ("قلّل التطبيقات (Apps) غير اللازمة + ضغط الصور", "Remove unneeded apps + compress images"),
        "og": ("theme.liquid أو تطبيق SEO", "theme.liquid or an SEO app"),
        "schema": ("theme.liquid (JSON-LD) أو تطبيق SEO", "theme.liquid (JSON-LD) or an SEO app"),
        "sitemap": ("Shopify ينشئ /sitemap.xml تلقائيًا", "Shopify creates /sitemap.xml automatically"),
        "viewport": ("theme.liquid ← وسم viewport في <head>", "theme.liquid > viewport tag in <head>"),
        "lang": ("theme.liquid ← lang في وسم <html>", "theme.liquid > lang on the <html> tag"),
        "alt": ("اضغط على صورة المنتج أو الصفحة ← Add alt text", "Click the product or page image > Add alt text"),
        "slug": ("URL handle ضمن Search engine listing", "URL handle inside Search engine listing"),
        "robots": ("robots.txt.liquid في القالب", "robots.txt.liquid in the theme"),
    },
    "wix": {
        "title": ("Pages ← النقاط الثلاث ← SEO basics", "Pages > three dots > SEO basics"),
        "desc": ("Pages ← النقاط الثلاث ← SEO basics", "Pages > three dots > SEO basics"),
        "canon": ("Wix يضيفه تلقائيًا؛ ويمكن تعديله من Page SEO ← Advanced SEO", "Wix adds it automatically; it can be edited in Page SEO > Advanced SEO"),
        "redirect": ("SEO Tools ← URL Redirect Manager", "SEO Tools > URL Redirect Manager"),
        "noindex": ("Pages ← SEO basics ← خيار السماح لمحركات البحث بفهرسة الصفحة", "Pages > SEO basics > the option that lets search engines index the page"),
        "https": ("Wix يفعّله تلقائيًا؛ راجع إعدادات الدومين", "Wix enables it automatically; check the domain settings"),
        "mixed": ("محتوى الصفحة: استبدل http:// بـ https://", "Page content: replace http:// with https://"),
        "slow": ("قلّل الوسائط والتطبيقات (Apps) وضغط الصور", "Reduce media and apps, compress images"),
        "og": ("Pages ← SEO basics ← Social share", "Pages > SEO basics > Social share"),
        "schema": ("Page SEO ← Advanced SEO ← Structured data markup", "Page SEO > Advanced SEO > Structured data markup"),
        "sitemap": ("Wix ينشئ خريطة الموقع تلقائيًا", "Wix creates the sitemap automatically"),
        "viewport": ("يتحكم به Wix تلقائيًا، وقد تكون النتيجة غير دقيقة", "Controlled by Wix automatically, so this result may be inaccurate"),
        "lang": ("يتحكم به Wix حسب لغة الموقع في الإعدادات", "Controlled by Wix from the site language in settings"),
        "alt": ("اضغط على الصورة ← Settings ← What's in the image? (Alt text)", "Click the image > Settings > What's in the image? (Alt text)"),
        "slug": ("Pages ← SEO basics ← URL slug", "Pages > SEO basics > URL slug"),
        "robots": ("SEO Tools ← Robots.txt Editor", "SEO Tools > Robots.txt Editor"),
    },
    "other": {   # منصة غير معروفة: نص عام بلا أسماء إضافات
        "title": ("حقل عنوان SEO للصفحة في لوحة تحكم موقعك أو وسم <title> في <head>", "The page's SEO title field in your admin panel, or the <title> tag in <head>"),
        "desc": ("حقل وصف SEO للصفحة في لوحة تحكم موقعك أو وسم <meta name='description'>", "The page's SEO description field in your admin panel, or the <meta name='description'> tag"),
        "canon": ("إعدادات SEO للصفحة أو وسم <link rel='canonical'> في <head>", "The page's SEO settings, or the <link rel='canonical'> tag in <head>"),
        "redirect": ("إعدادات إعادة التوجيه في منصتك أو الاستضافة", "Redirect settings in your platform or hosting"),
        "noindex": ("إعدادات فهرسة الصفحة في منصتك أو وسم robots في <head>", "The page's indexing setting in your platform, or the robots meta tag in <head>"),
        "https": ("الاستضافة (SSL) وإعدادات الدومين في منصتك", "Hosting (SSL) and domain settings in your platform"),
        "mixed": ("كود القالب أو محتوى الصفحة", "Theme code or page content"),
        "slow": ("الاستضافة + ضغط الصور + تقليل الإضافات أو التطبيقات", "Hosting + compress images + reduce plugins or apps"),
        "og": ("إعدادات المشاركة الاجتماعية في منصتك أو وسوم og في <head>", "Social sharing settings in your platform, or og tags in <head>"),
        "schema": ("إعدادات SEO أو سكربت JSON-LD في <head>", "SEO settings or a JSON-LD script in <head>"),
        "sitemap": ("إعدادات خريطة الموقع في منصتك أو ملف sitemap.xml", "Sitemap settings in your platform, or sitemap.xml"),
        "viewport": ("كود القالب: وسم viewport في <head>", "Theme code: the viewport tag in <head>"),
        "lang": ("كود القالب: خاصية lang في وسم <html>", "Theme code: the lang attribute on the <html> tag"),
        "alt": ("إعدادات الصورة (النص البديل) في منصتك أو الوسم <img alt=''>", "The image's alt text setting in your platform, or the <img alt=''> attribute"),
        "slug": ("الرابط الدائم (Slug) للصفحة في منصتك", "The page slug / permalink in your platform"),
        "robots": ("ملف robots.txt في جذر الموقع أو إعدادات منصتك", "robots.txt in the site root, or your platform's settings"),
    },
}
PW["squarespace"] = PW["other"]
PW["webflow"] = PW["other"]
# بعض الحلول (وليس المكان فقط) تتغير حسب المنصة
PWF = {
    "blogger": {"HTTP_5XX": ("هذا غالبًا خطأ مؤقت من خوادم Blogger. أعد الفحص لاحقًا، وإن استمر فراجع حالة الخدمة.", "This is usually a temporary error on Blogger's servers. Re-run the audit later, and check the service status if it persists.")},
    "shopify": {"HTTP_5XX": ("راجع حالة Shopify وتطبيقات المتجر والقالب؛ وتواصل مع دعم Shopify إن استمر الخطأ.", "Check Shopify status, the store apps and the theme; contact Shopify support if it persists.")},
    "wix": {"HTTP_5XX": ("راجع حالة Wix وتواصل مع دعم Wix إن استمر الخطأ.", "Check Wix status and contact Wix support if the error persists.")},
}
PLATFORM_NAMES = {"blogger": "Blogger", "wordpress": "WordPress", "shopify": "Shopify", "wix": "Wix",
                  "squarespace": "Squarespace", "webflow": "Webflow"}


def platform_label():
    return PLATFORM_NAMES.get(PLATFORM) or tr("plat_unknown")


def detect_platform(html, url):
    h = html.lower()
    host = urlparse(url).netloc.lower()
    gen = ""
    try:
        g = BeautifulSoup(html, "html.parser").find("meta", attrs={"name": re.compile(r"^generator$", re.I)})
        gen = (g.get("content") or "").lower() if g else ""
    except Exception:
        pass
    if "blogger" in gen or "blogspot." in host:
        return "blogger"
    if "wordpress" in gen or "/wp-content/" in h or "/wp-includes/" in h:
        return "wordpress"
    if "shopify" in gen or "cdn.shopify.com" in h:
        return "shopify"
    if "wix.com" in gen or "wixstatic.com" in h:
        return "wix"
    if "squarespace" in h:
        return "squarespace"
    if "webflow" in gen or "webflow.com" in h:
        return "webflow"
    return "other"


def cat(code):
    sev, d, f, w = CATALOG[code]
    i = 0 if LANG == "ar" else 1
    fix, where = f[i], w[i]
    if PLATFORM != "wordpress":
        fx = PWF.get(PLATFORM, {}).get(code)
        if fx:
            fix = fx[i]
        grp = CODE_GROUP.get(code)
        pw = PW.get(PLATFORM, PW["other"]).get(grp) if grp else None
        if pw:
            where = pw[i]
    return sev, d[i], fix, where


# ------------------------------------------------------------------ الشبكة / Network
def make_session():
    s = requests.Session()
    s.headers.update(HEADERS)
    return s


def fetch(session, url, timeout, retries=2):
    last = None
    for i in range(retries + 1):
        try:
            return session.get(url, timeout=timeout, allow_redirects=True)
        except Exception as e:  # noqa
            last = e
            time.sleep(1 + i)
    raise last


# ------------------------------------------------------------------ Sitemap
def read_locs(content, url):
    if url.endswith(".gz") or content[:2] == b"\x1f\x8b":
        content = gzip.decompress(content)
    text = content.decode("utf-8", errors="ignore")
    is_index = "<sitemapindex" in text
    locs = [re.sub(r"<!\[CDATA\[|\]\]>", "", m).strip()
            for m in re.findall(r"<loc>\s*(.*?)\s*</loc>", text, re.S)]
    return is_index, locs


def expand_sitemap(session, url, timeout, depth=0, seen=None):
    seen = seen if seen is not None else set()
    if url in seen or depth > 3:
        return []
    seen.add(url)
    try:
        r = fetch(session, url, timeout, retries=1)
        if r.status_code != 200:
            return []
        is_index, locs = read_locs(r.content, url)
    except Exception:
        return []
    if not is_index:
        return locs
    urls = []
    for child in locs[:50]:
        urls += expand_sitemap(session, child, timeout, depth + 1, seen)
    return urls


def discover(session, base, timeout):
    robots_txt = ""
    candidates = []
    try:
        r = session.get(base + "/robots.txt", timeout=timeout)
        if r.status_code == 200 and "<html" not in r.text[:500].lower():
            robots_txt = r.text
            for line in robots_txt.splitlines():
                if line.lower().startswith("sitemap:"):
                    candidates.append(line.split(":", 1)[1].strip())
    except Exception:
        pass
    for p in ("/sitemap.xml", "/sitemap_index.xml", "/wp-sitemap.xml"):
        candidates.append(base + p)
    for c in dict.fromkeys(candidates):
        urls = expand_sitemap(session, c, timeout)
        if urls:
            return c, list(dict.fromkeys(urls)), robots_txt
    return None, [], robots_txt


# ------------------------------------------------------------------ فحص صفحة / Page analysis
def norm(u):
    u = u.split("#")[0].strip().lower()
    return u[:-1] if u.endswith("/") else u


ASSET_RE = re.compile(r"\.(jpe?g|png|gif|webp|svg|ico|pdf|zip|rar|css|js|mp4|mp3|docx?|xlsx?|pptx?)(\?.*)?$", re.I)


def looks_like_site(s):
    """True for 'https://x', 'http://x' or a bare domain like 'example.com' (not a file path)."""
    s = (s or "").strip()
    if s.lower().startswith(("http://", "https://")):
        return True
    return bool(re.match(r"^([a-z0-9-]+\.)+[a-z]{2,}(:\d+)?(/.*)?$", s, re.I)) and not s.lower().endswith(".json")


def same_host(a, b):
    ha = urlparse(a).netloc.lower().replace("www.", "")
    hb = urlparse(b).netloc.lower().replace("www.", "")
    return ha == hb


def check_link(session, url, timeout=10):
    """Returns the HTTP status of a link (or 'ERROR'). Doubtful results (timeouts, 5xx) are re-checked once."""
    def once():
        try:
            r = session.head(url, timeout=timeout, allow_redirects=True)
            if r.status_code >= 400:
                r = session.get(url, timeout=timeout, allow_redirects=True, stream=True)
                r.close()
            return r.status_code
        except Exception:
            return "ERROR"

    st = once()
    if st == "ERROR" or (isinstance(st, int) and st >= 500):
        time.sleep(1.5)
        st = once()
    return st


def is_broken(st):
    # 401/403/429 are usually bot protection, not a truly broken link
    return st == "ERROR" or st in (404, 410) or (isinstance(st, int) and 500 <= st < 600)


PSI_CAT = {
    "ar": {"FAST": "سريع", "AVERAGE": "متوسط", "SLOW": "بطيء"},
    "en": {"FAST": "Fast", "AVERAGE": "Average", "SLOW": "Slow"},
}


def parse_psi(data):
    """Extract [score, FCP s, LCP s, CLS, TBT ms, TTFB ms, real-user rating, real-user INP ms] from a PSI response."""
    lh = data["lighthouseResult"]
    a = lh["audits"]

    def num(key, scale=1, nd=None):
        try:
            v = a[key]["numericValue"] / scale
            return round(v, nd) if nd is not None else round(v)
        except Exception:
            return "n/a"

    score = round(lh["categories"]["performance"]["score"] * 100)
    field = data.get("loadingExperience") or {}
    cat = PSI_CAT[LANG].get(field.get("overall_category"), "n/a")
    inp = (field.get("metrics", {}).get("INTERACTION_TO_NEXT_PAINT") or {}).get("percentile", "n/a")
    return [score,
            num("first-contentful-paint", 1000, 2),
            num("largest-contentful-paint", 1000, 2),
            num("cumulative-layout-shift", 1, 3),
            num("total-blocking-time"),
            num("server-response-time"),
            cat, inp]


def pagespeed(session, url, key="", strategy="mobile", timeout=90):
    """Google PageSpeed Insights API (mobile or desktop). Returns (values, error)."""
    params = {"url": url, "strategy": strategy, "category": "performance"}
    if key:
        params["key"] = key
    try:
        r = session.get("https://www.googleapis.com/pagespeedonline/v5/runPagespeed",
                        params=params, timeout=timeout)
        if r.status_code != 200:
            return None, f"HTTP {r.status_code}"
        return parse_psi(r.json()), ""
    except Exception as e:  # noqa
        return None, str(e)[:80]


def analyze(url, session, timeout, delay, rp):
    issues = []
    row = {"URL": url}
    try:
        r = fetch(session, url, timeout)
    except Exception as e:  # noqa
        row["Status"] = "ERROR"
        issues.append(("FETCH_FAIL", str(e)[:150]))
        return row, issues
    time.sleep(delay)

    final = r.url
    status = r.status_code
    row.update({
        "Status": status,
        "Final URL": final,
        "Redirects": len(r.history),
        "Response (s)": round(r.elapsed.total_seconds(), 2),
        "Size (KB)": round(len(r.content) / 1024),
    })

    if rp is not None and not rp.can_fetch("*", url):
        issues.append(("ROBOTS_BLOCKED", tr("robots_dis")))
    if len(r.history) >= 2:
        issues.append(("REDIRECT_CHAIN", tr("chain", n=len(r.history), u=final)))
    elif len(r.history) == 1:
        issues.append(("REDIRECT", tr("redir", u=final)))
    if 400 <= status < 500:
        issues.append(("HTTP_4XX", tr("status_n", n=status)))
        return row, issues
    if status >= 500:
        issues.append(("HTTP_5XX", tr("status_n", n=status)))
        return row, issues

    if not final.startswith("https://"):
        issues.append(("NOT_HTTPS", final))
    if row["Response (s)"] > 1.5:
        issues.append(("SLOW", tr("sec", n=row["Response (s)"])))
    if len(r.content) > 1_500_000:
        issues.append(("BIG_PAGE", f"{row['Size (KB)']} KB"))

    ctype = r.headers.get("Content-Type", "")
    if "html" not in ctype.lower():
        return row, issues

    soup = BeautifulSoup(r.content, "html.parser")

    # URL
    path = urlparse(final)
    if len(final) > 115:
        issues.append(("URL_LONG", tr("chars", n=len(final))))
    if re.search(r"[A-Z]", path.path) or "_" in path.path or path.query:
        issues.append(("URL_STYLE", final))

    # Title
    t = soup.find("title")
    title = t.get_text(strip=True) if t else ""
    row["Title"] = title
    row["Title len"] = len(title)
    if not title:
        issues.append(("TITLE_MISSING", ""))
    elif len(title) < 30:
        issues.append(("TITLE_SHORT", f"{tr('chars', n=len(title))}: {title}"))
    elif len(title) > 60:
        issues.append(("TITLE_LONG", f"{tr('chars', n=len(title))}: {title}"))

    # Meta description
    m = soup.find("meta", attrs={"name": re.compile(r"^description$", re.I)})
    desc = (m.get("content") or "").strip() if m else ""
    row["Meta description"] = desc
    row["Desc len"] = len(desc)
    if not desc:
        issues.append(("DESC_MISSING", ""))
    elif len(desc) < 70:
        issues.append(("DESC_SHORT", tr("chars", n=len(desc))))
    elif len(desc) > 160:
        issues.append(("DESC_LONG", tr("chars", n=len(desc))))

    # H1
    h1s = [h.get_text(" ", strip=True) for h in soup.find_all("h1")]
    h1s = [h for h in h1s if h]
    row["H1"] = h1s[0] if h1s else ""
    row["H1 count"] = len(h1s)
    if not h1s:
        issues.append(("H1_MISSING", ""))
    elif len(h1s) > 1:
        issues.append(("H1_MULTI", f"{len(h1s)} H1"))

    # Canonical
    can = soup.find("link", rel=lambda v: v and "canonical" in
                    ([x.lower() for x in v] if isinstance(v, list) else [v.lower()]))
    canonical = (can.get("href") or "").strip() if can else ""
    row["Canonical"] = canonical
    if not canonical:
        issues.append(("CANON_MISSING", ""))
    elif norm(urljoin(final, canonical)) != norm(final):
        issues.append(("CANON_OTHER", f"canonical = {canonical}"))

    # Noindex
    rb = soup.find("meta", attrs={"name": re.compile(r"^robots$", re.I)})
    robots_meta = (rb.get("content") or "").lower() if rb else ""
    xrt = r.headers.get("X-Robots-Tag", "").lower()
    noindex = "noindex" in robots_meta or "noindex" in xrt
    row["Noindex"] = tr("yes") if noindex else tr("no")
    if noindex:
        issues.append(("NOINDEX", robots_meta or xrt))

    # Mobile / lang / social / schema
    if not soup.find("meta", attrs={"name": re.compile(r"^viewport$", re.I)}):
        issues.append(("NO_VIEWPORT", ""))
    if not (soup.html and soup.html.get("lang")):
        issues.append(("NO_LANG", ""))
    if not soup.find("meta", property="og:title"):
        issues.append(("NO_OG", ""))
    if not (soup.find("script", type="application/ld+json") or soup.find(attrs={"itemtype": True})):
        issues.append(("NO_SCHEMA", ""))

    # Images
    imgs = soup.find_all("img")
    no_alt = [i for i in imgs if i.get("alt") is None]
    row["Images"] = len(imgs)
    row["Images no alt"] = len(no_alt)
    if no_alt:
        issues.append(("IMG_ALT", tr("img", a=len(no_alt), b=len(imgs))))

    # Mixed content
    if final.startswith("https://"):
        bad = 0
        for tag in soup.find_all(["img", "script", "iframe", "source", "link"]):
            u = tag.get("src") or (tag.get("href") if tag.name == "link" else None)
            if u and u.startswith("http://"):
                bad += 1
        if bad:
            issues.append(("MIXED_CONTENT", tr("mixed", n=bad)))

    # Links (internal + external), kept for the crawl / broken-link phases
    links, seen_links = [], set()
    for a in soup.find_all("a", href=True):
        href = a["href"].strip()
        if not href or href.startswith(("#", "mailto:", "tel:", "javascript:", "sms:", "whatsapp:")):
            continue
        absu = urldefrag(urljoin(final, href))[0]
        if not absu.startswith(("http://", "https://")) or absu in seen_links:
            continue
        seen_links.add(absu)
        links.append((absu, same_host(absu, final), a.get_text(" ", strip=True)[:60]))
    row["_links"] = links
    internal = sum(1 for _, i, _t in links if i)
    row["Internal links"] = internal
    row["External links"] = len(links) - internal
    if internal == 0:
        issues.append(("NO_INTERNAL_LINKS", ""))

    # Heading order (e.g. H1 -> H3 without H2)
    levels = [int(h.name[1]) for h in soup.find_all(re.compile(r"^h[1-6]$"))]
    if any(b - a > 1 for a, b in zip(levels, levels[1:])):
        issues.append(("HEADING_SKIP", "H" + " > H".join(str(x) for x in levels[:12])))

    # Word count
    n_scripts = len(soup.find_all("script"))
    js_root = soup.find(id=re.compile(r"^(root|app|__next|__nuxt)$"))
    for tag in soup(["script", "style", "noscript"]):
        tag.decompose()
    words = len(soup.get_text(" ", strip=True).split())
    row["Words"] = words
    if words < 300:
        issues.append(("THIN", tr("words", n=words)))
    if words < 50 and (js_root is not None or n_scripts >= 8):
        issues.append(("JS_SUSPECT", tr("js_d")))

    return row, issues


# ------------------------------------------------------------------ التقييم والتقرير / Scoring & report
def robots_where():
    pw = PW.get(PLATFORM, PW["other"]).get("robots") if PLATFORM != "wordpress" else None
    return pw[0 if LANG == "ar" else 1] if pw else tr("no_robots_where")


def grade(score):
    if score >= 90:
        return tr("g90")
    if score >= 75:
        return tr("g75")
    if score >= 50:
        return tr("g50")
    return tr("g0")


def style_sheet(ws, widths, sev_col=None):
    from openpyxl.styles import Alignment, Font, PatternFill
    head = PatternFill("solid", fgColor="1F4E78")
    for c in ws[1]:
        c.font = Font(bold=True, color="FFFFFF")
        c.fill = head
        c.alignment = Alignment(vertical="center", wrap_text=True)
    for col, w in widths.items():
        ws.column_dimensions[col].width = w
    ws.freeze_panes = "A2"
    ws.auto_filter.ref = ws.dimensions
    ws.sheet_view.rightToLeft = S[LANG]["rtl"]
    raw = {"Critical": "F8CBAD", "High": "FFE699", "Medium": "FFF2CC", "Low": "E2EFDA", "Info": "DDEBF7"}
    colors = {SEVLBL[LANG][k]: v for k, v in raw.items()}
    for row in ws.iter_rows(min_row=2):
        for c in row:
            c.alignment = Alignment(vertical="top", wrap_text=True)
        if sev_col is not None:
            v = row[sev_col].value
            if v in colors:
                row[sev_col].fill = PatternFill("solid", fgColor=colors[v])


# ------------------------------------------------------------------ تقرير PDF ملوّن / Colourful PDF report
def _pip(pkg, mod):
    try:
        __import__(mod)
    except ImportError:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", pkg])


def build_pdf(path, base, site_score, n_pages, n_broken, sev_count, top_list, speed_rows, n_issues, cap_note=""):
    _pip("reportlab", "reportlab")
    ar = LANG == "ar"
    if ar:
        _pip("arabic-reshaper", "arabic_reshaper")
        _pip("python-bidi", "bidi")
        import arabic_reshaper
        try:
            from bidi import get_display
        except ImportError:
            from bidi.algorithm import get_display
    from reportlab.lib.colors import HexColor, white
    from reportlab.lib.pagesizes import A4
    from reportlab.pdfbase import pdfmetrics
    from reportlab.pdfbase.ttfonts import TTFont
    from reportlab.pdfgen import canvas

    FONT, FONT_B = "Helvetica", "Helvetica-Bold"
    if ar:
        regs = ["/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf", "/usr/share/fonts/dejavu/DejaVuSans.ttf"]
        bolds = ["/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf", "/usr/share/fonts/dejavu/DejaVuSans-Bold.ttf"]
        reg = next((x for x in regs if os.path.exists(x)), None)
        bold = next((x for x in bolds if os.path.exists(x)), reg)
        if not reg:
            print(tr("pdf_nofont"))
            return False
        pdfmetrics.registerFont(TTFont("AR", reg))
        pdfmetrics.registerFont(TTFont("AR-B", bold))
        FONT, FONT_B = "AR", "AR-B"

        def shape(t):
            return get_display(arabic_reshaper.reshape(str(t)))
    else:
        def shape(t):
            return str(t)

    NAVY, TEAL, TXT, GREY, SOFT = HexColor("#1F3A6E"), HexColor("#16A2B8"), HexColor("#222222"), HexColor("#666666"), HexColor("#EAF1FB")
    SEV_COL = {"Critical": HexColor("#D9534F"), "High": HexColor("#F0883E"), "Medium": HexColor("#F2C94C"),
               "Low": HexColor("#2E9E5B"), "Info": HexColor("#5B9BD5")}
    lbl_to_sev = {v: k for k, v in SEVLBL[LANG].items()}
    W, H = A4
    M = 42
    SX = W - M if ar else M

    def put(c, t, y, font, size, color, x=None):
        c.setFont(font, size)
        c.setFillColor(color)
        (c.drawRightString if ar else c.drawString)(SX if x is None else x, y, shape(t))

    def wrap(t, font, size, maxw):
        lines, cur = [], ""
        for w in str(t).split(" "):
            trial = (cur + " " + w).strip()
            if not cur or pdfmetrics.stringWidth(shape(trial), font, size) <= maxw:
                cur = trial
            else:
                lines.append(cur)
                cur = w
        if cur:
            lines.append(cur)
        return lines

    c = canvas.Canvas(path, pagesize=A4)
    c.setTitle(f"{tr('pdf_title')} - {base}")
    st = {"y": H, "n": 1}

    def footer():
        c.setFont(FONT, 8)
        c.setFillColor(GREY)
        c.drawCentredString(W / 2, 22, shape(f"{tr('pdf_title')}  |  {st['n']}"))

    def new_page():
        footer()
        c.showPage()
        st["n"] += 1
        st["y"] = H - 50

    def ensure(h):
        if st["y"] - h < 50:
            new_page()

    # ---- header
    c.setFillColor(NAVY)
    c.rect(0, H - 135, W, 135, fill=1, stroke=0)
    c.setFillColor(TEAL)
    c.rect(0, H - 141, W, 6, fill=1, stroke=0)
    put(c, tr("pdf_title"), H - 56, FONT_B, 25, white)
    put(c, base, H - 82, FONT, 12, HexColor("#CFE3F3"))
    meta = (f"{tr('pdf_date')}: {datetime.date.today().isoformat()}   |   {tr('pdf_platform')}: {platform_label()}"
            f"   |   {tr('pdf_prepared')}: {AUTHOR_NAME}")
    put(c, meta, H - 112, FONT, 9, HexColor("#CFE3F3"))

    # ---- score circle
    col = (SEV_COL["Low"] if site_score >= 90 else HexColor("#7CB342") if site_score >= 75
           else SEV_COL["High"] if site_score >= 50 else SEV_COL["Critical"])
    cx, cy, r = W / 2, H - 215, 54
    c.setFillColor(SOFT)
    c.circle(cx, cy, r + 9, fill=1, stroke=0)
    c.setFillColor(col)
    c.circle(cx, cy, r, fill=1, stroke=0)
    c.setFillColor(white)
    c.setFont("Helvetica-Bold", 36)
    c.drawCentredString(cx, cy - 4, f"{site_score:g}")
    c.setFont("Helvetica", 11)
    c.drawCentredString(cx, cy - 22, "/ 100")
    verdict = grade(site_score)
    c.setFillColor(col)
    c.setFont(FONT_B, 13)
    c.drawCentredString(cx, cy - r - 26, shape(verdict))

    # ---- 4 stat cards
    cards = [(tr("pdf_pages"), n_pages, NAVY), (tr("pdf_broken"), n_broken, SEV_COL["High"]),
             (tr("pdf_crit"), sev_count.get("Critical", 0), SEV_COL["Critical"]),
             (tr("pdf_high"), sev_count.get("High", 0), SEV_COL["High"])]
    gap = 10
    bw = (W - 2 * M - 3 * gap) / 4
    by = cy - r - 112
    for i, (lab, val, ccol) in enumerate(cards):
        bx = M + (3 - i if ar else i) * (bw + gap)
        c.setFillColor(SOFT)
        c.roundRect(bx, by, bw, 58, 7, fill=1, stroke=0)
        c.setFillColor(ccol)
        c.roundRect(bx, by, 5, 58, 2, fill=1, stroke=0)
        c.setFont("Helvetica-Bold", 20)
        c.drawCentredString(bx + bw / 2, by + 30, str(val))
        c.setFillColor(GREY)
        c.setFont(FONT, 8.5)
        c.drawCentredString(bx + bw / 2, by + 12, shape(lab))
    st["y"] = by - 28

    # ---- severity bars
    put(c, tr("pdf_sev"), st["y"], FONT_B, 13, NAVY)
    st["y"] -= 20
    mx = max([sev_count.get(k, 0) for k in ("Critical", "High", "Medium", "Low")] + [1])
    barw = W - 2 * M - 120
    for k in ("Critical", "High", "Medium", "Low"):
        n = sev_count.get(k, 0)
        put(c, SEVLBL[LANG][k], st["y"], FONT, 10, TXT)
        x0 = (W - M - 70 - barw) if ar else (M + 70)
        yb = st["y"] - 2
        c.setFillColor(SOFT)
        c.roundRect(x0, yb, barw, 10, 5, fill=1, stroke=0)
        fw = max(barw * n / mx, 2) if n else 0
        if fw:
            c.setFillColor(SEV_COL[k])
            c.roundRect((x0 + barw - fw) if ar else x0, yb, fw, 10, 5, fill=1, stroke=0)
        c.setFillColor(TXT)
        c.setFont("Helvetica-Bold", 10)
        (c.drawString if ar else c.drawRightString)(M if ar else W - M, yb + 1, str(n))
        st["y"] -= 22

    # ---- PageSpeed (mobile) card, only when available
    mob = next((r for r in speed_rows if len(r) > 4 and isinstance(r[2], (int, float))), None)
    if mob:
        st["y"] -= 8
        ensure(70)
        put(c, tr("pdf_speed"), st["y"], FONT_B, 13, NAVY)
        st["y"] -= 18
        sc = mob[2]
        scol = SEV_COL["Low"] if sc >= 90 else SEV_COL["Medium"] if sc >= 50 else SEV_COL["Critical"]
        c.setFillColor(scol)
        c.roundRect(M, st["y"] - 28, W - 2 * M, 36, 7, fill=1, stroke=0)
        c.setFillColor(white)
        c.setFont("Helvetica-Bold", 12)
        c.drawCentredString(W / 2, st["y"] - 14, f"{sc}/100   |   LCP {mob[4]}s   |   CLS {mob[5]}   |   TBT {mob[6]}ms")
        st["y"] -= 46

    # ---- executive summary
    st["y"] -= 8
    ensure(120)
    put(c, tr("pdf_exec"), st["y"], FONT_B, 13, NAVY)
    st["y"] -= 18
    if top_list:
        sep = "، " if ar else ", "
        top = sep.join(x[1] for x in top_list[:3])
        text = tr("exec_tpl", pages=n_pages, site=base, platform=platform_label(), score=f"{site_score:g}", grade=verdict,
                  n_issues=n_issues, n_crit=sev_count.get("Critical", 0), n_high=sev_count.get("High", 0), top=top)
    else:
        text = tr("pdf_none")
    if cap_note:
        text += " " + cap_note
    for ln in wrap(text, FONT, 10, W - 2 * M):
        ensure(16)
        put(c, ln, st["y"], FONT, 10, TXT)
        st["y"] -= 15

    # ---- top fixes
    st["y"] -= 10
    ensure(70)
    put(c, tr("pdf_top"), st["y"], FONT_B, 13, NAVY)
    st["y"] -= 20
    for sev_l, desc, cnt, pct, fix, where in top_list[:8]:
        sk = lbl_to_sev.get(sev_l, "Low")
        lines = [(tr("pdf_how") + ": " + fix, TXT), (tr("pdf_where") + ": " + where, GREY)]
        body = []
        for t, colr in lines:
            body += [(ln, colr) for ln in wrap(t, FONT, 8.5, W - 2 * M - 18)][:4]
        ensure(34 + 11.5 * len(body))
        dx = (W - M - 4) if ar else (M + 4)
        c.setFillColor(SEV_COL[sk])
        c.circle(dx, st["y"] + 3, 4.5, fill=1, stroke=0)
        tx = (W - M - 16) if ar else (M + 16)
        for i, ln in enumerate(wrap(f"{desc}  ({cnt} {tr('pdf_pg')})", FONT_B, 10, W - 2 * M - 18)[:2]):
            put(c, ln, st["y"], FONT_B, 10, TXT, x=tx)
            st["y"] -= 13
        for ln, colr in body:
            put(c, ln, st["y"], FONT, 8.5, colr, x=tx)
            st["y"] -= 11.5
        st["y"] -= 9

    # ---- notes
    notes = [tr("note"), tr("menu_note")] + ([tr("plat_note")] if PLATFORM in ("other", "squarespace", "webflow") else [])
    st["y"] -= 2
    for nt in notes:
        for ln in wrap(nt, FONT, 8, W - 2 * M):
            ensure(12)
            put(c, ln, st["y"], FONT, 8, GREY)
            st["y"] -= 11
    footer()
    c.save()
    return True


def main():
    global LANG, PLATFORM
    ap = argparse.ArgumentParser(description="SEO Audit v5 (Technical + On-Page)")
    ap.add_argument("url", nargs="?", help="website URL")
    ap.add_argument("--lang", choices=["ar", "en"], default=REPORT_LANG, help="report language: ar (Arabic) or en (English)")
    ap.add_argument("--max", type=int, default=100, help="number of sitemap pages to audit (0 = all)")
    ap.add_argument("--crawl", type=int, default=50,
                    help="extra pages to discover through internal links (0 = off)")
    ap.add_argument("--links-max", type=int, default=300, help="max links to check for broken ones (0 = off)")
    ap.add_argument("--psi", type=int, default=1, help="pages to test with Google PageSpeed (default 1 = homepage, 0 = off)")
    ap.add_argument("--psi-key", default=PSI_API_KEY, help="optional Google PageSpeed API key")
    ap.add_argument("--workers", type=int, default=5)
    ap.add_argument("--delay", type=float, default=0.3, help="delay between requests (seconds)")
    ap.add_argument("--timeout", type=int, default=15)
    ap.add_argument("--out", default="", help="Excel file name (default: seo_audit_<domain>.xlsx)")
    args, _ = ap.parse_known_args()  # works in Colab/Jupyter too
    LANG = args.lang if args.lang in ("ar", "en") else "ar"

    # In Colab/Jupyter, sys.argv contains "-f /path/kernel-xxx.json", which argparse may mistake
    # for the URL. Only accept an argument that really looks like a website.
    arg_url = args.url if looks_like_site(args.url) else ""
    url = (arg_url or SITE_URL or input(tr("p_enter"))).strip()
    if not url.startswith("http"):
        url = "https://" + url
    p = urlparse(url)
    if not p.netloc:
        print("❌ Invalid website URL / رابط الموقع غير صالح:", repr(url))
        sys.exit(1)
    base = f"{p.scheme}://{p.netloc}"
    if not args.out:  # one file per website, so reports never overwrite each other
        args.out = "seo_audit_" + re.sub(r"[^a-z0-9]+", "_", p.netloc.lower()).strip("_") + ".xlsx"

    session = make_session()
    try:  # detect the site platform from the homepage
        hp = fetch(session, base, args.timeout, retries=1)
        PLATFORM = detect_platform(hp.text, hp.url)
    except Exception:
        PLATFORM = "other"
    print(tr("p_platform", p=platform_label()))
    print(tr("p_search", b=base))
    sm_url, urls, robots_txt = discover(session, base, args.timeout)
    if not urls:
        print(tr("p_nosm"))
        sys.exit(1)
    total = len(urls)
    print(tr("p_found", sm=sm_url, n=total))
    sitemap_all = {norm(u) for u in urls}

    sampled = False
    if args.max and total > args.max:
        step = total / args.max
        urls = [urls[int(i * step)] for i in range(args.max)]  # evenly spread sample
        sampled = True
        print(tr("p_sample", m=args.max))
    sitemap_pages = set(urls)

    rp = None
    if robots_txt:
        rp = RobotFileParser()
        rp.parse(robots_txt.splitlines())

    def run(url_list):
        out = []
        with ThreadPoolExecutor(max_workers=args.workers) as ex:
            futs = {ex.submit(analyze, u, session, args.timeout, args.delay, rp): u for u in url_list}
            for n, f in enumerate(as_completed(futs), 1):
                try:
                    out.append(f.result())
                except Exception as e:  # noqa
                    out.append(({"URL": futs[f], "Status": "ERROR"}, [("FETCH_FAIL", str(e)[:150])]))
                if n % 10 == 0 or n == len(futs):
                    print(tr("p_prog", n=n, m=len(futs)))
        return out

    def known(res):
        s = set()
        for row, _ in res:
            s.add(norm(row["URL"]))
            if row.get("Final URL"):
                s.add(norm(row["Final URL"]))
        return s

    # ---- Phase 1: sitemap pages
    results = run(urls)
    analyzed = known(results)

    # ---- Phase 2: pages discovered through internal links (not in the sitemap)
    crawl_added = 0
    crawl_truncated = False
    if args.crawl:
        cand = {}
        for row, _ in results:
            for lnk, internal, _t in row.get("_links", []):
                n = norm(lnk)
                if (internal and n not in analyzed and n not in sitemap_all and n not in cand
                        and not ASSET_RE.search(lnk) and (rp is None or rp.can_fetch("*", lnk))):
                    cand[n] = lnk
        cand_urls = list(cand.values())
        crawl_truncated = len(cand_urls) > args.crawl
        cand_urls = cand_urls[:args.crawl]
        if cand_urls:
            print(tr("p_crawl", n=len(cand_urls)))
            res2 = run(cand_urls)
            for row, issues in res2:
                if row.get("Status") == 200 and "Title" in row and not row.get("Redirects"):
                    issues.append(("NOT_IN_SITEMAP", ""))
            results += res2
            analyzed |= known(res2)
            crawl_added = len(res2)

    # ---- Phase 3: broken links
    status_map = {}
    for row, _ in results:
        for key in (row["URL"], row.get("Final URL")):
            if key:
                status_map[norm(key)] = row.get("Status")
    targets = {}
    for row, _ in results:
        for lnk, internal, _t in row.get("_links", []):
            n = norm(lnk)
            if n not in status_map and n not in targets:
                targets[n] = (lnk, internal)
    ordered = sorted(targets.values(), key=lambda x: not x[1])  # internal links first
    ordered = ordered[:args.links_max] if args.links_max else []
    checked = {}
    if ordered:
        print(tr("p_links", n=len(ordered)))
        with ThreadPoolExecutor(max_workers=8) as ex:
            futs = {ex.submit(check_link, session, lnk): norm(lnk) for lnk, _i in ordered}
            for f in as_completed(futs):
                checked[futs[f]] = f.result()
    status_map.update(checked)

    broken_rows = []
    for row, issues in results:
        n_int = n_ext = 0
        for lnk, internal, text in row.get("_links", []):
            st = status_map.get(norm(lnk))
            if st is not None and is_broken(st):
                broken_rows.append([row["URL"], lnk, st, tr("int") if internal else tr("ext"), text])
                if internal:
                    n_int += 1
                else:
                    n_ext += 1
        if n_int:
            issues.append(("BROKEN_INTERNAL", tr("broken_n", n=n_int)))
        if n_ext:
            issues.append(("BROKEN_EXTERNAL", tr("broken_n", n=n_ext)))

    # ---- Phase 4: orphan pages (only when every sitemap page was audited and the crawl was complete)
    orphan_note = ""
    if sampled or crawl_truncated or not args.crawl:
        orphan_note = tr("orphan_skip")
    else:
        inbound = set()
        for row, _ in results:
            me = {norm(row["URL"]), norm(row.get("Final URL") or row["URL"])}
            for lnk, internal, _t in row.get("_links", []):
                n = norm(lnk)
                if internal and n not in me:
                    inbound.add(n)
        home = {norm(base)}
        for row, issues in results:
            if row["URL"] in sitemap_pages and row.get("Status") == 200 and "Title" in row:
                keys = {norm(row["URL"]), norm(row.get("Final URL") or row["URL"])}
                if not (keys & inbound) and not (keys & home):
                    issues.append(("ORPHAN", ""))

    # ---- Phase 5: Google PageSpeed (optional)
    speed_rows = []
    if args.psi:
        print(tr("p_psi", n=args.psi))
        idx = {r[0]["URL"]: r for r in results}
        psi_targets = [r[0]["URL"] for r in results if r[0].get("Status") == 200 and "Title" in r[0]]
        psi_targets.sort(key=lambda u: norm(u) != norm(base))  # homepage first
        for u in psi_targets[:args.psi]:
            for dev in ("mobile", "desktop"):
                label = tr("dev_" + dev)
                res, err = pagespeed(session, u, args.psi_key, dev)
                if res is None:
                    speed_rows.append([u, label] + ["n/a"] * 8)
                    print(tr("p_psi_err", d=label, u=u, e=err))
                else:
                    speed_rows.append([u, label, *res])
                    if dev == "mobile":  # score deductions use the mobile result
                        if res[0] < 50:
                            idx[u][1].append(("PSI_POOR", tr("psi_d", s=res[0])))
                        elif res[0] < 90:
                            idx[u][1].append(("PSI_AVG", tr("psi_d", s=res[0])))
                time.sleep(1)

    # ---- Duplicate Title / Description / H1
    def dups(key):
        c = Counter(r[0].get(key) for r in results if r[0].get(key))
        return {k for k, v in c.items() if v > 1}

    d_title, d_desc, d_h1 = dups("Title"), dups("Meta description"), dups("H1")
    for row, issues in results:
        if row.get("Title") in d_title:
            issues.append(("TITLE_DUP", row["Title"]))
        if row.get("Meta description") in d_desc:
            issues.append(("DESC_DUP", row["Meta description"][:100]))
        if row.get("H1") in d_h1:
            issues.append(("H1_DUP", row["H1"]))

    # ---- Scoring
    issue_rows, page_rows = [], []
    for row, issues in results:
        row.pop("_links", None)
        codes = {}
        for code, detail in issues:
            codes.setdefault(code, detail)
        penalty = sum(WEIGHTS[CATALOG[c][0]] for c in codes)
        row["Score"] = max(0, 100 - penalty)
        row["Issues"] = len(codes)
        page_rows.append(row)
        for code, detail in codes.items():
            sev, desc, fix, where = cat(code)
            issue_rows.append((SEV_ORDER[sev], code,
                               [row["URL"], SEVLBL[LANG][sev], desc, detail, fix, where, code]))
    issue_rows.sort(key=lambda x: (x[0], x[1]))
    issue_list = [x[2] for x in issue_rows]

    site_penalty = 0
    site_issue_rows = []
    if not robots_txt:
        site_penalty += 5
        site_issue_rows.append((tr("site_issue"),
                                f"{tr('no_robots')} -> {tr('no_robots_fix')} ({robots_where()})"))
    avg = sum(r["Score"] for r in page_rows) / len(page_rows)
    raw_score = max(0, round(avg - site_penalty, 1))
    # The page average can hide serious problems, so serious problems put a ceiling on the score.
    crit_pages = sum(1 for _row, _iss in results if any(CATALOG[c][0] == "Critical" for c, _d in _iss))
    crit_pct = crit_pages * 100 / len(page_rows)
    mob = next((r_ for r_ in speed_rows if r_[1] == tr("dev_mobile") and isinstance(r_[2], (int, float))), None)
    cap, reasons = 100, []
    if crit_pages:
        cap = min(cap, 89)
        if raw_score > 89:
            reasons.append(tr("cap_any"))
    if crit_pct >= 5:
        cap = min(cap, 74)
        reasons.append(tr("cap_crit", p=round(crit_pct, 1)))
    if crit_pct >= 25:
        cap = min(cap, 49)
    if mob and mob[2] < 50:
        cap = min(cap, 74)
        reasons.append(tr("cap_psi", s=mob[2]))
    if mob and mob[8] == PSI_CAT[LANG]["SLOW"]:
        cap = min(cap, 74)
        reasons.append(tr("cap_cwv"))
    site_score = min(raw_score, cap)
    cap_note = tr("cap_line", cap=cap, r=("؛ " if LANG == "ar" else "; ").join(reasons)) if site_score < raw_score and reasons else ""

    # Aggregated top issues
    agg = Counter(x[1] for x in issue_rows)
    top_rows = []
    for code, cnt in agg.items():
        sev, desc, fix, where = cat(code)
        top_rows.append((SEV_ORDER[sev], -cnt,
                         [SEVLBL[LANG][sev], desc, cnt, round(cnt * 100 / len(page_rows), 1), fix, where]))
    top_rows.sort(key=lambda x: (x[0], x[1]))
    top_list = [x[2] for x in top_rows]
    page_rows.sort(key=lambda x: x["Score"])

    sev_count = Counter(CATALOG[x[1]][0] for x in issue_rows)
    summary = [
        (tr("site"), base),
        (tr("platform"), platform_label()),
        (tr("sitemap"), sm_url),
        (tr("sm_total"), total),
        (tr("checked"), len(page_rows)),
        (tr("crawl_found"), crawl_added),
        (tr("links_checked"), len(checked)),
        (tr("broken_total"), len(broken_rows)),
        (tr("score"), site_score),
        (tr("verdict"), grade(site_score)),
        (tr("raw_score"), raw_score),
        (tr("crit_pages"), f"{crit_pages} ({crit_pct:.1f}%)"),
        (tr("coverage"), f"{len(sitemap_pages)}/{total} ({len(sitemap_pages) * 100 / total:.0f}%)"),
    ]
    for sev in ("Critical", "High", "Medium", "Low"):
        summary.append((tr("issues_of", s=SEVLBL[LANG][sev]), sev_count.get(sev, 0)))
    summary.append((tr("note_l"), tr("note")))
    summary.append((tr("note_l"), tr("menu_note")))
    if PLATFORM in ("other", "squarespace", "webflow"):
        summary.append((tr("note_l"), tr("plat_note")))
    if orphan_note:
        summary.append((tr("note_l"), orphan_note))
    if cap_note:
        summary.append((tr("note_l"), cap_note))
    for u_, n_ in Counter(r_[1] for r_ in broken_rows).most_common(2):
        if n_ >= 3:
            summary.append((tr("root_cause"), tr("root_url", n=n_, u=u_[:90])))
    for a_, n_ in Counter(r_[4] for r_ in broken_rows if r_[4]).most_common(1):
        if n_ >= 5:
            summary.append((tr("root_cause"), tr("root_anchor", a=a_, n=n_)))
    for (t_, w_), n_ in Counter((r_.get("Title"), r_.get("Words")) for r_ in page_rows if r_.get("Title")).most_common(1):
        if n_ >= 5:
            summary.append((tr("root_cause"), tr("dup_group", n=n_, t=str(t_)[:60], w=w_)))
    summary.append((tr("note_l"), tr("verify_note")))
    summary += site_issue_rows

    sh = S[LANG]["sheets"]
    with pd.ExcelWriter(args.out, engine="openpyxl") as w:
        pd.DataFrame(summary, columns=S[LANG]["sum_cols"]).to_excel(w, sheet_name=sh[0], index=False)
        pd.DataFrame(top_list, columns=S[LANG]["top_cols"]).to_excel(w, sheet_name=sh[1], index=False)
        pd.DataFrame(issue_list, columns=S[LANG]["iss_cols"]).to_excel(w, sheet_name=sh[2], index=False)
        pd.DataFrame(page_rows).rename(columns=S[LANG]["page_cols"]).to_excel(w, sheet_name=sh[3], index=False)
        pd.DataFrame(broken_rows, columns=S[LANG]["bl_cols"]).to_excel(w, sheet_name=sh[4], index=False)
        if speed_rows:
            pd.DataFrame(speed_rows, columns=S[LANG]["speed_cols"]).to_excel(w, sheet_name=sh[5], index=False)
        style_sheet(w.sheets[sh[0]], {"A": 36, "B": 90})
        style_sheet(w.sheets[sh[1]], {"A": 12, "B": 48, "C": 14, "D": 12, "E": 60, "F": 55}, sev_col=0)
        style_sheet(w.sheets[sh[2]], {"A": 50, "B": 12, "C": 45, "D": 40, "E": 55, "F": 50, "G": 16}, sev_col=1)
        style_sheet(w.sheets[sh[3]], {"A": 50, "B": 9, "C": 40})
        style_sheet(w.sheets[sh[4]], {"A": 50, "B": 50, "C": 10, "D": 12, "E": 30})
        if speed_rows:
            style_sheet(w.sheets[sh[5]], {"A": 50, "B": 12, "C": 20, "D": 12, "E": 12, "F": 10, "G": 16, "H": 16, "I": 28, "J": 22})

    print("\n" + "=" * 50)
    print(tr("p_score", s=site_score, g=grade(site_score)))
    print("   " + " | ".join(f"{SEVLBL[LANG][s]}: {sev_count.get(s, 0)}"
                              for s in ("Critical", "High", "Medium", "Low")))
    print(f"   {tr('broken_total')}: {len(broken_rows)}")
    print(tr("p_top"))
    for x in top_list[:5]:
        print(tr("p_top_row", sev=x[0], issue=x[1], n=x[2]))
    print(tr("p_file", f=args.out))

    pdf_path = re.sub(r"\.xlsx$", "", args.out) + ".pdf"
    pdf_ok = False
    try:
        pdf_ok = build_pdf(pdf_path, base, site_score, len(page_rows), len(broken_rows), sev_count,
                           top_list, speed_rows, len(issue_list), cap_note)
        if pdf_ok:
            print(tr("pdf_file", f=pdf_path))
    except Exception as e:  # the Excel report is already saved; never lose it because of the PDF
        print("PDF error:", e)

    try:  # In Google Colab the files download automatically
        from google.colab import files  # type: ignore
        files.download(args.out)
        if pdf_ok:
            files.download(pdf_path)
    except Exception:
        pass


if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
SEO + AEO/GEO Audit from sitemap.xml  |  فحص SEO تقني وجاهزية AEO/GEO من خريطة الموقع   (v4)

- Reads robots.txt and sitemap.xml (including sitemap index files)
- Detects the website platform (WordPress, Blogger, Shopify, Wix, ...) and gives fixes that fit that platform
- Checks pages one by one, discovers extra pages via internal links, finds broken links and orphan pages
- AEO/GEO checks: AI crawler access in robots.txt, llms.txt, entity schema, author/date/sources, question headings
- Optional Google PageSpeed test (--psi N) and a warning for JavaScript-rendered pages
- Two scores out of 100: technical SEO score and AEO/GEO readiness score (heuristic estimates, not official metrics)
- Exports an Excel report (issues, how to fix, where to fix) and a one-page PDF summary
- Report language: Modern Standard Arabic (--lang ar) or English (--lang en)

Install:
    pip install requests beautifulsoup4 pandas openpyxl weasyprint      (weasyprint is only needed for the PDF)

Usage:
    python seo_audit_v4.py https://example.com
    python seo_audit_v4.py https://example.com --lang en --max 200 --out report.xlsx
    python seo_audit_v4.py https://example.com --max 0 --crawl 200   (all sitemap pages + 200 discovered pages)
    python seo_audit_v4.py https://example.com --psi 5                (Google PageSpeed on 5 pages; default 1 = homepage; 0 = off)
    python seo_audit_v4.py https://example.com --pdf 0                (skip the PDF)
"""
import argparse
import gzip
import html as html_lib
import json
import math
import re
import sys
import time
from datetime import date
from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from urllib.parse import urldefrag, urljoin, urlparse
from urllib.robotparser import RobotFileParser

import pandas as pd
import requests
from bs4 import BeautifulSoup

# ======================================================================
# 👇👇👇  هنا كتبدل رابط الموقع اللي بغيتي تفحصو (الموقع الحالي: odysseedelaterre.fr)
#         CHANGE THE WEBSITE TO AUDIT HERE (the only line you need to edit)
# ======================================================================
SITE_URL = "https://odysseedelaterre.fr"

# 👇 لغة التقرير: "en" = عربية فصحى  |  "en" = English
#    REPORT LANGUAGE: "ar" = Arabic  |  "en" = English
REPORT_LANG = "en"

# 👇 (اختياري) مفتاح Google PageSpeed. خليه فارغ "" إلا ما عندكش
PSI_API_KEY = "zaSyD0XbcuK74WPFo0umfMJTEHas5ppSDm3nkAI"
# ======================================================================
# (إلا كتبتي رابط فالأمر %run هو اللي كيتطبق، وإلا ماكتبتيش كيتطبق SITE_URL)

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/124.0 Safari/537.36 SEOAuditBot/1.0"
}

# Points deducted from a page's score for each issue, by severity
WEIGHTS = {"Critical": 25, "High": 12, "Medium": 5, "Low": 2, "Info": 0}
SEV_ORDER = {"Critical": 0, "High": 1, "Medium": 2, "Low": 3, "Info": 4}
SEVLBL = {
    "ar": {"Critical": "حرج", "High": "عالٍ", "Medium": "متوسط", "Low": "منخفض", "Info": "ملاحظة"},
    "en": {"Critical": "Critical", "High": "High", "Medium": "Medium", "Low": "Low", "Info": "Info"},
}

# Points deducted for AEO/GEO issues (fewer checks than SEO, so each one weighs more)
AI_WEIGHTS = {"Critical": 30, "High": 15, "Medium": 8, "Low": 4, "Info": 0}

LANG = "ar"
PLATFORM = "other"  # detected website platform, set in main()


def cat_of(code):
    """Issue family: AEO (answer engines), GEO (generative engines) or SEO (technical / on-page)."""
    return "AEO" if code.startswith("AEO_") else "GEO" if code.startswith("GEO_") else "SEO"


def is_ai(code):
    return code.startswith(("AEO_", "GEO_"))


# ------------------------------------------------------------------ النصوص / UI strings
S = {
    "ar": {
        "chars": "{n} حرفًا", "words": "{n} كلمة", "sec": "{n} ثانية",
        "robots_dis": "ممنوعة بقاعدة Disallow في robots.txt",
        "chain": "{n} عمليات إعادة توجيه ← {u}",
        "redir": "يعيد التوجيه إلى {u}",
        "img": "{a} من أصل {b} صورة بلا نص بديل",
        "mixed": "{n} ملفًا يُحمَّل عبر http",
        "yes": "نعم", "no": "لا",
        "g90": "ممتاز", "g75": "جيد", "g50": "متوسط (يحتاج إلى تحسينات)", "g0": "ضعيف (مشكلات كبيرة)",
        "sheets": ["الملخص", "أهم الإصلاحات", "كل المشكلات", "الصفحات", "الروابط المكسورة", "السرعة", "AEO-GEO"],
        "status_n": "رمز الاستجابة {n}",
        "page_cols": {"URL": "الرابط", "Status": "الحالة", "Final URL": "الرابط النهائي", "Redirects": "عدد إعادات التوجيه",
                      "Response (s)": "زمن الاستجابة (ثانية)", "Size (KB)": "الحجم (كيلوبايت)", "Title": "عنوان الصفحة",
                      "Title len": "طول العنوان", "Meta description": "وصف الميتا", "Desc len": "طول الوصف",
                      "H1": "العنوان الرئيسي H1", "H1 count": "عدد H1", "Canonical": "Canonical", "Noindex": "Noindex",
                      "Images": "عدد الصور", "Images no alt": "صور بلا نص بديل", "Internal links": "روابط داخلية",
                      "External links": "روابط خارجية", "Words": "عدد الكلمات", "Schema types": "أنواع Schema",
                      "Question headings": "عناوين على شكل أسئلة", "Lists & tables": "قوائم وجداول",
                      "Author": "الكاتب", "Date": "التاريخ", "Score": "تقييم SEO (من 100)",
                      "AI score": "تقييم AEO/GEO (من 100)", "Issues": "عدد المشكلات"},
        "bl_cols": ["الصفحة المصدر", "الرابط المكسور", "الحالة", "النوع", "نص الرابط"],
        "int": "داخلي", "ext": "خارجي",
        "speed_cols": ["الرابط", "الجهاز", "الأداء (من 100)", "FCP (ثانية)", "LCP (ثانية)", "CLS", "TBT (مللي ثانية)",
                       "TTFB (مللي ثانية)", "تقييم المستخدمين الحقيقيين (28 يومًا)", "INP للمستخدمين (مللي ثانية)"],
        "broken_n": "{n} رابط مكسور",
        "js_d": "نص قليل مع عناصر JavaScript كثيرة",
        "psi_d": "الأداء {s}/100",
        "crawl_found": "صفحات أُضيفت للفحص عبر الروابط الداخلية (خارج الخريطة)",
        "links_checked": "عدد الروابط التي جرى فحصها",
        "broken_total": "عدد الروابط المكسورة",
        "orphan_skip": "تم تخطي كشف الصفحات اليتيمة لأن الفحص لم يشمل كل الصفحات. استخدم --max 0 مع --crawl أكبر.",
        "p_crawl": "🕸️ تم اكتشاف {n} صفحة إضافية عبر الروابط الداخلية، جارٍ فحصها ...",
        "p_links": "🔗 جارٍ فحص {n} رابط بحثًا عن الروابط المكسورة ...",
        "p_psi": "⚡ جارٍ فحص السرعة عبر Google PageSpeed ({n} صفحة، على الجوال والكمبيوتر، قد يستغرق عدة دقائق) ...",
        "dev_mobile": "جوال", "dev_desktop": "كمبيوتر",
        "p_psi_err": "⚠️ تعذّر فحص PageSpeed ({d}) للرابط {u} ({e}). إن كان الخطأ 429 فالحصة المجانية استُنفدت: جرّب لاحقًا أو استعمل --psi-key، أو افحص الرابط يدويًا في pagespeed.web.dev",
        "sum_cols": ["البند", "القيمة"],
        "top_cols": ["النوع", "الخطورة", "المشكلة", "عدد الصفحات المتأثرة", "% من الصفحات", "طريقة الإصلاح", "مكان الإصلاح"],
        "iss_cols": ["الرابط", "النوع", "الخطورة", "المشكلة", "التفاصيل", "طريقة الإصلاح", "مكان الإصلاح", "الرمز"],
        "site": "الموقع", "sitemap": "خريطة الموقع (Sitemap)", "sm_total": "عدد الروابط في الخريطة",
        "checked": "عدد الصفحات المفحوصة", "score": "تقييم SEO التقني (من 100)", "verdict": "حكم SEO التقني",
        "issues_of": "عدد المشكلات ({s})",
        "note_l": "ملاحظة",
        "note": "التقييم من 100: كلما ارتفع الرقم كان الموقع أفضل. هذا تقييم تقديري من هذه الأداة وليس مقياسًا رسميًا من Google.",
        "site_issue": "مشكلة على مستوى الموقع",
        "no_robots": "لا يوجد ملف robots.txt",
        "no_robots_fix": "أنشئ ملف robots.txt في جذر الموقع وأضف إليه رابط خريطة الموقع.",
        "no_robots_where": "جذر الموقع: /robots.txt",
        "p_enter": "🔗 أدخل رابط الموقع (مثال: https://example.com): ",
        "p_search": "⏳ جارٍ البحث عن خريطة الموقع لـ {b} ...",
        "p_nosm": "❌ لم يتم العثور على خريطة موقع (robots.txt أو /sitemap.xml أو /sitemap_index.xml أو /wp-sitemap.xml).",
        "p_found": "✅ خريطة الموقع: {sm} | {n} رابط",
        "p_sample": "ℹ️ سيتم فحص {m} صفحة (عينة موزعة على الموقع). استخدم --max 0 لفحص الكل.",
        "p_prog": "   تم فحص {n}/{m} صفحة",
        "p_score": "🏁 تقييم SEO التقني: {s}/100 ← {g}",
        "p_top": "🔧 أهم المشكلات التي يجب إصلاحها:",
        "p_top_row": "   [{sev}] {issue} ({n} صفحة)",
        "p_file": "📄 التقرير الكامل: {f}",
        # ---- v4: platform, AEO/GEO, PDF
        "platform": "المنصة (CMS)", "platform_note_l": "ملاحظة حسب المنصة",
        "p_platform": "🧩 المنصة المكتشفة: {p}",
        "ai_score": "جاهزية AEO/GEO (من 100)", "ai_verdict": "حكم AEO/GEO",
        "geo_note": "جاهزية AEO/GEO تقيس الاستعداد التقني فقط (وصول الزواحف، البيانات المنظمة، بنية المحتوى، إشارات الثقة). هي لا تتنبأ بالظهور أو الاستشهاد داخل إجابات الذكاء الاصطناعي.",
        "p_ai": "🤖 جارٍ فحص جاهزية AEO/GEO (زواحف الذكاء الاصطناعي، llms.txt، بيانات الكيان) ...",
        "p_ai_score": "🤖 جاهزية AEO/GEO: {s}/100 ← {g}",
        "site_wide": "على مستوى الموقع",
        "geo_cols": ["الفحص", "الحالة", "التفاصيل"],
        "st_ok": "سليم", "st_bad": "مشكلة", "st_warn": "تنبيه", "st_info": "معلومة",
        "bot_row": "زاحف ذكاء اصطناعي: {b}",
        "bt": {"search": "بحث وإجابات", "user": "بطلب المستخدم", "train": "تدريب النماذج"},
        "bot_allowed": "مسموح", "bot_blocked": "محظور",
        "llms_l": "ملف llms.txt (اقتراح اختياري غير رسمي)", "llms_yes": "موجود", "llms_no": "غير موجود",
        "entity_l": "بيانات الكيان في الصفحة الرئيسية (Organization / Person / LocalBusiness)",
        "entity_no": "لم يُعثر على بيانات منظمة للكيان", "sameas_l": "روابط sameAs إلى الحسابات الرسمية",
        "about_l": "صفحة «من نحن» أو «اتصل بنا» مرتبطة من الصفحة الرئيسية",
        "found": "موجود", "missing": "غير موجود",
        "stat_q": "صفحات المحتوى التي فيها عناوين على شكل أسئلة", "stat_l": "صفحات المحتوى التي فيها قوائم أو جداول",
        "stat_a": "صفحات المقالات التي فيها كاتب ظاهر", "stat_d": "صفحات المقالات التي فيها تاريخ نشر أو تحديث",
        "n_of": "{a} من {b}",
        "p_psi_hint": "💡 الخطأ 403 أو 400 يعني غالبًا أن PageSpeed Insights API غير مفعّل أو أن المفتاح مقيّد بواجهة أخرى: افتح Google Cloud ← API et services ← Identifiants ← المفتاح ← Restrictions relatives aux API.",
        "p_pdf": "📑 جارٍ إنشاء ملخص PDF من صفحة واحدة ...",
        "p_pdf_ok": "📑 ملخص PDF: {f}",
        "p_pdf_fail": "⚠️ تعذّر إنشاء PDF ({e}). حُفظت نسخة HTML بدلًا منه: {h} (افتحها واضغط Ctrl+P ثم احفظها PDF).",
        "rtl": True,
    },
    "en": {
        "chars": "{n} chars", "words": "{n} words", "sec": "{n} s",
        "robots_dis": "Disallowed by robots.txt",
        "chain": "{n} redirects -> {u}",
        "redir": "Redirects to {u}",
        "img": "{a} of {b} images have no alt text",
        "mixed": "{n} resources loaded over http",
        "yes": "Yes", "no": "No",
        "g90": "Excellent", "g75": "Good", "g50": "Average (needs work)", "g0": "Poor (major issues)",
        "sheets": ["Summary", "Top Fixes", "All Issues", "Pages", "Broken Links", "Speed", "AEO-GEO"],
        "status_n": "Status {n}",
        "page_cols": {"Score": "SEO score", "AI score": "AEO/GEO score"},
        "bl_cols": ["Source page", "Broken link", "Status", "Type", "Anchor text"],
        "int": "Internal", "ext": "External",
        "speed_cols": ["URL", "Device", "Performance (/100)", "FCP (s)", "LCP (s)", "CLS", "TBT (ms)",
                       "TTFB (ms)", "Real-user rating (28 days)", "Real-user INP (ms)"],
        "broken_n": "{n} broken links",
        "js_d": "Very little text but many JavaScript elements",
        "psi_d": "Performance {s}/100",
        "crawl_found": "Pages added through internal links (not in sitemap)",
        "links_checked": "Links checked",
        "broken_total": "Broken links found",
        "orphan_skip": "Orphan-page detection skipped because not all pages were audited. Use --max 0 with a larger --crawl.",
        "p_crawl": "🕸️ Found {n} extra pages through internal links, auditing them ...",
        "p_links": "🔗 Checking {n} links for broken ones ...",
        "p_psi": "⚡ Running Google PageSpeed ({n} pages, mobile and desktop, may take several minutes) ...",
        "dev_mobile": "Mobile", "dev_desktop": "Desktop",
        "p_psi_err": "⚠️ PageSpeed failed ({d}) for {u} ({e}). If it is 429 the free quota is used up: try later or use --psi-key, or test the URL manually at pagespeed.web.dev",
        "sum_cols": ["Item", "Value"],
        "top_cols": ["Type", "Severity", "Issue", "Pages affected", "% of pages", "How to fix", "Where to fix"],
        "iss_cols": ["URL", "Type", "Severity", "Issue", "Detail", "How to fix", "Where to fix", "Code"],
        "site": "Website", "sitemap": "Sitemap", "sm_total": "URLs in sitemap",
        "checked": "Pages audited", "score": "Technical SEO score (out of 100)", "verdict": "Technical SEO verdict",
        "issues_of": "Issues ({s})",
        "note_l": "Note",
        "note": "Score is out of 100: the higher the better. It is this tool's own estimate, not an official Google metric.",
        "site_issue": "Site-wide issue",
        "no_robots": "No robots.txt file",
        "no_robots_fix": "Create a robots.txt file in the site root and add the sitemap URL to it.",
        "no_robots_where": "Site root: /robots.txt",
        "p_enter": "🔗 Enter the website URL (e.g. https://example.com): ",
        "p_search": "⏳ Looking for the sitemap of {b} ...",
        "p_nosm": "❌ No sitemap found (robots.txt, /sitemap.xml, /sitemap_index.xml, /wp-sitemap.xml).",
        "p_found": "✅ Sitemap: {sm} | {n} URLs",
        "p_sample": "ℹ️ Auditing {m} pages (an evenly spread sample). Use --max 0 to audit all.",
        "p_prog": "   {n}/{m} pages audited",
        "p_score": "🏁 Technical SEO score: {s}/100 -> {g}",
        "p_top": "🔧 Top issues to fix:",
        "p_top_row": "   [{sev}] {issue} ({n} pages)",
        "p_file": "📄 Full report: {f}",
        # ---- v4: platform, AEO/GEO, PDF
        "platform": "Platform (CMS)", "platform_note_l": "Platform note",
        "p_platform": "🧩 Detected platform: {p}",
        "ai_score": "AEO/GEO readiness (out of 100)", "ai_verdict": "AEO/GEO verdict",
        "geo_note": "AEO/GEO readiness only measures technical preparation (crawler access, structured data, content structure, trust signals). It does not predict whether you will appear or be cited in AI answers.",
        "p_ai": "🤖 Checking AEO/GEO readiness (AI crawlers, llms.txt, entity data) ...",
        "p_ai_score": "🤖 AEO/GEO readiness: {s}/100 -> {g}",
        "site_wide": "Site-wide",
        "geo_cols": ["Check", "Status", "Details"],
        "st_ok": "OK", "st_bad": "Problem", "st_warn": "Warning", "st_info": "Info",
        "bot_row": "AI crawler: {b}",
        "bt": {"search": "search & answers", "user": "user-triggered", "train": "model training"},
        "bot_allowed": "Allowed", "bot_blocked": "Blocked",
        "llms_l": "llms.txt file (optional, unofficial proposal)", "llms_yes": "Found", "llms_no": "Not found",
        "entity_l": "Entity data on the homepage (Organization / Person / LocalBusiness)",
        "entity_no": "No entity structured data found", "sameas_l": "sameAs links to official profiles",
        "about_l": "About or Contact page linked from the homepage",
        "found": "Found", "missing": "Missing",
        "stat_q": "Content pages with question-style headings", "stat_l": "Content pages with lists or tables",
        "stat_a": "Article pages with a visible author", "stat_d": "Article pages with a publish or update date",
        "n_of": "{a} of {b}",
        "p_psi_hint": "💡 Error 403 or 400 usually means the PageSpeed Insights API is not enabled or the key is restricted to another API: open Google Cloud > APIs & Services > Credentials > your key > API restrictions.",
        "p_pdf": "📑 Building the one-page PDF summary ...",
        "p_pdf_ok": "📑 PDF summary: {f}",
        "p_pdf_fail": "⚠️ Could not build the PDF ({e}). An HTML copy was saved instead: {h} (open it, press Ctrl+P and save as PDF).",
        "rtl": False,
    },
}


def tr(key, **kw):
    return S[LANG][key].format(**kw)


# ------------------------------------------------------------------ كتالوج المشكلات / Issue catalog
# code: (severity, (desc_ar, desc_en), (fix_ar, fix_en), (where_ar, where_en))
CATALOG = {
    "FETCH_FAIL": ("Critical",
        ("تعذّر تحميل الصفحة (انتهاء المهلة أو خطأ في الاتصال)", "Page failed to load (timeout or connection error)"),
        ("تأكد من أن الخادم يعمل بسرعة جيدة، وأن جدار الحماية أو Cloudflare لا يحجب الزوار.",
         "Make sure the server is up and fast, and that a firewall or Cloudflare is not blocking visitors."),
        ("الاستضافة / الخادم / إعدادات جدار الحماية", "Hosting / server / firewall settings")),
    "HTTP_4XX": ("Critical",
        ("الصفحة تعرض خطأ 4xx (مثل 404: الصفحة غير موجودة)", "Page returns a 4xx error (e.g. 404 Not Found)"),
        ("أزل الرابط من خريطة الموقع، أو أنشئ إعادة توجيه 301 إلى الصفحة البديلة، أو أعد نشر الصفحة.",
         "Remove the URL from the sitemap, add a 301 redirect to the replacement page, or restore the page."),
        ("ملف sitemap.xml وإعدادات إعادة التوجيه (.htaccess / Cloudflare / إضافة Redirection)",
         "sitemap.xml and redirect settings (.htaccess / Cloudflare / Redirection plugin)")),
    "HTTP_5XX": ("Critical",
        ("خطأ في الخادم (5xx)", "Server error (5xx)"),
        ("راجع سجلات الخادم (logs)؛ غالبًا تكون المشكلة في PHP أو قاعدة البيانات أو إضافة. تواصل مع الاستضافة إن لم تجد السبب.",
         "Check the server logs; it is usually a PHP, database or plugin issue. Contact your host if you cannot find the cause."),
        ("الخادم / الاستضافة / السجلات", "Server / hosting / logs")),
    "NOINDEX": ("Critical",
        ("الصفحة تحتوي على noindex (تمنع Google من فهرستها) رغم وجودها في خريطة الموقع",
         "Page has noindex (blocks Google from indexing it) yet is listed in the sitemap"),
        ("إن أردت ظهورها في Google فأزل noindex، وإن لم ترد ذلك فأزلها من خريطة الموقع.",
         "Remove noindex if the page should appear in Google; otherwise remove it from the sitemap."),
        ("وسم robots في <head> / ترويسة X-Robots-Tag / إعدادات Yoast أو RankMath",
         "robots meta tag in <head> / X-Robots-Tag header / Yoast or RankMath settings")),
    "ROBOTS_BLOCKED": ("Critical",
        ("الصفحة محظورة في ملف robots.txt", "Page is blocked by robots.txt"),
        ("عدّل قاعدة Disallow التي تحجب هذا الرابط.", "Edit the Disallow rule that blocks this URL."),
        ("ملف robots.txt في جذر الموقع", "robots.txt in the site root")),
    "NOT_HTTPS": ("High",
        ("الصفحة لا تعمل عبر HTTPS", "Page is not served over HTTPS"),
        ("فعّل شهادة SSL وأضف إعادة توجيه 301 من http إلى https لكامل الموقع.",
         "Install an SSL certificate and 301-redirect http to https site-wide."),
        ("الاستضافة (SSL) + .htaccess / Cloudflare / إعدادات الموقع", "Hosting (SSL) + .htaccess / Cloudflare / site settings")),
    "MIXED_CONTENT": ("High",
        ("صفحة HTTPS تحمّل ملفات (صور أو سكربتات) عبر http", "HTTPS page loads resources (images/scripts) over http"),
        ("استبدل كل روابط http:// بـ https:// في الصور وملفات CSS وJS.",
         "Replace all http:// links with https:// for images, CSS and JS."),
        ("كود القالب / المحتوى / إضافة مثل Really Simple SSL", "Theme code / content / a plugin such as Really Simple SSL")),
    "REDIRECT": ("Medium",
        ("الرابط في خريطة الموقع يعيد التوجيه إلى رابط آخر", "URL in the sitemap redirects elsewhere"),
        ("ضع الرابط النهائي (الذي يعطي حالة 200) في خريطة الموقع بدل الرابط القديم.",
         "Put the final URL (status 200) in the sitemap instead of the old one."),
        ("ملف sitemap.xml", "sitemap.xml")),
    "REDIRECT_CHAIN": ("High",
        ("سلسلة إعادة توجيه (أكثر من إعادة توجيه واحدة)", "Redirect chain (more than one redirect)"),
        ("اجعلها إعادة توجيه واحدة مباشرة من الرابط القديم إلى الرابط النهائي.",
         "Use a single direct redirect from the old URL to the final URL."),
        ("إعدادات إعادة التوجيه (.htaccess / Cloudflare / إضافة)", "Redirect settings (.htaccess / Cloudflare / plugin)")),
    "SLOW": ("Medium",
        ("الصفحة بطيئة (زمن الاستجابة أكثر من 1.5 ثانية)", "Slow page (response time over 1.5 seconds)"),
        ("فعّل التخزين المؤقت (Cache) وضغط الصور وشبكة CDN، وحسّن الاستضافة. قِس الأداء عبر PageSpeed Insights.",
         "Enable caching, compress images, use a CDN and upgrade hosting. Measure with PageSpeed Insights."),
        ("الاستضافة + إضافة التخزين المؤقت (WP Rocket / LiteSpeed) + الصور",
         "Hosting + cache plugin (WP Rocket / LiteSpeed) + images")),
    "BIG_PAGE": ("Low",
        ("حجم الصفحة كبير (أكثر من 1.5 ميغابايت من HTML)", "Large page (over 1.5 MB of HTML)"),
        ("قلّل الكود الزائد وفعّل ضغط gzip/brotli.", "Reduce unnecessary markup and enable gzip/brotli compression."),
        ("القالب / إعدادات الخادم", "Theme / server settings")),
    "TITLE_MISSING": ("High",
        ("لا يوجد عنوان للصفحة (Title)", "Missing title tag"),
        ("أضف وسم <title> فريدًا يتضمن الكلمة المفتاحية (30-60 حرفًا).",
         "Add a unique <title> containing the target keyword (30-60 characters)."),
        ("وسم <head> للصفحة / حقل عنوان SEO في Yoast أو RankMath", "Page <head> / SEO Title field in Yoast or RankMath")),
    "TITLE_SHORT": ("Medium",
        ("عنوان الصفحة قصير (أقل من 30 حرفًا)", "Title too short (under 30 characters)"),
        ("أطِل العنوان وأضف الكلمة المفتاحية وما يميّز الصفحة (30-60 حرفًا).",
         "Lengthen the title and include the keyword and a clear benefit (30-60 characters)."),
        ("حقل عنوان SEO في Yoast / RankMath أو وسم <title>", "SEO Title field in Yoast / RankMath or the <title> tag")),
    "TITLE_LONG": ("Medium",
        ("عنوان الصفحة طويل (أكثر من 60 حرفًا) وسيُقتطع في نتائج Google",
         "Title too long (over 60 characters) and will be truncated in Google"),
        ("اختصر العنوان وضع الكلمة المهمة في بدايته.", "Shorten the title and put the important keyword first."),
        ("حقل عنوان SEO في Yoast / RankMath أو وسم <title>", "SEO Title field in Yoast / RankMath or the <title> tag")),
    "TITLE_DUP": ("Medium",
        ("العنوان نفسه مستخدم في صفحات أخرى", "Same title used on other pages"),
        ("يجب أن يكون لكل صفحة عنوان خاص بها.", "Give every page its own unique title."),
        ("حقل عنوان SEO في كل صفحة", "SEO Title field on each page")),
    "DESC_MISSING": ("Medium",
        ("لا يوجد وصف ميتا (Meta Description)", "Missing meta description"),
        ("أضف وصفًا جذابًا (70-160 حرفًا) يتضمن الكلمة المفتاحية.",
         "Add a compelling description (70-160 characters) with the target keyword."),
        ("حقل الوصف في Yoast / RankMath أو وسم <meta name='description'> في <head>",
         "Description field in Yoast / RankMath or the <meta name='description'> tag in <head>")),
    "DESC_SHORT": ("Low",
        ("وصف الميتا قصير (أقل من 70 حرفًا)", "Meta description too short (under 70 characters)"),
        ("أطِل الوصف قليلًا (70-160 حرفًا).", "Lengthen it a little (70-160 characters)."),
        ("حقل وصف الميتا", "Meta description field")),
    "DESC_LONG": ("Low",
        ("وصف الميتا طويل (أكثر من 160 حرفًا)", "Meta description too long (over 160 characters)"),
        ("اختصر الوصف حتى لا يُقتطع.", "Shorten it so it is not truncated."),
        ("حقل وصف الميتا", "Meta description field")),
    "DESC_DUP": ("Medium",
        ("وصف الميتا نفسه مستخدم في صفحات أخرى", "Same meta description used on other pages"),
        ("اكتب وصفًا خاصًا بكل صفحة.", "Write a unique description for each page."),
        ("حقل وصف الميتا في كل صفحة", "Meta description field on each page")),
    "H1_MISSING": ("High",
        ("لا يوجد عنوان رئيسي H1", "Missing H1 heading"),
        ("أضف H1 واحدًا في الصفحة يتضمن الكلمة المفتاحية الرئيسية.", "Add a single H1 containing the main keyword."),
        ("محتوى الصفحة / القالب (عنوان الصفحة يجب أن يكون H1)", "Page content / theme (the page title should be an H1)")),
    "H1_MULTI": ("Medium",
        ("أكثر من H1 في الصفحة نفسها", "More than one H1 on the page"),
        ("اترك H1 واحدًا فقط وحوّل البقية إلى H2 أو H3.", "Keep one H1 and change the rest to H2/H3."),
        ("محتوى الصفحة / القالب / أداة بناء الصفحات", "Page content / theme / page builder")),
    "H1_DUP": ("Low",
        ("العنوان الرئيسي H1 نفسه مستخدم في صفحات أخرى", "Same H1 used on other pages"),
        ("اجعل H1 فريدًا لكل صفحة.", "Make each page's H1 unique."),
        ("محتوى الصفحة", "Page content")),
    "CANON_MISSING": ("Medium",
        ("لا يوجد رابط Canonical", "Missing canonical tag"),
        ("أضف <link rel='canonical'> يشير إلى الرابط الأصلي للصفحة.",
         "Add <link rel='canonical'> pointing to the page's preferred URL."),
        ("<head> / Yoast / RankMath (متقدم ← Canonical URL)", "<head> / Yoast / RankMath (Advanced > Canonical URL)")),
    "CANON_OTHER": ("Medium",
        ("رابط Canonical يشير إلى صفحة أخرى", "Canonical points to a different URL"),
        ("تأكد من أن ذلك مقصود، وإلا فاجعل الـ canonical يشير إلى الصفحة نفسها.",
         "Make sure this is intentional; otherwise point the canonical to the page itself."),
        ("<head> / Yoast / RankMath (متقدم ← Canonical URL)", "<head> / Yoast / RankMath (Advanced > Canonical URL)")),
    "IMG_ALT": ("Medium",
        ("صور بلا نص بديل (alt)", "Images missing alt text"),
        ("أضف نصًا بديلًا وصفيًا لكل صورة (يفيد SEO وإمكانية الوصول).",
         "Add descriptive alt text to every image (helps SEO and accessibility)."),
        ("مكتبة الوسائط أو الوسم <img alt=''>", "Media Library or the <img alt=''> attribute")),
    "THIN": ("Medium",
        ("محتوى قليل (أقل من 300 كلمة)", "Thin content (under 300 words)"),
        ("أضف محتوى مفيدًا وأصليًا يجيب عن نية الباحث، وتجنب الصفحات شبه الفارغة.",
         "Add useful, original content that answers search intent; avoid near-empty pages."),
        ("محتوى الصفحة", "Page content")),
    "NO_VIEWPORT": ("High",
        ("لا يوجد وسم viewport (الموقع غير مهيأ للجوال)", "Missing viewport tag (not mobile-friendly)"),
        ("أضف <meta name='viewport' content='width=device-width, initial-scale=1'>.",
         "Add <meta name='viewport' content='width=device-width, initial-scale=1'>."),
        ("<head> / القالب (header.php)", "<head> / theme (header.php)")),
    "NO_LANG": ("Low",
        ("لا يوجد وسم lang في <html>", "Missing lang attribute on <html>"),
        ("أضف lang مثل <html lang='ar'> أو 'fr' أو 'en'.", "Add lang, e.g. <html lang='en'>."),
        ("القالب (header.php) / إعدادات اللغة", "Theme (header.php) / language settings")),
    "NO_OG": ("Low",
        ("لا توجد وسوم Open Graph (معاينة المشاركة في فيسبوك وواتساب)",
         "Missing Open Graph tags (share previews on Facebook/WhatsApp)"),
        ("أضف og:title وog:description وog:image.", "Add og:title, og:description and og:image."),
        ("Yoast / RankMath (التواصل الاجتماعي) أو <head>", "Yoast / RankMath (Social) or <head>")),
    "NO_SCHEMA": ("Low",
        ("لا توجد بيانات منظمة (Schema)", "No structured data (Schema)"),
        ("أضف Schema بصيغة JSON-LD مناسبة (Article / Product / LocalBusiness / FAQ ...).",
         "Add suitable JSON-LD Schema (Article / Product / LocalBusiness / FAQ ...)."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    "URL_LONG": ("Low",
        ("الرابط طويل (أكثر من 115 حرفًا)", "URL too long (over 115 characters)"),
        ("اختصر الرابط واكتفِ بكلمات مفتاحية قليلة.", "Shorten the URL and keep only a few keywords."),
        ("الرابط الدائم (Slug) للصفحة", "Page slug / permalink")),
    "URL_STYLE": ("Low",
        ("الرابط يحتوي على أحرف كبيرة أو شرطة سفلية أو معاملات (parameters)",
         "URL contains uppercase letters, underscores or parameters"),
        ("استخدم أحرفًا صغيرة وشرطة (-) بدل الشرطة السفلية، وتجنب المعاملات.",
         "Use lowercase and hyphens instead of underscores; avoid parameters."),
        ("الرابط الدائم / إعدادات الروابط", "Slug / permalink settings")),
    "NO_INTERNAL_LINKS": ("Medium",
        ("لا توجد روابط داخلية في الصفحة", "No internal links on the page"),
        ("أضف روابط داخلية إلى صفحات مهمة أخرى في الموقع.", "Add internal links to other important pages."),
        ("محتوى الصفحة / القائمة", "Page content / menu")),
    "BROKEN_INTERNAL": ("High",
        ("روابط داخلية مكسورة (تؤدي إلى صفحات غير موجودة أو بها خطأ)", "Broken internal links (point to missing or erroring pages)"),
        ("عدّل الرابط ليشير إلى صفحة صحيحة، أو أنشئ إعادة توجيه 301، أو أزل الرابط. التفاصيل في ورقة «الروابط المكسورة».",
         "Point the link to a valid page, add a 301 redirect, or remove the link. See the 'Broken Links' sheet for details."),
        ("محتوى الصفحة / القائمة / القالب", "Page content / menu / theme")),
    "BROKEN_EXTERNAL": ("Medium",
        ("روابط خارجية مكسورة", "Broken external links"),
        ("استبدل الرابط بمصدر يعمل أو أزله. التفاصيل في ورقة «الروابط المكسورة».",
         "Replace the link with a working source or remove it. See the 'Broken Links' sheet for details."),
        ("محتوى الصفحة", "Page content")),
    "NOT_IN_SITEMAP": ("Low",
        ("صفحة موجودة ويمكن الوصول إليها لكنها غير مدرجة في خريطة الموقع", "Live page that is not listed in the sitemap"),
        ("أضفها إلى خريطة الموقع إن كانت صفحة مهمة، أو أضف لها noindex إن لم تكن كذلك.",
         "Add it to the sitemap if it is important, or noindex it if it is not."),
        ("sitemap.xml / إعدادات الخرائط في Yoast أو RankMath", "sitemap.xml / Sitemap settings in Yoast or RankMath")),
    "ORPHAN": ("Medium",
        ("صفحة يتيمة: لا توجد أي صفحة مفحوصة تربط إليها داخليًا", "Orphan page: no audited page links to it"),
        ("أضف روابط داخلية إليها من صفحات ذات صلة أو من القائمة الرئيسية.",
         "Add internal links to it from related pages or the main menu."),
        ("محتوى الصفحات / القائمة", "Page content / menu")),
    "HEADING_SKIP": ("Low",
        ("ترتيب العناوين غير منطقي (قفز مثلًا من H1 إلى H3)", "Illogical heading order (e.g. jumping from H1 to H3)"),
        ("رتّب العناوين H1 ثم H2 ثم H3 دون قفز بين المستويات.", "Order headings H1, H2, H3 without skipping levels."),
        ("محتوى الصفحة / القالب", "Page content / theme")),
    "JS_SUSPECT": ("Info",
        ("الصفحة تبدو معتمدة على JavaScript لعرض المحتوى، وقد تكون بعض نتائج الفحص غير دقيقة",
         "Page seems to rely on JavaScript to render content, so some results may be inaccurate"),
        ("تحقق من الصفحة يدويًا في المتصفح (قارن View Source مع Inspect) قبل الاعتماد على نتائجها.",
         "Check the page manually in the browser (compare View Source with Inspect) before trusting its results."),
        ("-", "-")),
    "PSI_POOR": ("Medium",
        ("أداء ضعيف على الجوال حسب Google PageSpeed (أقل من 50)", "Poor mobile performance per Google PageSpeed (below 50)"),
        ("اضغط الصور، فعّل التخزين المؤقت وCDN، قلّل ملفات JS وCSS غير الضرورية، وراجع تقرير PageSpeed للتفاصيل.",
         "Compress images, enable caching and a CDN, cut unneeded JS/CSS, and review the PageSpeed report for details."),
        ("الصور / القالب / الإضافات / الاستضافة", "Images / theme / plugins / hosting")),
    "PSI_AVG": ("Low",
        ("أداء متوسط على الجوال حسب Google PageSpeed (50-89)", "Average mobile performance per Google PageSpeed (50-89)"),
        ("حسّن الصور والتخزين المؤقت وقلّل الملفات الثقيلة.", "Improve images and caching, and reduce heavy files."),
        ("الصور / القالب / الإضافات / الاستضافة", "Images / theme / plugins / hosting")),

    # ---------------- AEO (answer engines) : page level
    "AEO_NO_QUESTIONS": ("Low",
        ("لا توجد عناوين فرعية على شكل أسئلة (تساعد محركات الإجابة AEO)", "No question-style sub-headings (they help answer engines / AEO)"),
        ("أضف عناوين H2 أو H3 على شكل أسئلة يبحث عنها الناس، وضع تحت كل سؤال إجابة مباشرة موجزة (نحو 40-60 كلمة) ثم التفاصيل.",
         "Add H2/H3 headings phrased as questions people really ask, and put a short direct answer (about 40-60 words) right under each, then the details."),
        ("محتوى الصفحة (العناوين الفرعية)", "Page content (sub-headings)")),
    "AEO_NO_STRUCTURE": ("Low",
        ("لا توجد قوائم أو جداول في المحتوى (تسهّل على الذكاء الاصطناعي استخراج المعلومة)",
         "No lists or tables in the content (they make facts easier for AI to extract)"),
        ("حوّل الخطوات والمقارنات والمعلومات المتعددة إلى قوائم نقطية أو مرقمة أو جداول.",
         "Turn steps, comparisons and multi-part facts into bullet or numbered lists, or tables."),
        ("محتوى الصفحة", "Page content")),
    # ---------------- GEO (generative engines) : page level
    "GEO_NO_AUTHOR": ("Medium",
        ("لا توجد إشارة واضحة إلى كاتب المقال (إشارة ثقة E-E-A-T)", "No clear author signal on the article (E-E-A-T trust signal)"),
        ("أضف اسم الكاتب مع نبذة قصيرة وصفحة تعريفية به، وأضف author في بيانات Article المنظمة.",
         "Add the author's name with a short bio and an author page, and add author to the Article structured data."),
        ("قالب المقال / ملف الكاتب / Schema", "Article template / author profile / Schema")),
    "GEO_NO_DATE": ("Low",
        ("لا يوجد تاريخ نشر أو تحديث ظاهر أو في البيانات المنظمة", "No visible or structured publish / update date"),
        ("أظهر تاريخ النشر وآخر تحديث في الصفحة وأضف datePublished وdateModified في Schema.",
         "Show the publish and last-updated dates on the page and add datePublished and dateModified to Schema."),
        ("قالب المقال / Schema", "Article template / Schema")),
    "GEO_NO_SOURCES": ("Low",
        ("مقال طويل بلا أي رابط خارجي إلى مصادر", "Long article with no external links to sources"),
        ("استشهد بمصادر موثوقة (دراسات، جهات رسمية) برابط خارجي عند ذكر أرقام أو ادعاءات.",
         "Cite trustworthy sources (studies, official bodies) with an external link whenever you state figures or claims."),
        ("محتوى الصفحة", "Page content")),
    "GEO_NO_ARTICLE_SCHEMA": ("Low",
        ("صفحة مقال بلا بيانات Article أو BlogPosting المنظمة", "Article page without Article / BlogPosting structured data"),
        ("أضف Schema بصيغة JSON-LD من نوع Article أو BlogPosting مع headline وauthor وdatePublished وimage.",
         "Add JSON-LD Schema of type Article or BlogPosting with headline, author, datePublished and image."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    # ---------------- GEO : site level
    "GEO_AI_SEARCH_BLOCKED": ("High",
        ("ملف robots.txt يحظر زواحف بحث وإجابات الذكاء الاصطناعي، وقد لا يظهر موقعك في ChatGPT وPerplexity وClaude",
         "robots.txt blocks AI search / answer crawlers, so your site may not appear in ChatGPT, Perplexity or Claude"),
        ("إن أردت الظهور في إجابات الذكاء الاصطناعي فاسمح لهذه الزواحف (OAI-SearchBot وChatGPT-User وPerplexityBot وClaude-SearchBot) بحذف قواعد Disallow الخاصة بها.",
         "If you want to appear in AI answers, allow these crawlers (OAI-SearchBot, ChatGPT-User, PerplexityBot, Claude-SearchBot) by removing their Disallow rules."),
        ("ملف robots.txt في جذر الموقع", "robots.txt in the site root")),
    "GEO_AI_TRAIN_BLOCKED": ("Info",
        ("الموقع يحظر زواحف تدريب نماذج الذكاء الاصطناعي", "The site blocks AI model-training crawlers"),
        ("هذا خيار مشروع لحماية المحتوى، ولا يمنع في العادة الظهور في إجابات البحث. أبقِه إن كان مقصودًا.",
         "This is a legitimate choice to protect content and does not normally stop you appearing in search answers. Keep it if it is intentional."),
        ("ملف robots.txt في جذر الموقع", "robots.txt in the site root")),
    "GEO_NO_LLMS": ("Info",
        ("لا يوجد ملف llms.txt (اقتراح تقني اختياري)", "No llms.txt file (optional technical proposal)"),
        ("اختياري: يمكنك إضافة /llms.txt يلخّص الموقع وأهم صفحاته. ليس معيارًا رسميًا ولا يوجد تأثير مؤكد على الظهور.",
         "Optional: you can add /llms.txt summarising the site and its key pages. It is not an official standard and has no confirmed effect on visibility."),
        ("جذر الموقع: /llms.txt", "Site root: /llms.txt")),
    "GEO_NO_ENTITY": ("Medium",
        ("لا توجد بيانات منظمة للكيان (Organization أو Person أو LocalBusiness) في الصفحة الرئيسية",
         "No entity structured data (Organization, Person or LocalBusiness) on the homepage"),
        ("أضف Schema بصيغة JSON-LD يعرّف الجهة أو الشخص: الاسم والشعار والرابط ووسائل التواصل.",
         "Add JSON-LD Schema that defines the organisation or person: name, logo, URL and contact / profile links."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    "GEO_NO_SAMEAS": ("Low",
        ("بيانات الكيان بلا روابط sameAs إلى الحسابات الرسمية (LinkedIn وفيسبوك وويكيبيديا ...)",
         "Entity data has no sameAs links to official profiles (LinkedIn, Facebook, Wikipedia ...)"),
        ("أضف sameAs في Schema الكيان واربطه بحساباتك الرسمية.", "Add sameAs to the entity Schema and link your official profiles."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    "GEO_NO_ABOUT": ("Low",
        ("لا توجد صفحة «من نحن» أو «اتصل بنا» مرتبطة من الصفحة الرئيسية (إشارة ثقة)",
         "No About or Contact page linked from the homepage (trust signal)"),
        ("أنشئ صفحة تعريفية وصفحة تواصل واربطهما من القائمة أو التذييل.",
         "Create an About page and a Contact page and link them from the menu or footer."),
        ("القائمة / التذييل", "Menu / footer")),
}

# Issues that are checked once for the whole site (not per page)
SITE_GEO_CODES = {"GEO_AI_SEARCH_BLOCKED", "GEO_AI_TRAIN_BLOCKED", "GEO_NO_LLMS", "GEO_NO_ENTITY",
                  "GEO_NO_SAMEAS", "GEO_NO_ABOUT"}

# ------------------------------------------------------------------ المنصة / Platform-aware advice
PLAT_NAMES = {
    "wordpress": ("WordPress", "WordPress"), "blogger": ("Blogger", "Blogger"), "shopify": ("Shopify", "Shopify"),
    "wix": ("Wix", "Wix"), "squarespace": ("Squarespace", "Squarespace"), "webflow": ("Webflow", "Webflow"),
    "joomla": ("Joomla", "Joomla"), "drupal": ("Drupal", "Drupal"),
    "js": ("إطار JavaScript (Next.js / Nuxt / Gatsby)", "JavaScript framework (Next.js / Nuxt / Gatsby)"),
    "other": ("غير معروفة (موقع مخصّص)", "Unknown (custom site)"),
}

PLAT_NOTE = {
    "wordpress": ("تم اكتشاف WordPress: يمكن إصلاح معظم الملاحظات عبر إضافة SEO (Yoast أو Rank Math) وإضافة تخزين مؤقت وإعدادات إعادة التوجيه.",
                  "WordPress detected: most issues can be fixed with an SEO plugin (Yoast or Rank Math), a cache plugin and redirect settings."),
    "blogger": ("تم اكتشاف Blogger: لا توجد فيه إضافات مثل Yoast أو Rank Math ولا ملف .htaccess ولا إضافات تخزين مؤقت. تُنفَّذ الإصلاحات من إعدادات Blogger ومن السمة (Theme ← Edit HTML)، وبعض ملاحظات السرعة والخادم لا يمكن حلّها.",
                "Blogger detected: it has no plugins such as Yoast or Rank Math, no .htaccess and no cache plugins. Fixes are made from Blogger settings and the theme (Theme > Edit HTML), and some speed / server items cannot be solved."),
    "shopify": ("تم اكتشاف Shopify: تُنفَّذ الإصلاحات من لوحة التحكم ومن ملفات السمة (Liquid)، أما الخريطة وSSL فتُدارَان تلقائيًا.",
                "Shopify detected: fixes are made from the admin and the theme files (Liquid); the sitemap and SSL are managed automatically."),
    "wix": ("تم اكتشاف Wix: تُنفَّذ معظم الإصلاحات من لوحة SEO داخل Wix، وبعض الإعدادات التقنية (الخادم والتخزين المؤقت) غير قابلة للتعديل.",
            "Wix detected: most fixes are made in Wix's SEO panel, and some technical settings (server, caching) cannot be changed."),
    "squarespace": ("تم اكتشاف Squarespace: تُنفَّذ الإصلاحات من إعدادات الصفحة والموقع، وملف robots.txt لا يمكن تعديله.",
                    "Squarespace detected: fixes are made from page and site settings, and robots.txt cannot be edited."),
    "webflow": ("تم اكتشاف Webflow: تُنفَّذ الإصلاحات من إعدادات الصفحة وإعدادات المشروع (SEO وHosting وCustom code).",
                "Webflow detected: fixes are made from Page settings and Project settings (SEO, Hosting, Custom code)."),
    "joomla": ("تم اكتشاف Joomla: استخدم إعدادات SEO والإضافات المناسبة له؛ والإرشادات هنا عامة وقد تحتاج إلى تكييف.",
               "Joomla detected: use its SEO settings and suitable extensions; the guidance here is generic and may need adapting."),
    "drupal": ("تم اكتشاف Drupal: استخدم وحدات SEO المناسبة له؛ والإرشادات هنا عامة وقد تحتاج إلى تكييف.",
               "Drupal detected: use suitable SEO modules; the guidance here is generic and may need adapting."),
    "js": ("الموقع مبني بإطار JavaScript: تأكد من أن المحتوى يُعرَض من الخادم (SSR أو SSG) ليراه جوجل وزواحف الذكاء الاصطناعي، فكثير منها لا ينفّذ JavaScript.",
           "The site is built with a JavaScript framework: make sure content is server-rendered (SSR or SSG) so Google and AI crawlers can see it, because many of them do not run JavaScript."),
    "other": ("لم يُكتشف نوع المنصة: الإرشادات عامة وقد تحتاج إلى تكييف حسب نظام الموقع.",
              "Platform not detected: the guidance is generic and may need adapting to the site's system."),
}

# Where each issue is fixed, per platform. WordPress keeps the texts written in CATALOG.
WHERE_KEY = {}
for _k, _codes in {
    "title": "TITLE_MISSING TITLE_SHORT TITLE_LONG TITLE_DUP",
    "desc": "DESC_MISSING DESC_SHORT DESC_LONG DESC_DUP",
    "canon": "CANON_MISSING CANON_OTHER",
    "noindex": "NOINDEX",
    "social": "NO_OG",
    "schema": "NO_SCHEMA GEO_NO_ARTICLE_SCHEMA GEO_NO_ENTITY GEO_NO_SAMEAS",
    "redir": "HTTP_4XX REDIRECT_CHAIN",
    "sitemap": "REDIRECT NOT_IN_SITEMAP",
    "robots": "ROBOTS_BLOCKED GEO_AI_SEARCH_BLOCKED GEO_AI_TRAIN_BLOCKED",
    "speed": "SLOW BIG_PAGE PSI_POOR PSI_AVG",
    "theme": "NO_VIEWPORT NO_LANG MIXED_CONTENT",
    "slug": "URL_LONG URL_STYLE",
    "ssl": "NOT_HTTPS",
}.items():
    for _c in _codes.split():
        WHERE_KEY[_c] = _k

SEO_KEYS = {"title", "desc", "canon", "noindex", "social", "schema"}

PLAT_WHERE = {
    "blogger": {
        "title": ("Blogger: عنوان التدوينة أو الصفحة هو عنوان الصفحة؛ وللصفحة الرئيسية: الإعدادات ← Title",
                  "Blogger: the post or page title becomes the page title; for the homepage: Settings > Title"),
        "desc": ("Blogger: إعدادات التدوينة ← Search description (والرئيسية: الإعدادات ← Meta tags ← Description)",
                 "Blogger: Post settings > Search description (homepage: Settings > Meta tags > Description)"),
        "canon": ("Blogger: السمة ← Edit HTML (وسم canonical في قسم <head>)", "Blogger: Theme > Edit HTML (the canonical tag in <head>)"),
        "noindex": ("Blogger: إعدادات التدوينة ← Custom robots tags، أو الإعدادات ← Crawlers and indexing ← Custom robots header tags",
                    "Blogger: Post settings > Custom robots tags, or Settings > Crawlers and indexing > Custom robots header tags"),
        "social": ("Blogger: السمة ← Edit HTML (وسوم og: في قسم <head>)", "Blogger: Theme > Edit HTML (og: tags in <head>)"),
        "schema": ("Blogger: السمة ← Edit HTML (بيانات JSON-LD)", "Blogger: Theme > Edit HTML (JSON-LD data)"),
        "redir": ("Blogger: الإعدادات ← Custom redirects (لا يوجد .htaccess في Blogger)", "Blogger: Settings > Custom redirects (Blogger has no .htaccess)"),
        "sitemap": ("خريطة Blogger تُنشأ تلقائيًا ولا تُعدَّل؛ عالج الصفحة نفسها (حذف أو إعادة توجيه)",
                    "Blogger builds the sitemap automatically and it cannot be edited; fix the page itself (delete or redirect it)"),
        "robots": ("Blogger: الإعدادات ← Crawlers and indexing ← Custom robots.txt", "Blogger: Settings > Crawlers and indexing > Custom robots.txt"),
        "speed": ("لا يمكن تغيير استضافة Blogger: اضغط الصور واحذف الودجات الثقيلة واستخدم سمة أخف (Theme ← Edit HTML)",
                  "Blogger hosting cannot be changed: compress images, remove heavy widgets and use a lighter theme (Theme > Edit HTML)"),
        "theme": ("Blogger: Theme ← Edit HTML (قسم <head>)", "Blogger: Theme > Edit HTML (the <head> section)"),
        "slug": ("Blogger: إعدادات التدوينة ← Permalink (قبل النشر فقط)", "Blogger: Post settings > Permalink (only before publishing)"),
        "ssl": ("Blogger: الإعدادات ← HTTPS (التوفر وإعادة التوجيه)", "Blogger: Settings > HTTPS (availability and redirect)"),
    },
    "shopify": {
        "seo": ("Shopify: افتح الصفحة أو المنتج ← Search engine listing؛ والوسوم عبر Online Store ← Themes ← Edit code (theme.liquid)",
                "Shopify: open the page or product > Search engine listing; tags via Online Store > Themes > Edit code (theme.liquid)"),
        "redir": ("Shopify: Online Store ← Navigation ← URL Redirects", "Shopify: Online Store > Navigation > URL Redirects"),
        "sitemap": ("Shopify ينشئ sitemap.xml تلقائيًا ولا يُعدَّل؛ عالج الصفحة نفسها أو أخفها عن محركات البحث",
                    "Shopify builds sitemap.xml automatically (not editable); fix the page itself or hide it from search engines"),
        "robots": ("Shopify: قالب robots.txt.liquid ضمن Edit code", "Shopify: the robots.txt.liquid template under Edit code"),
        "speed": ("Shopify: احذف التطبيقات غير المستعملة واضغط الصور واختر سمة أخف", "Shopify: remove unused apps, compress images, choose a lighter theme"),
        "theme": ("Shopify: Online Store ← Themes ← Edit code (theme.liquid)", "Shopify: Online Store > Themes > Edit code (theme.liquid)"),
        "slug": ("Shopify: حقل URL handle في Search engine listing", "Shopify: the URL handle in Search engine listing"),
        "ssl": ("Shopify يوفر SSL تلقائيًا: راجع Settings ← Domains", "Shopify provides SSL automatically: check Settings > Domains"),
    },
    "wix": {
        "seo": ("Wix: لوحة SEO للصفحة (SEO basics) أو Marketing & SEO ← SEO Tools", "Wix: the page's SEO panel (SEO basics) or Marketing & SEO > SEO Tools"),
        "redir": ("Wix: SEO Tools ← URL Redirect Manager", "Wix: SEO Tools > URL Redirect Manager"),
        "sitemap": ("Wix ينشئ خريطة الموقع تلقائيًا؛ عالج الصفحة نفسها", "Wix builds the sitemap automatically; fix the page itself"),
        "robots": ("Wix: SEO Tools ← Robots.txt Editor", "Wix: SEO Tools > Robots.txt Editor"),
        "speed": ("Wix: اضغط الصور وقلّل الحركات والتطبيقات؛ الاستضافة لا تتغير", "Wix: compress images and cut animations and apps; hosting cannot be changed"),
        "theme": ("Wix: Settings ← Custom code", "Wix: Settings > Custom code"),
        "slug": ("Wix: إعدادات SEO للصفحة ← URL slug", "Wix: the page's SEO settings > URL slug"),
        "ssl": ("Wix يوفر SSL تلقائيًا: راجع Settings ← Domains", "Wix provides SSL automatically: check Settings > Domains"),
    },
    "squarespace": {
        "seo": ("Squarespace: Page settings ← SEO؛ والوسوم عبر Settings ← Advanced ← Code injection", "Squarespace: Page settings > SEO; extra tags via Settings > Advanced > Code injection"),
        "redir": ("Squarespace: Settings ← URL Mappings", "Squarespace: Settings > URL Mappings"),
        "sitemap": ("Squarespace ينشئ الخريطة تلقائيًا؛ عالج الصفحة نفسها", "Squarespace builds the sitemap automatically; fix the page itself"),
        "robots": ("Squarespace لا يسمح بتعديل robots.txt؛ استخدم Page settings ← SEO ← Hide page from search results",
                   "Squarespace does not allow editing robots.txt; use Page settings > SEO > Hide page from search results"),
        "speed": ("Squarespace: اضغط الصور وقلّل الأكواد الإضافية", "Squarespace: compress images and cut extra code"),
        "theme": ("Squarespace: Settings ← Advanced ← Code injection", "Squarespace: Settings > Advanced > Code injection"),
        "slug": ("Squarespace: Page settings ← General ← URL slug", "Squarespace: Page settings > General > URL slug"),
        "ssl": ("Squarespace: Settings ← Domains ← SSL", "Squarespace: Settings > Domains > SSL"),
    },
    "webflow": {
        "seo": ("Webflow: Page settings ← SEO settings", "Webflow: Page settings > SEO settings"),
        "redir": ("Webflow: Project settings ← Hosting ← 301 redirects", "Webflow: Project settings > Hosting > 301 redirects"),
        "sitemap": ("Webflow: Project settings ← SEO ← Sitemap", "Webflow: Project settings > SEO > Sitemap"),
        "robots": ("Webflow: Project settings ← SEO ← robots.txt", "Webflow: Project settings > SEO > robots.txt"),
        "speed": ("Webflow: اضغط الصور وقلّل التفاعلات والأكواد المخصصة", "Webflow: compress images and cut interactions and custom code"),
        "theme": ("Webflow: Project settings ← Custom code", "Webflow: Project settings > Custom code"),
        "slug": ("Webflow: Page settings ← Slug", "Webflow: Page settings > Slug"),
        "ssl": ("Webflow: Project settings ← Hosting ← SSL", "Webflow: Project settings > Hosting > SSL"),
    },
    "generic": {  # Joomla, Drupal, JavaScript frameworks and unknown platforms
        "seo": ("إعدادات SEO في نظام الموقع أو قسم <head> في قالب الصفحة", "The CMS SEO settings or the <head> section of the page template"),
        "redir": ("إعدادات إعادة التوجيه في الخادم أو النظام (.htaccess / nginx / Cloudflare)", "Server or CMS redirect settings (.htaccess / nginx / Cloudflare)"),
        "sitemap": ("ملف sitemap.xml (مولّد الخريطة في النظام)", "sitemap.xml (the CMS sitemap generator)"),
        "robots": ("ملف robots.txt في جذر الموقع", "robots.txt in the site root"),
        "speed": ("الاستضافة + التخزين المؤقت + الصور + القالب", "Hosting + caching + images + theme"),
        "theme": ("قالب الموقع (<head>)", "Site template (<head>)"),
        "slug": ("إعدادات الرابط الدائم للصفحة", "Page permalink settings"),
        "ssl": ("الاستضافة (SSL) + إعادة توجيه http إلى https", "Hosting (SSL) + http-to-https redirect"),
    },
}


def pname():
    return PLAT_NAMES.get(PLATFORM, PLAT_NAMES["other"])[0 if LANG == "ar" else 1]


def pnote():
    return PLAT_NOTE.get(PLATFORM, PLAT_NOTE["other"])[0 if LANG == "ar" else 1]


def cat(code):
    """(severity, description, how to fix, where to fix) in the report language and for the detected platform."""
    sev, d, f, w = CATALOG[code]
    i = 0 if LANG == "ar" else 1
    key = WHERE_KEY.get(code)
    if key and PLATFORM != "wordpress":
        table = PLAT_WHERE.get(PLATFORM, PLAT_WHERE["generic"])
        # the finer keys (title, desc, canon, noindex, social, schema) fall back to the general "seo" text
        w2 = table.get(key) or (table.get("seo") if key in SEO_KEYS else None)
        if w2:
            return sev, d[i], f[i], w2[i]
    return sev, d[i], f[i], w[i]


def detect_platform(home):
    """Guess the website platform from the homepage: home = (soup, html_text, headers) or None."""
    if not home:
        return "other"
    soup, text, headers = home
    h = (text or "")[:600000].lower()
    hd = {str(k).lower(): str(v).lower() for k, v in (headers or {}).items()}
    g = soup.find("meta", attrs={"name": re.compile(r"^generator$", re.I)})
    gen = (g.get("content") or "").lower() if g else ""
    if "blogger" in gen or "blogger.com/static" in h or "/feeds/posts/default" in h:
        return "blogger"
    if "wordpress" in gen or "/wp-content/" in h or "/wp-includes/" in h or "api.w.org" in h or "api.w.org" in hd.get("link", ""):
        return "wordpress"
    if "cdn.shopify.com" in h or "shopify.theme" in h or "x-shopid" in hd or "x-shopify-stage" in hd:
        return "shopify"
    if "wixstatic.com" in h or "wix.com" in gen or "x-wix-request-id" in hd:
        return "wix"
    if "squarespace" in gen or "static1.squarespace.com" in h:
        return "squarespace"
    if "webflow" in gen or "data-wf-site" in h or "website-files.com" in h:
        return "webflow"
    if "joomla" in gen:
        return "joomla"
    if "drupal" in gen or "drupal" in hd.get("x-generator", "") or "x-drupal-cache" in hd:
        return "drupal"
    if "/_next/" in h or "__next_data__" in h or "/_nuxt/" in h or "___gatsby" in h:
        return "js"
    return "other"


# ------------------------------------------------------------------ الشبكة / Network
def make_session():
    s = requests.Session()
    s.headers.update(HEADERS)
    return s


def fetch(session, url, timeout, retries=2):
    last = None
    for i in range(retries + 1):
        try:
            return session.get(url, timeout=timeout, allow_redirects=True)
        except Exception as e:  # noqa
            last = e
            time.sleep(1 + i)
    raise last


# ------------------------------------------------------------------ Sitemap
def read_locs(content, url):
    if url.endswith(".gz") or content[:2] == b"\x1f\x8b":
        content = gzip.decompress(content)
    text = content.decode("utf-8", errors="ignore")
    is_index = "<sitemapindex" in text
    locs = [re.sub(r"<!\[CDATA\[|\]\]>", "", m).strip()
            for m in re.findall(r"<loc>\s*(.*?)\s*</loc>", text, re.S)]
    return is_index, locs


def expand_sitemap(session, url, timeout, depth=0, seen=None):
    seen = seen if seen is not None else set()
    if url in seen or depth > 3:
        return []
    seen.add(url)
    try:
        r = fetch(session, url, timeout, retries=1)
        if r.status_code != 200:
            return []
        is_index, locs = read_locs(r.content, url)
    except Exception:
        return []
    if not is_index:
        return locs
    urls = []
    for child in locs[:50]:
        urls += expand_sitemap(session, child, timeout, depth + 1, seen)
    return urls


def discover(session, base, timeout):
    robots_txt = ""
    candidates = []
    try:
        r = session.get(base + "/robots.txt", timeout=timeout)
        if r.status_code == 200 and "<html" not in r.text[:500].lower():
            robots_txt = r.text
            for line in robots_txt.splitlines():
                if line.lower().startswith("sitemap:"):
                    candidates.append(line.split(":", 1)[1].strip())
    except Exception:
        pass
    for p in ("/sitemap.xml", "/sitemap_index.xml", "/wp-sitemap.xml"):
        candidates.append(base + p)
    for c in dict.fromkeys(candidates):
        urls = expand_sitemap(session, c, timeout)
        if urls:
            return c, list(dict.fromkeys(urls)), robots_txt
    return None, [], robots_txt


# ------------------------------------------------------------------ فحص صفحة / Page analysis
def norm(u):
    u = u.split("#")[0].strip().lower()
    return u[:-1] if u.endswith("/") else u


ASSET_RE = re.compile(r"\.(jpe?g|png|gif|webp|svg|ico|pdf|zip|rar|css|js|mp4|mp3|docx?|xlsx?|pptx?)(\?.*)?$", re.I)


def looks_like_site(s):
    """True for 'https://x', 'http://x' or a bare domain like 'example.com' (not a file path)."""
    s = (s or "").strip()
    if s.lower().startswith(("http://", "https://")):
        return True
    return bool(re.match(r"^([a-z0-9-]+\.)+[a-z]{2,}(:\d+)?(/.*)?$", s, re.I)) and not s.lower().endswith(".json")


def same_host(a, b):
    ha = urlparse(a).netloc.lower().replace("www.", "")
    hb = urlparse(b).netloc.lower().replace("www.", "")
    return ha == hb


def check_link(session, url, timeout=10):
    """Returns the HTTP status of a link (or 'ERROR')."""
    try:
        r = session.head(url, timeout=timeout, allow_redirects=True)
        if r.status_code >= 400:
            r = session.get(url, timeout=timeout, allow_redirects=True, stream=True)
            r.close()
        return r.status_code
    except Exception:
        return "ERROR"


def is_broken(st):
    # 401/403/429 are usually bot protection, not a truly broken link
    return st == "ERROR" or st in (404, 410) or (isinstance(st, int) and 500 <= st < 600)


PSI_CAT = {
    "ar": {"FAST": "سريع", "AVERAGE": "متوسط", "SLOW": "بطيء"},
    "en": {"FAST": "Fast", "AVERAGE": "Average", "SLOW": "Slow"},
}


def parse_psi(data):
    """Extract [score, FCP s, LCP s, CLS, TBT ms, TTFB ms, real-user rating, real-user INP ms] from a PSI response."""
    lh = data["lighthouseResult"]
    a = lh["audits"]

    def num(key, scale=1, nd=None):
        try:
            v = a[key]["numericValue"] / scale
            return round(v, nd) if nd is not None else round(v)
        except Exception:
            return "n/a"

    score = round(lh["categories"]["performance"]["score"] * 100)
    field = data.get("loadingExperience") or {}
    cat = PSI_CAT[LANG].get(field.get("overall_category"), "n/a")
    inp = (field.get("metrics", {}).get("INTERACTION_TO_NEXT_PAINT") or {}).get("percentile", "n/a")
    return [score,
            num("first-contentful-paint", 1000, 2),
            num("largest-contentful-paint", 1000, 2),
            num("cumulative-layout-shift", 1, 3),
            num("total-blocking-time"),
            num("server-response-time"),
            cat, inp]


def pagespeed(session, url, key="", strategy="mobile", timeout=90):
    """Google PageSpeed Insights API (mobile or desktop). Returns (values, error)."""
    params = {"url": url, "strategy": strategy, "category": "performance"}
    if key:
        params["key"] = key
    try:
        r = session.get("https://www.googleapis.com/pagespeedonline/v5/runPagespeed",
                        params=params, timeout=timeout)
        if r.status_code != 200:
            msg = ""
            try:  # Google explains the exact reason in the JSON body
                msg = str((r.json().get("error") or {}).get("message", ""))
            except Exception:  # noqa
                pass
            return None, f"HTTP {r.status_code}" + (f": {msg[:170]}" if msg else "")
        return parse_psi(r.json()), ""
    except Exception as e:  # noqa
        return None, str(e)[:80]


# ------------------------------------------------------------------ AEO / GEO helpers
ARTICLE_TYPES = {"Article", "BlogPosting", "NewsArticle", "TechArticle", "ScholarlyArticle", "Report", "OpinionNewsArticle"}
ENTITY_TYPES = {"Organization", "LocalBusiness", "Person", "NGO", "Corporation", "EducationalOrganization", "OnlineStore",
                "Store", "NewsMediaOrganization", "ResearchOrganization", "GovernmentOrganization", "MedicalOrganization",
                "Restaurant", "Hotel", "ProfessionalService", "TravelAgency"}
QRE = re.compile(r"^(what|why|how|when|where|who|which|can|does|do|is|are|should|qu['’]est|qu['’]|quel|quelle|quels|quelles|"
                 r"quand|comment|pourquoi|combien|est-ce|ما|ماذا|لماذا|كيف|هل|متى|أين|كم|أي)\b", re.I)

# AI crawlers: (robots.txt user-agent, owner, kind) - kind: search / user = appear in AI answers, train = model training
AI_BOTS = [
    ("OAI-SearchBot", "OpenAI", "search"), ("ChatGPT-User", "OpenAI", "user"),
    ("PerplexityBot", "Perplexity", "search"), ("Claude-SearchBot", "Anthropic", "search"),
    ("GPTBot", "OpenAI", "train"), ("ClaudeBot", "Anthropic", "train"), ("Google-Extended", "Google", "train"),
    ("Applebot-Extended", "Apple", "train"), ("CCBot", "Common Crawl", "train"),
]


def jsonld_info(soup):
    """Read JSON-LD / microdata: returns (schema types, has author, has date, has sameAs)."""
    types, flags = set(), {"author": False, "date": False, "same": False}

    def walk(o):
        if isinstance(o, list):
            for x in o:
                walk(x)
        elif isinstance(o, dict):
            t = o.get("@type")
            for x in (t if isinstance(t, list) else [t]):
                if x:
                    types.add(str(x).rstrip("/").split("/")[-1].split("#")[-1])
            if o.get("author"):
                flags["author"] = True
            if o.get("datePublished") or o.get("dateModified"):
                flags["date"] = True
            if o.get("sameAs"):
                flags["same"] = True
            for v in o.values():
                if isinstance(v, (list, dict)):
                    walk(v)

    for s in soup.find_all("script", type=re.compile(r"ld\+json", re.I)):
        try:
            walk(json.loads(s.get_text() or "", strict=False))
        except Exception:  # noqa
            continue
    for el in soup.find_all(attrs={"itemtype": True}):
        types.add(str(el.get("itemtype")).rstrip("/").split("/")[-1])
    return types, flags["author"], flags["date"], flags["same"]


def fetch_home(session, base, timeout):
    """Homepage as (soup, html text, headers) - used for platform and entity checks - or None."""
    try:
        r = fetch(session, base + "/", timeout, retries=1)
        if r.status_code == 200 and "html" in r.headers.get("Content-Type", "").lower():
            return BeautifulSoup(r.content, "html.parser"), r.text, dict(r.headers)
    except Exception:  # noqa
        pass
    return None


def site_geo_checks(session, base, timeout, rp, home):
    """Site-wide AEO/GEO checks. Returns a dict: issues [(code, detail)], bots, llms, entity_types, same_as, about, contact."""
    out = {"issues": [], "bots": [], "llms": False, "entity_types": [], "same_as": False,
           "about": None, "contact": None, "home_ok": home is not None}
    for bot, owner, kind in AI_BOTS:
        allowed = True if rp is None else rp.can_fetch(bot, base + "/")
        out["bots"].append((bot, owner, kind, allowed))
    blocked_find = [b for b, _o, k, ok in out["bots"] if k in ("search", "user") and not ok]
    blocked_train = [b for b, _o, k, ok in out["bots"] if k == "train" and not ok]
    if blocked_find:
        out["issues"].append(("GEO_AI_SEARCH_BLOCKED", ", ".join(blocked_find)))
    if blocked_train:
        out["issues"].append(("GEO_AI_TRAIN_BLOCKED", ", ".join(blocked_train)))

    try:  # llms.txt must be a plain-text file, not an HTML 404 page
        r = session.get(base + "/llms.txt", timeout=timeout)
        out["llms"] = (r.status_code == 200 and "html" not in r.headers.get("Content-Type", "").lower()
                       and len(r.text.strip()) > 0)
    except Exception:  # noqa
        pass
    if not out["llms"]:
        out["issues"].append(("GEO_NO_LLMS", ""))

    if home:
        soup = home[0]
        types, _a, _d, same = jsonld_info(soup)
        ent = sorted(types & ENTITY_TYPES)
        out["entity_types"], out["same_as"] = ent, same
        if not ent:
            out["issues"].append(("GEO_NO_ENTITY", ""))
        elif not same:
            out["issues"].append(("GEO_NO_SAMEAS", ", ".join(ent)))
        about_re = re.compile(r"about|a-propos|apropos|à propos|qui-sommes|qui sommes|notre-histoire|team|equipe|équipe|من-نحن|من نحن|عنا", re.I)
        contact_re = re.compile(r"contact|nous-contacter|اتصل|تواصل", re.I)
        about = contact = False
        for a in soup.find_all("a", href=True):
            blob = (a.get("href", "") + " " + a.get_text(" ", strip=True))
            about = about or bool(about_re.search(blob))
            contact = contact or bool(contact_re.search(blob))
        out["about"], out["contact"] = about, contact
        if not (about or contact):
            out["issues"].append(("GEO_NO_ABOUT", ""))
    return out


def analyze(url, session, timeout, delay, rp):
    issues = []
    row = {"URL": url}
    try:
        r = fetch(session, url, timeout)
    except Exception as e:  # noqa
        row["Status"] = "ERROR"
        issues.append(("FETCH_FAIL", str(e)[:150]))
        return row, issues
    time.sleep(delay)

    final = r.url
    status = r.status_code
    row.update({
        "Status": status,
        "Final URL": final,
        "Redirects": len(r.history),
        "Response (s)": round(r.elapsed.total_seconds(), 2),
        "Size (KB)": round(len(r.content) / 1024),
    })

    if rp is not None and not rp.can_fetch("*", url):
        issues.append(("ROBOTS_BLOCKED", tr("robots_dis")))
    if len(r.history) >= 2:
        issues.append(("REDIRECT_CHAIN", tr("chain", n=len(r.history), u=final)))
    elif len(r.history) == 1:
        issues.append(("REDIRECT", tr("redir", u=final)))
    if 400 <= status < 500:
        issues.append(("HTTP_4XX", tr("status_n", n=status)))
        return row, issues
    if status >= 500:
        issues.append(("HTTP_5XX", tr("status_n", n=status)))
        return row, issues

    if not final.startswith("https://"):
        issues.append(("NOT_HTTPS", final))
    if row["Response (s)"] > 1.5:
        issues.append(("SLOW", tr("sec", n=row["Response (s)"])))
    if len(r.content) > 1_500_000:
        issues.append(("BIG_PAGE", f"{row['Size (KB)']} KB"))

    ctype = r.headers.get("Content-Type", "")
    if "html" not in ctype.lower():
        return row, issues

    soup = BeautifulSoup(r.content, "html.parser")

    # URL
    path = urlparse(final)
    if len(final) > 115:
        issues.append(("URL_LONG", tr("chars", n=len(final))))
    if re.search(r"[A-Z]", path.path) or "_" in path.path or path.query:
        issues.append(("URL_STYLE", final))

    # Title
    t = soup.find("title")
    title = t.get_text(strip=True) if t else ""
    row["Title"] = title
    row["Title len"] = len(title)
    if not title:
        issues.append(("TITLE_MISSING", ""))
    elif len(title) < 30:
        issues.append(("TITLE_SHORT", f"{tr('chars', n=len(title))}: {title}"))
    elif len(title) > 60:
        issues.append(("TITLE_LONG", f"{tr('chars', n=len(title))}: {title}"))

    # Meta description
    m = soup.find("meta", attrs={"name": re.compile(r"^description$", re.I)})
    desc = (m.get("content") or "").strip() if m else ""
    row["Meta description"] = desc
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
SEO + AEO/GEO Audit from sitemap.xml  |  فحص SEO تقني وجاهزية AEO/GEO من خريطة الموقع   (v4)

- Reads robots.txt and sitemap.xml (including sitemap index files)
- Detects the website platform (WordPress, Blogger, Shopify, Wix, ...) and gives fixes that fit that platform
- Checks pages one by one, discovers extra pages via internal links, finds broken links and orphan pages
- AEO/GEO checks: AI crawler access in robots.txt, llms.txt, entity schema, author/date/sources, question headings
- Optional Google PageSpeed test (--psi N) and a warning for JavaScript-rendered pages
- Two scores out of 100: technical SEO score and AEO/GEO readiness score (heuristic estimates, not official metrics)
- Exports an Excel report (issues, how to fix, where to fix) and a one-page PDF summary
- Report language: Modern Standard Arabic (--lang ar) or English (--lang en)

Install:
    pip install requests beautifulsoup4 pandas openpyxl weasyprint      (weasyprint is only needed for the PDF)

Usage:
    python seo_audit_v4.py https://example.com
    python seo_audit_v4.py https://example.com --lang en --max 200 --out report.xlsx
    python seo_audit_v4.py https://example.com --max 0 --crawl 200   (all sitemap pages + 200 discovered pages)
    python seo_audit_v4.py https://example.com --psi 5                (Google PageSpeed on 5 pages; default 1 = homepage; 0 = off)
    python seo_audit_v4.py https://example.com --pdf 0                (skip the PDF)
"""
import argparse
import gzip
import html as html_lib
import json
import math
import re
import sys
import time
from datetime import date
from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from urllib.parse import urldefrag, urljoin, urlparse
from urllib.robotparser import RobotFileParser

import pandas as pd
import requests
from bs4 import BeautifulSoup

# ======================================================================
# 👇👇👇  هنا كتبدل رابط الموقع اللي بغيتي تفحصو (الموقع الحالي: odysseedelaterre.fr)
#         CHANGE THE WEBSITE TO AUDIT HERE (the only line you need to edit)
# ======================================================================
SITE_URL = "https://odysseedelaterre.fr"

# 👇 لغة التقرير: "en" = عربية فصحى  |  "en" = English
#    REPORT LANGUAGE: "ar" = Arabic  |  "en" = English
REPORT_LANG = "en"

# 👇 (اختياري) مفتاح Google PageSpeed. خليه فارغ "" إلا ما عندكش
PSI_API_KEY = "zaSyD0XbcuK74WPFo0umfMJTEHas5ppSDm3nkAI"
# ======================================================================
# (إلا كتبتي رابط فالأمر %run هو اللي كيتطبق، وإلا ماكتبتيش كيتطبق SITE_URL)

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/124.0 Safari/537.36 SEOAuditBot/1.0"
}

# Points deducted from a page's score for each issue, by severity
WEIGHTS = {"Critical": 25, "High": 12, "Medium": 5, "Low": 2, "Info": 0}
SEV_ORDER = {"Critical": 0, "High": 1, "Medium": 2, "Low": 3, "Info": 4}
SEVLBL = {
    "ar": {"Critical": "حرج", "High": "عالٍ", "Medium": "متوسط", "Low": "منخفض", "Info": "ملاحظة"},
    "en": {"Critical": "Critical", "High": "High", "Medium": "Medium", "Low": "Low", "Info": "Info"},
}

# Points deducted for AEO/GEO issues (fewer checks than SEO, so each one weighs more)
AI_WEIGHTS = {"Critical": 30, "High": 15, "Medium": 8, "Low": 4, "Info": 0}

LANG = "ar"
PLATFORM = "other"  # detected website platform, set in main()


def cat_of(code):
    """Issue family: AEO (answer engines), GEO (generative engines) or SEO (technical / on-page)."""
    return "AEO" if code.startswith("AEO_") else "GEO" if code.startswith("GEO_") else "SEO"


def is_ai(code):
    return code.startswith(("AEO_", "GEO_"))


# ------------------------------------------------------------------ النصوص / UI strings
S = {
    "ar": {
        "chars": "{n} حرفًا", "words": "{n} كلمة", "sec": "{n} ثانية",
        "robots_dis": "ممنوعة بقاعدة Disallow في robots.txt",
        "chain": "{n} عمليات إعادة توجيه ← {u}",
        "redir": "يعيد التوجيه إلى {u}",
        "img": "{a} من أصل {b} صورة بلا نص بديل",
        "mixed": "{n} ملفًا يُحمَّل عبر http",
        "yes": "نعم", "no": "لا",
        "g90": "ممتاز", "g75": "جيد", "g50": "متوسط (يحتاج إلى تحسينات)", "g0": "ضعيف (مشكلات كبيرة)",
        "sheets": ["الملخص", "أهم الإصلاحات", "كل المشكلات", "الصفحات", "الروابط المكسورة", "السرعة", "AEO-GEO"],
        "status_n": "رمز الاستجابة {n}",
        "page_cols": {"URL": "الرابط", "Status": "الحالة", "Final URL": "الرابط النهائي", "Redirects": "عدد إعادات التوجيه",
                      "Response (s)": "زمن الاستجابة (ثانية)", "Size (KB)": "الحجم (كيلوبايت)", "Title": "عنوان الصفحة",
                      "Title len": "طول العنوان", "Meta description": "وصف الميتا", "Desc len": "طول الوصف",
                      "H1": "العنوان الرئيسي H1", "H1 count": "عدد H1", "Canonical": "Canonical", "Noindex": "Noindex",
                      "Images": "عدد الصور", "Images no alt": "صور بلا نص بديل", "Internal links": "روابط داخلية",
                      "External links": "روابط خارجية", "Words": "عدد الكلمات", "Schema types": "أنواع Schema",
                      "Question headings": "عناوين على شكل أسئلة", "Lists & tables": "قوائم وجداول",
                      "Author": "الكاتب", "Date": "التاريخ", "Score": "تقييم SEO (من 100)",
                      "AI score": "تقييم AEO/GEO (من 100)", "Issues": "عدد المشكلات"},
        "bl_cols": ["الصفحة المصدر", "الرابط المكسور", "الحالة", "النوع", "نص الرابط"],
        "int": "داخلي", "ext": "خارجي",
        "speed_cols": ["الرابط", "الجهاز", "الأداء (من 100)", "FCP (ثانية)", "LCP (ثانية)", "CLS", "TBT (مللي ثانية)",
                       "TTFB (مللي ثانية)", "تقييم المستخدمين الحقيقيين (28 يومًا)", "INP للمستخدمين (مللي ثانية)"],
        "broken_n": "{n} رابط مكسور",
        "js_d": "نص قليل مع عناصر JavaScript كثيرة",
        "psi_d": "الأداء {s}/100",
        "crawl_found": "صفحات أُضيفت للفحص عبر الروابط الداخلية (خارج الخريطة)",
        "links_checked": "عدد الروابط التي جرى فحصها",
        "broken_total": "عدد الروابط المكسورة",
        "orphan_skip": "تم تخطي كشف الصفحات اليتيمة لأن الفحص لم يشمل كل الصفحات. استخدم --max 0 مع --crawl أكبر.",
        "p_crawl": "🕸️ تم اكتشاف {n} صفحة إضافية عبر الروابط الداخلية، جارٍ فحصها ...",
        "p_links": "🔗 جارٍ فحص {n} رابط بحثًا عن الروابط المكسورة ...",
        "p_psi": "⚡ جارٍ فحص السرعة عبر Google PageSpeed ({n} صفحة، على الجوال والكمبيوتر، قد يستغرق عدة دقائق) ...",
        "dev_mobile": "جوال", "dev_desktop": "كمبيوتر",
        "p_psi_err": "⚠️ تعذّر فحص PageSpeed ({d}) للرابط {u} ({e}). إن كان الخطأ 429 فالحصة المجانية استُنفدت: جرّب لاحقًا أو استعمل --psi-key، أو افحص الرابط يدويًا في pagespeed.web.dev",
        "sum_cols": ["البند", "القيمة"],
        "top_cols": ["النوع", "الخطورة", "المشكلة", "عدد الصفحات المتأثرة", "% من الصفحات", "طريقة الإصلاح", "مكان الإصلاح"],
        "iss_cols": ["الرابط", "النوع", "الخطورة", "المشكلة", "التفاصيل", "طريقة الإصلاح", "مكان الإصلاح", "الرمز"],
        "site": "الموقع", "sitemap": "خريطة الموقع (Sitemap)", "sm_total": "عدد الروابط في الخريطة",
        "checked": "عدد الصفحات المفحوصة", "score": "تقييم SEO التقني (من 100)", "verdict": "حكم SEO التقني",
        "issues_of": "عدد المشكلات ({s})",
        "note_l": "ملاحظة",
        "note": "التقييم من 100: كلما ارتفع الرقم كان الموقع أفضل. هذا تقييم تقديري من هذه الأداة وليس مقياسًا رسميًا من Google.",
        "site_issue": "مشكلة على مستوى الموقع",
        "no_robots": "لا يوجد ملف robots.txt",
        "no_robots_fix": "أنشئ ملف robots.txt في جذر الموقع وأضف إليه رابط خريطة الموقع.",
        "no_robots_where": "جذر الموقع: /robots.txt",
        "p_enter": "🔗 أدخل رابط الموقع (مثال: https://example.com): ",
        "p_search": "⏳ جارٍ البحث عن خريطة الموقع لـ {b} ...",
        "p_nosm": "❌ لم يتم العثور على خريطة موقع (robots.txt أو /sitemap.xml أو /sitemap_index.xml أو /wp-sitemap.xml).",
        "p_found": "✅ خريطة الموقع: {sm} | {n} رابط",
        "p_sample": "ℹ️ سيتم فحص {m} صفحة (عينة موزعة على الموقع). استخدم --max 0 لفحص الكل.",
        "p_prog": "   تم فحص {n}/{m} صفحة",
        "p_score": "🏁 تقييم SEO التقني: {s}/100 ← {g}",
        "p_top": "🔧 أهم المشكلات التي يجب إصلاحها:",
        "p_top_row": "   [{sev}] {issue} ({n} صفحة)",
        "p_file": "📄 التقرير الكامل: {f}",
        # ---- v4: platform, AEO/GEO, PDF
        "platform": "المنصة (CMS)", "platform_note_l": "ملاحظة حسب المنصة",
        "p_platform": "🧩 المنصة المكتشفة: {p}",
        "ai_score": "جاهزية AEO/GEO (من 100)", "ai_verdict": "حكم AEO/GEO",
        "geo_note": "جاهزية AEO/GEO تقيس الاستعداد التقني فقط (وصول الزواحف، البيانات المنظمة، بنية المحتوى، إشارات الثقة). هي لا تتنبأ بالظهور أو الاستشهاد داخل إجابات الذكاء الاصطناعي.",
        "p_ai": "🤖 جارٍ فحص جاهزية AEO/GEO (زواحف الذكاء الاصطناعي، llms.txt، بيانات الكيان) ...",
        "p_ai_score": "🤖 جاهزية AEO/GEO: {s}/100 ← {g}",
        "site_wide": "على مستوى الموقع",
        "geo_cols": ["الفحص", "الحالة", "التفاصيل"],
        "st_ok": "سليم", "st_bad": "مشكلة", "st_warn": "تنبيه", "st_info": "معلومة",
        "bot_row": "زاحف ذكاء اصطناعي: {b}",
        "bt": {"search": "بحث وإجابات", "user": "بطلب المستخدم", "train": "تدريب النماذج"},
        "bot_allowed": "مسموح", "bot_blocked": "محظور",
        "llms_l": "ملف llms.txt (اقتراح اختياري غير رسمي)", "llms_yes": "موجود", "llms_no": "غير موجود",
        "entity_l": "بيانات الكيان في الصفحة الرئيسية (Organization / Person / LocalBusiness)",
        "entity_no": "لم يُعثر على بيانات منظمة للكيان", "sameas_l": "روابط sameAs إلى الحسابات الرسمية",
        "about_l": "صفحة «من نحن» أو «اتصل بنا» مرتبطة من الصفحة الرئيسية",
        "found": "موجود", "missing": "غير موجود",
        "stat_q": "صفحات المحتوى التي فيها عناوين على شكل أسئلة", "stat_l": "صفحات المحتوى التي فيها قوائم أو جداول",
        "stat_a": "صفحات المقالات التي فيها كاتب ظاهر", "stat_d": "صفحات المقالات التي فيها تاريخ نشر أو تحديث",
        "n_of": "{a} من {b}",
        "p_psi_hint": "💡 الخطأ 403 أو 400 يعني غالبًا أن PageSpeed Insights API غير مفعّل أو أن المفتاح مقيّد بواجهة أخرى: افتح Google Cloud ← API et services ← Identifiants ← المفتاح ← Restrictions relatives aux API.",
        "p_pdf": "📑 جارٍ إنشاء ملخص PDF من صفحة واحدة ...",
        "p_pdf_ok": "📑 ملخص PDF: {f}",
        "p_pdf_fail": "⚠️ تعذّر إنشاء PDF ({e}). حُفظت نسخة HTML بدلًا منه: {h} (افتحها واضغط Ctrl+P ثم احفظها PDF).",
        "rtl": True,
    },
    "en": {
        "chars": "{n} chars", "words": "{n} words", "sec": "{n} s",
        "robots_dis": "Disallowed by robots.txt",
        "chain": "{n} redirects -> {u}",
        "redir": "Redirects to {u}",
        "img": "{a} of {b} images have no alt text",
        "mixed": "{n} resources loaded over http",
        "yes": "Yes", "no": "No",
        "g90": "Excellent", "g75": "Good", "g50": "Average (needs work)", "g0": "Poor (major issues)",
        "sheets": ["Summary", "Top Fixes", "All Issues", "Pages", "Broken Links", "Speed", "AEO-GEO"],
        "status_n": "Status {n}",
        "page_cols": {"Score": "SEO score", "AI score": "AEO/GEO score"},
        "bl_cols": ["Source page", "Broken link", "Status", "Type", "Anchor text"],
        "int": "Internal", "ext": "External",
        "speed_cols": ["URL", "Device", "Performance (/100)", "FCP (s)", "LCP (s)", "CLS", "TBT (ms)",
                       "TTFB (ms)", "Real-user rating (28 days)", "Real-user INP (ms)"],
        "broken_n": "{n} broken links",
        "js_d": "Very little text but many JavaScript elements",
        "psi_d": "Performance {s}/100",
        "crawl_found": "Pages added through internal links (not in sitemap)",
        "links_checked": "Links checked",
        "broken_total": "Broken links found",
        "orphan_skip": "Orphan-page detection skipped because not all pages were audited. Use --max 0 with a larger --crawl.",
        "p_crawl": "🕸️ Found {n} extra pages through internal links, auditing them ...",
        "p_links": "🔗 Checking {n} links for broken ones ...",
        "p_psi": "⚡ Running Google PageSpeed ({n} pages, mobile and desktop, may take several minutes) ...",
        "dev_mobile": "Mobile", "dev_desktop": "Desktop",
        "p_psi_err": "⚠️ PageSpeed failed ({d}) for {u} ({e}). If it is 429 the free quota is used up: try later or use --psi-key, or test the URL manually at pagespeed.web.dev",
        "sum_cols": ["Item", "Value"],
        "top_cols": ["Type", "Severity", "Issue", "Pages affected", "% of pages", "How to fix", "Where to fix"],
        "iss_cols": ["URL", "Type", "Severity", "Issue", "Detail", "How to fix", "Where to fix", "Code"],
        "site": "Website", "sitemap": "Sitemap", "sm_total": "URLs in sitemap",
        "checked": "Pages audited", "score": "Technical SEO score (out of 100)", "verdict": "Technical SEO verdict",
        "issues_of": "Issues ({s})",
        "note_l": "Note",
        "note": "Score is out of 100: the higher the better. It is this tool's own estimate, not an official Google metric.",
        "site_issue": "Site-wide issue",
        "no_robots": "No robots.txt file",
        "no_robots_fix": "Create a robots.txt file in the site root and add the sitemap URL to it.",
        "no_robots_where": "Site root: /robots.txt",
        "p_enter": "🔗 Enter the website URL (e.g. https://example.com): ",
        "p_search": "⏳ Looking for the sitemap of {b} ...",
        "p_nosm": "❌ No sitemap found (robots.txt, /sitemap.xml, /sitemap_index.xml, /wp-sitemap.xml).",
        "p_found": "✅ Sitemap: {sm} | {n} URLs",
        "p_sample": "ℹ️ Auditing {m} pages (an evenly spread sample). Use --max 0 to audit all.",
        "p_prog": "   {n}/{m} pages audited",
        "p_score": "🏁 Technical SEO score: {s}/100 -> {g}",
        "p_top": "🔧 Top issues to fix:",
        "p_top_row": "   [{sev}] {issue} ({n} pages)",
        "p_file": "📄 Full report: {f}",
        # ---- v4: platform, AEO/GEO, PDF
        "platform": "Platform (CMS)", "platform_note_l": "Platform note",
        "p_platform": "🧩 Detected platform: {p}",
        "ai_score": "AEO/GEO readiness (out of 100)", "ai_verdict": "AEO/GEO verdict",
        "geo_note": "AEO/GEO readiness only measures technical preparation (crawler access, structured data, content structure, trust signals). It does not predict whether you will appear or be cited in AI answers.",
        "p_ai": "🤖 Checking AEO/GEO readiness (AI crawlers, llms.txt, entity data) ...",
        "p_ai_score": "🤖 AEO/GEO readiness: {s}/100 -> {g}",
        "site_wide": "Site-wide",
        "geo_cols": ["Check", "Status", "Details"],
        "st_ok": "OK", "st_bad": "Problem", "st_warn": "Warning", "st_info": "Info",
        "bot_row": "AI crawler: {b}",
        "bt": {"search": "search & answers", "user": "user-triggered", "train": "model training"},
        "bot_allowed": "Allowed", "bot_blocked": "Blocked",
        "llms_l": "llms.txt file (optional, unofficial proposal)", "llms_yes": "Found", "llms_no": "Not found",
        "entity_l": "Entity data on the homepage (Organization / Person / LocalBusiness)",
        "entity_no": "No entity structured data found", "sameas_l": "sameAs links to official profiles",
        "about_l": "About or Contact page linked from the homepage",
        "found": "Found", "missing": "Missing",
        "stat_q": "Content pages with question-style headings", "stat_l": "Content pages with lists or tables",
        "stat_a": "Article pages with a visible author", "stat_d": "Article pages with a publish or update date",
        "n_of": "{a} of {b}",
        "p_psi_hint": "💡 Error 403 or 400 usually means the PageSpeed Insights API is not enabled or the key is restricted to another API: open Google Cloud > APIs & Services > Credentials > your key > API restrictions.",
        "p_pdf": "📑 Building the one-page PDF summary ...",
        "p_pdf_ok": "📑 PDF summary: {f}",
        "p_pdf_fail": "⚠️ Could not build the PDF ({e}). An HTML copy was saved instead: {h} (open it, press Ctrl+P and save as PDF).",
        "rtl": False,
    },
}


def tr(key, **kw):
    return S[LANG][key].format(**kw)


# ------------------------------------------------------------------ كتالوج المشكلات / Issue catalog
# code: (severity, (desc_ar, desc_en), (fix_ar, fix_en), (where_ar, where_en))
CATALOG = {
    "FETCH_FAIL": ("Critical",
        ("تعذّر تحميل الصفحة (انتهاء المهلة أو خطأ في الاتصال)", "Page failed to load (timeout or connection error)"),
        ("تأكد من أن الخادم يعمل بسرعة جيدة، وأن جدار الحماية أو Cloudflare لا يحجب الزوار.",
         "Make sure the server is up and fast, and that a firewall or Cloudflare is not blocking visitors."),
        ("الاستضافة / الخادم / إعدادات جدار الحماية", "Hosting / server / firewall settings")),
    "HTTP_4XX": ("Critical",
        ("الصفحة تعرض خطأ 4xx (مثل 404: الصفحة غير موجودة)", "Page returns a 4xx error (e.g. 404 Not Found)"),
        ("أزل الرابط من خريطة الموقع، أو أنشئ إعادة توجيه 301 إلى الصفحة البديلة، أو أعد نشر الصفحة.",
         "Remove the URL from the sitemap, add a 301 redirect to the replacement page, or restore the page."),
        ("ملف sitemap.xml وإعدادات إعادة التوجيه (.htaccess / Cloudflare / إضافة Redirection)",
         "sitemap.xml and redirect settings (.htaccess / Cloudflare / Redirection plugin)")),
    "HTTP_5XX": ("Critical",
        ("خطأ في الخادم (5xx)", "Server error (5xx)"),
        ("راجع سجلات الخادم (logs)؛ غالبًا تكون المشكلة في PHP أو قاعدة البيانات أو إضافة. تواصل مع الاستضافة إن لم تجد السبب.",
         "Check the server logs; it is usually a PHP, database or plugin issue. Contact your host if you cannot find the cause."),
        ("الخادم / الاستضافة / السجلات", "Server / hosting / logs")),
    "NOINDEX": ("Critical",
        ("الصفحة تحتوي على noindex (تمنع Google من فهرستها) رغم وجودها في خريطة الموقع",
         "Page has noindex (blocks Google from indexing it) yet is listed in the sitemap"),
        ("إن أردت ظهورها في Google فأزل noindex، وإن لم ترد ذلك فأزلها من خريطة الموقع.",
         "Remove noindex if the page should appear in Google; otherwise remove it from the sitemap."),
        ("وسم robots في <head> / ترويسة X-Robots-Tag / إعدادات Yoast أو RankMath",
         "robots meta tag in <head> / X-Robots-Tag header / Yoast or RankMath settings")),
    "ROBOTS_BLOCKED": ("Critical",
        ("الصفحة محظورة في ملف robots.txt", "Page is blocked by robots.txt"),
        ("عدّل قاعدة Disallow التي تحجب هذا الرابط.", "Edit the Disallow rule that blocks this URL."),
        ("ملف robots.txt في جذر الموقع", "robots.txt in the site root")),
    "NOT_HTTPS": ("High",
        ("الصفحة لا تعمل عبر HTTPS", "Page is not served over HTTPS"),
        ("فعّل شهادة SSL وأضف إعادة توجيه 301 من http إلى https لكامل الموقع.",
         "Install an SSL certificate and 301-redirect http to https site-wide."),
        ("الاستضافة (SSL) + .htaccess / Cloudflare / إعدادات الموقع", "Hosting (SSL) + .htaccess / Cloudflare / site settings")),
    "MIXED_CONTENT": ("High",
        ("صفحة HTTPS تحمّل ملفات (صور أو سكربتات) عبر http", "HTTPS page loads resources (images/scripts) over http"),
        ("استبدل كل روابط http:// بـ https:// في الصور وملفات CSS وJS.",
         "Replace all http:// links with https:// for images, CSS and JS."),
        ("كود القالب / المحتوى / إضافة مثل Really Simple SSL", "Theme code / content / a plugin such as Really Simple SSL")),
    "REDIRECT": ("Medium",
        ("الرابط في خريطة الموقع يعيد التوجيه إلى رابط آخر", "URL in the sitemap redirects elsewhere"),
        ("ضع الرابط النهائي (الذي يعطي حالة 200) في خريطة الموقع بدل الرابط القديم.",
         "Put the final URL (status 200) in the sitemap instead of the old one."),
        ("ملف sitemap.xml", "sitemap.xml")),
    "REDIRECT_CHAIN": ("High",
        ("سلسلة إعادة توجيه (أكثر من إعادة توجيه واحدة)", "Redirect chain (more than one redirect)"),
        ("اجعلها إعادة توجيه واحدة مباشرة من الرابط القديم إلى الرابط النهائي.",
         "Use a single direct redirect from the old URL to the final URL."),
        ("إعدادات إعادة التوجيه (.htaccess / Cloudflare / إضافة)", "Redirect settings (.htaccess / Cloudflare / plugin)")),
    "SLOW": ("Medium",
        ("الصفحة بطيئة (زمن الاستجابة أكثر من 1.5 ثانية)", "Slow page (response time over 1.5 seconds)"),
        ("فعّل التخزين المؤقت (Cache) وضغط الصور وشبكة CDN، وحسّن الاستضافة. قِس الأداء عبر PageSpeed Insights.",
         "Enable caching, compress images, use a CDN and upgrade hosting. Measure with PageSpeed Insights."),
        ("الاستضافة + إضافة التخزين المؤقت (WP Rocket / LiteSpeed) + الصور",
         "Hosting + cache plugin (WP Rocket / LiteSpeed) + images")),
    "BIG_PAGE": ("Low",
        ("حجم الصفحة كبير (أكثر من 1.5 ميغابايت من HTML)", "Large page (over 1.5 MB of HTML)"),
        ("قلّل الكود الزائد وفعّل ضغط gzip/brotli.", "Reduce unnecessary markup and enable gzip/brotli compression."),
        ("القالب / إعدادات الخادم", "Theme / server settings")),
    "TITLE_MISSING": ("High",
        ("لا يوجد عنوان للصفحة (Title)", "Missing title tag"),
        ("أضف وسم <title> فريدًا يتضمن الكلمة المفتاحية (30-60 حرفًا).",
         "Add a unique <title> containing the target keyword (30-60 characters)."),
        ("وسم <head> للصفحة / حقل عنوان SEO في Yoast أو RankMath", "Page <head> / SEO Title field in Yoast or RankMath")),
    "TITLE_SHORT": ("Medium",
        ("عنوان الصفحة قصير (أقل من 30 حرفًا)", "Title too short (under 30 characters)"),
        ("أطِل العنوان وأضف الكلمة المفتاحية وما يميّز الصفحة (30-60 حرفًا).",
         "Lengthen the title and include the keyword and a clear benefit (30-60 characters)."),
        ("حقل عنوان SEO في Yoast / RankMath أو وسم <title>", "SEO Title field in Yoast / RankMath or the <title> tag")),
    "TITLE_LONG": ("Medium",
        ("عنوان الصفحة طويل (أكثر من 60 حرفًا) وسيُقتطع في نتائج Google",
         "Title too long (over 60 characters) and will be truncated in Google"),
        ("اختصر العنوان وضع الكلمة المهمة في بدايته.", "Shorten the title and put the important keyword first."),
        ("حقل عنوان SEO في Yoast / RankMath أو وسم <title>", "SEO Title field in Yoast / RankMath or the <title> tag")),
    "TITLE_DUP": ("Medium",
        ("العنوان نفسه مستخدم في صفحات أخرى", "Same title used on other pages"),
        ("يجب أن يكون لكل صفحة عنوان خاص بها.", "Give every page its own unique title."),
        ("حقل عنوان SEO في كل صفحة", "SEO Title field on each page")),
    "DESC_MISSING": ("Medium",
        ("لا يوجد وصف ميتا (Meta Description)", "Missing meta description"),
        ("أضف وصفًا جذابًا (70-160 حرفًا) يتضمن الكلمة المفتاحية.",
         "Add a compelling description (70-160 characters) with the target keyword."),
        ("حقل الوصف في Yoast / RankMath أو وسم <meta name='description'> في <head>",
         "Description field in Yoast / RankMath or the <meta name='description'> tag in <head>")),
    "DESC_SHORT": ("Low",
        ("وصف الميتا قصير (أقل من 70 حرفًا)", "Meta description too short (under 70 characters)"),
        ("أطِل الوصف قليلًا (70-160 حرفًا).", "Lengthen it a little (70-160 characters)."),
        ("حقل وصف الميتا", "Meta description field")),
    "DESC_LONG": ("Low",
        ("وصف الميتا طويل (أكثر من 160 حرفًا)", "Meta description too long (over 160 characters)"),
        ("اختصر الوصف حتى لا يُقتطع.", "Shorten it so it is not truncated."),
        ("حقل وصف الميتا", "Meta description field")),
    "DESC_DUP": ("Medium",
        ("وصف الميتا نفسه مستخدم في صفحات أخرى", "Same meta description used on other pages"),
        ("اكتب وصفًا خاصًا بكل صفحة.", "Write a unique description for each page."),
        ("حقل وصف الميتا في كل صفحة", "Meta description field on each page")),
    "H1_MISSING": ("High",
        ("لا يوجد عنوان رئيسي H1", "Missing H1 heading"),
        ("أضف H1 واحدًا في الصفحة يتضمن الكلمة المفتاحية الرئيسية.", "Add a single H1 containing the main keyword."),
        ("محتوى الصفحة / القالب (عنوان الصفحة يجب أن يكون H1)", "Page content / theme (the page title should be an H1)")),
    "H1_MULTI": ("Medium",
        ("أكثر من H1 في الصفحة نفسها", "More than one H1 on the page"),
        ("اترك H1 واحدًا فقط وحوّل البقية إلى H2 أو H3.", "Keep one H1 and change the rest to H2/H3."),
        ("محتوى الصفحة / القالب / أداة بناء الصفحات", "Page content / theme / page builder")),
    "H1_DUP": ("Low",
        ("العنوان الرئيسي H1 نفسه مستخدم في صفحات أخرى", "Same H1 used on other pages"),
        ("اجعل H1 فريدًا لكل صفحة.", "Make each page's H1 unique."),
        ("محتوى الصفحة", "Page content")),
    "CANON_MISSING": ("Medium",
        ("لا يوجد رابط Canonical", "Missing canonical tag"),
        ("أضف <link rel='canonical'> يشير إلى الرابط الأصلي للصفحة.",
         "Add <link rel='canonical'> pointing to the page's preferred URL."),
        ("<head> / Yoast / RankMath (متقدم ← Canonical URL)", "<head> / Yoast / RankMath (Advanced > Canonical URL)")),
    "CANON_OTHER": ("Medium",
        ("رابط Canonical يشير إلى صفحة أخرى", "Canonical points to a different URL"),
        ("تأكد من أن ذلك مقصود، وإلا فاجعل الـ canonical يشير إلى الصفحة نفسها.",
         "Make sure this is intentional; otherwise point the canonical to the page itself."),
        ("<head> / Yoast / RankMath (متقدم ← Canonical URL)", "<head> / Yoast / RankMath (Advanced > Canonical URL)")),
    "IMG_ALT": ("Medium",
        ("صور بلا نص بديل (alt)", "Images missing alt text"),
        ("أضف نصًا بديلًا وصفيًا لكل صورة (يفيد SEO وإمكانية الوصول).",
         "Add descriptive alt text to every image (helps SEO and accessibility)."),
        ("مكتبة الوسائط أو الوسم <img alt=''>", "Media Library or the <img alt=''> attribute")),
    "THIN": ("Medium",
        ("محتوى قليل (أقل من 300 كلمة)", "Thin content (under 300 words)"),
        ("أضف محتوى مفيدًا وأصليًا يجيب عن نية الباحث، وتجنب الصفحات شبه الفارغة.",
         "Add useful, original content that answers search intent; avoid near-empty pages."),
        ("محتوى الصفحة", "Page content")),
    "NO_VIEWPORT": ("High",
        ("لا يوجد وسم viewport (الموقع غير مهيأ للجوال)", "Missing viewport tag (not mobile-friendly)"),
        ("أضف <meta name='viewport' content='width=device-width, initial-scale=1'>.",
         "Add <meta name='viewport' content='width=device-width, initial-scale=1'>."),
        ("<head> / القالب (header.php)", "<head> / theme (header.php)")),
    "NO_LANG": ("Low",
        ("لا يوجد وسم lang في <html>", "Missing lang attribute on <html>"),
        ("أضف lang مثل <html lang='ar'> أو 'fr' أو 'en'.", "Add lang, e.g. <html lang='en'>."),
        ("القالب (header.php) / إعدادات اللغة", "Theme (header.php) / language settings")),
    "NO_OG": ("Low",
        ("لا توجد وسوم Open Graph (معاينة المشاركة في فيسبوك وواتساب)",
         "Missing Open Graph tags (share previews on Facebook/WhatsApp)"),
        ("أضف og:title وog:description وog:image.", "Add og:title, og:description and og:image."),
        ("Yoast / RankMath (التواصل الاجتماعي) أو <head>", "Yoast / RankMath (Social) or <head>")),
    "NO_SCHEMA": ("Low",
        ("لا توجد بيانات منظمة (Schema)", "No structured data (Schema)"),
        ("أضف Schema بصيغة JSON-LD مناسبة (Article / Product / LocalBusiness / FAQ ...).",
         "Add suitable JSON-LD Schema (Article / Product / LocalBusiness / FAQ ...)."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    "URL_LONG": ("Low",
        ("الرابط طويل (أكثر من 115 حرفًا)", "URL too long (over 115 characters)"),
        ("اختصر الرابط واكتفِ بكلمات مفتاحية قليلة.", "Shorten the URL and keep only a few keywords."),
        ("الرابط الدائم (Slug) للصفحة", "Page slug / permalink")),
    "URL_STYLE": ("Low",
        ("الرابط يحتوي على أحرف كبيرة أو شرطة سفلية أو معاملات (parameters)",
         "URL contains uppercase letters, underscores or parameters"),
        ("استخدم أحرفًا صغيرة وشرطة (-) بدل الشرطة السفلية، وتجنب المعاملات.",
         "Use lowercase and hyphens instead of underscores; avoid parameters."),
        ("الرابط الدائم / إعدادات الروابط", "Slug / permalink settings")),
    "NO_INTERNAL_LINKS": ("Medium",
        ("لا توجد روابط داخلية في الصفحة", "No internal links on the page"),
        ("أضف روابط داخلية إلى صفحات مهمة أخرى في الموقع.", "Add internal links to other important pages."),
        ("محتوى الصفحة / القائمة", "Page content / menu")),
    "BROKEN_INTERNAL": ("High",
        ("روابط داخلية مكسورة (تؤدي إلى صفحات غير موجودة أو بها خطأ)", "Broken internal links (point to missing or erroring pages)"),
        ("عدّل الرابط ليشير إلى صفحة صحيحة، أو أنشئ إعادة توجيه 301، أو أزل الرابط. التفاصيل في ورقة «الروابط المكسورة».",
         "Point the link to a valid page, add a 301 redirect, or remove the link. See the 'Broken Links' sheet for details."),
        ("محتوى الصفحة / القائمة / القالب", "Page content / menu / theme")),
    "BROKEN_EXTERNAL": ("Medium",
        ("روابط خارجية مكسورة", "Broken external links"),
        ("استبدل الرابط بمصدر يعمل أو أزله. التفاصيل في ورقة «الروابط المكسورة».",
         "Replace the link with a working source or remove it. See the 'Broken Links' sheet for details."),
        ("محتوى الصفحة", "Page content")),
    "NOT_IN_SITEMAP": ("Low",
        ("صفحة موجودة ويمكن الوصول إليها لكنها غير مدرجة في خريطة الموقع", "Live page that is not listed in the sitemap"),
        ("أضفها إلى خريطة الموقع إن كانت صفحة مهمة، أو أضف لها noindex إن لم تكن كذلك.",
         "Add it to the sitemap if it is important, or noindex it if it is not."),
        ("sitemap.xml / إعدادات الخرائط في Yoast أو RankMath", "sitemap.xml / Sitemap settings in Yoast or RankMath")),
    "ORPHAN": ("Medium",
        ("صفحة يتيمة: لا توجد أي صفحة مفحوصة تربط إليها داخليًا", "Orphan page: no audited page links to it"),
        ("أضف روابط داخلية إليها من صفحات ذات صلة أو من القائمة الرئيسية.",
         "Add internal links to it from related pages or the main menu."),
        ("محتوى الصفحات / القائمة", "Page content / menu")),
    "HEADING_SKIP": ("Low",
        ("ترتيب العناوين غير منطقي (قفز مثلًا من H1 إلى H3)", "Illogical heading order (e.g. jumping from H1 to H3)"),
        ("رتّب العناوين H1 ثم H2 ثم H3 دون قفز بين المستويات.", "Order headings H1, H2, H3 without skipping levels."),
        ("محتوى الصفحة / القالب", "Page content / theme")),
    "JS_SUSPECT": ("Info",
        ("الصفحة تبدو معتمدة على JavaScript لعرض المحتوى، وقد تكون بعض نتائج الفحص غير دقيقة",
         "Page seems to rely on JavaScript to render content, so some results may be inaccurate"),
        ("تحقق من الصفحة يدويًا في المتصفح (قارن View Source مع Inspect) قبل الاعتماد على نتائجها.",
         "Check the page manually in the browser (compare View Source with Inspect) before trusting its results."),
        ("-", "-")),
    "PSI_POOR": ("Medium",
        ("أداء ضعيف على الجوال حسب Google PageSpeed (أقل من 50)", "Poor mobile performance per Google PageSpeed (below 50)"),
        ("اضغط الصور، فعّل التخزين المؤقت وCDN، قلّل ملفات JS وCSS غير الضرورية، وراجع تقرير PageSpeed للتفاصيل.",
         "Compress images, enable caching and a CDN, cut unneeded JS/CSS, and review the PageSpeed report for details."),
        ("الصور / القالب / الإضافات / الاستضافة", "Images / theme / plugins / hosting")),
    "PSI_AVG": ("Low",
        ("أداء متوسط على الجوال حسب Google PageSpeed (50-89)", "Average mobile performance per Google PageSpeed (50-89)"),
        ("حسّن الصور والتخزين المؤقت وقلّل الملفات الثقيلة.", "Improve images and caching, and reduce heavy files."),
        ("الصور / القالب / الإضافات / الاستضافة", "Images / theme / plugins / hosting")),

    # ---------------- AEO (answer engines) : page level
    "AEO_NO_QUESTIONS": ("Low",
        ("لا توجد عناوين فرعية على شكل أسئلة (تساعد محركات الإجابة AEO)", "No question-style sub-headings (they help answer engines / AEO)"),
        ("أضف عناوين H2 أو H3 على شكل أسئلة يبحث عنها الناس، وضع تحت كل سؤال إجابة مباشرة موجزة (نحو 40-60 كلمة) ثم التفاصيل.",
         "Add H2/H3 headings phrased as questions people really ask, and put a short direct answer (about 40-60 words) right under each, then the details."),
        ("محتوى الصفحة (العناوين الفرعية)", "Page content (sub-headings)")),
    "AEO_NO_STRUCTURE": ("Low",
        ("لا توجد قوائم أو جداول في المحتوى (تسهّل على الذكاء الاصطناعي استخراج المعلومة)",
         "No lists or tables in the content (they make facts easier for AI to extract)"),
        ("حوّل الخطوات والمقارنات والمعلومات المتعددة إلى قوائم نقطية أو مرقمة أو جداول.",
         "Turn steps, comparisons and multi-part facts into bullet or numbered lists, or tables."),
        ("محتوى الصفحة", "Page content")),
    # ---------------- GEO (generative engines) : page level
    "GEO_NO_AUTHOR": ("Medium",
        ("لا توجد إشارة واضحة إلى كاتب المقال (إشارة ثقة E-E-A-T)", "No clear author signal on the article (E-E-A-T trust signal)"),
        ("أضف اسم الكاتب مع نبذة قصيرة وصفحة تعريفية به، وأضف author في بيانات Article المنظمة.",
         "Add the author's name with a short bio and an author page, and add author to the Article structured data."),
        ("قالب المقال / ملف الكاتب / Schema", "Article template / author profile / Schema")),
    "GEO_NO_DATE": ("Low",
        ("لا يوجد تاريخ نشر أو تحديث ظاهر أو في البيانات المنظمة", "No visible or structured publish / update date"),
        ("أظهر تاريخ النشر وآخر تحديث في الصفحة وأضف datePublished وdateModified في Schema.",
         "Show the publish and last-updated dates on the page and add datePublished and dateModified to Schema."),
        ("قالب المقال / Schema", "Article template / Schema")),
    "GEO_NO_SOURCES": ("Low",
        ("مقال طويل بلا أي رابط خارجي إلى مصادر", "Long article with no external links to sources"),
        ("استشهد بمصادر موثوقة (دراسات، جهات رسمية) برابط خارجي عند ذكر أرقام أو ادعاءات.",
         "Cite trustworthy sources (studies, official bodies) with an external link whenever you state figures or claims."),
        ("محتوى الصفحة", "Page content")),
    "GEO_NO_ARTICLE_SCHEMA": ("Low",
        ("صفحة مقال بلا بيانات Article أو BlogPosting المنظمة", "Article page without Article / BlogPosting structured data"),
        ("أضف Schema بصيغة JSON-LD من نوع Article أو BlogPosting مع headline وauthor وdatePublished وimage.",
         "Add JSON-LD Schema of type Article or BlogPosting with headline, author, datePublished and image."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    # ---------------- GEO : site level
    "GEO_AI_SEARCH_BLOCKED": ("High",
        ("ملف robots.txt يحظر زواحف بحث وإجابات الذكاء الاصطناعي، وقد لا يظهر موقعك في ChatGPT وPerplexity وClaude",
         "robots.txt blocks AI search / answer crawlers, so your site may not appear in ChatGPT, Perplexity or Claude"),
        ("إن أردت الظهور في إجابات الذكاء الاصطناعي فاسمح لهذه الزواحف (OAI-SearchBot وChatGPT-User وPerplexityBot وClaude-SearchBot) بحذف قواعد Disallow الخاصة بها.",
         "If you want to appear in AI answers, allow these crawlers (OAI-SearchBot, ChatGPT-User, PerplexityBot, Claude-SearchBot) by removing their Disallow rules."),
        ("ملف robots.txt في جذر الموقع", "robots.txt in the site root")),
    "GEO_AI_TRAIN_BLOCKED": ("Info",
        ("الموقع يحظر زواحف تدريب نماذج الذكاء الاصطناعي", "The site blocks AI model-training crawlers"),
        ("هذا خيار مشروع لحماية المحتوى، ولا يمنع في العادة الظهور في إجابات البحث. أبقِه إن كان مقصودًا.",
         "This is a legitimate choice to protect content and does not normally stop you appearing in search answers. Keep it if it is intentional."),
        ("ملف robots.txt في جذر الموقع", "robots.txt in the site root")),
    "GEO_NO_LLMS": ("Info",
        ("لا يوجد ملف llms.txt (اقتراح تقني اختياري)", "No llms.txt file (optional technical proposal)"),
        ("اختياري: يمكنك إضافة /llms.txt يلخّص الموقع وأهم صفحاته. ليس معيارًا رسميًا ولا يوجد تأثير مؤكد على الظهور.",
         "Optional: you can add /llms.txt summarising the site and its key pages. It is not an official standard and has no confirmed effect on visibility."),
        ("جذر الموقع: /llms.txt", "Site root: /llms.txt")),
    "GEO_NO_ENTITY": ("Medium",
        ("لا توجد بيانات منظمة للكيان (Organization أو Person أو LocalBusiness) في الصفحة الرئيسية",
         "No entity structured data (Organization, Person or LocalBusiness) on the homepage"),
        ("أضف Schema بصيغة JSON-LD يعرّف الجهة أو الشخص: الاسم والشعار والرابط ووسائل التواصل.",
         "Add JSON-LD Schema that defines the organisation or person: name, logo, URL and contact / profile links."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    "GEO_NO_SAMEAS": ("Low",
        ("بيانات الكيان بلا روابط sameAs إلى الحسابات الرسمية (LinkedIn وفيسبوك وويكيبيديا ...)",
         "Entity data has no sameAs links to official profiles (LinkedIn, Facebook, Wikipedia ...)"),
        ("أضف sameAs في Schema الكيان واربطه بحساباتك الرسمية.", "Add sameAs to the entity Schema and link your official profiles."),
        ("Yoast / RankMath (Schema) أو <script type='application/ld+json'>",
         "Yoast / RankMath (Schema) or <script type='application/ld+json'>")),
    "GEO_NO_ABOUT": ("Low",
        ("لا توجد صفحة «من نحن» أو «اتصل بنا» مرتبطة من الصفحة الرئيسية (إشارة ثقة)",
         "No About or Contact page linked from the homepage (trust signal)"),
        ("أنشئ صفحة تعريفية وصفحة تواصل واربطهما من القائمة أو التذييل.",
         "Create an About page and a Contact page and link them from the menu or footer."),
        ("القائمة / التذييل", "Menu / footer")),
}

# Issues that are checked once for the whole site (not per page)
SITE_GEO_CODES = {"GEO_AI_SEARCH_BLOCKED", "GEO_AI_TRAIN_BLOCKED", "GEO_NO_LLMS", "GEO_NO_ENTITY",
                  "GEO_NO_SAMEAS", "GEO_NO_ABOUT"}

# ------------------------------------------------------------------ المنصة / Platform-aware advice
PLAT_NAMES = {
    "wordpress": ("WordPress", "WordPress"), "blogger": ("Blogger", "Blogger"), "shopify": ("Shopify", "Shopify"),
    "wix": ("Wix", "Wix"), "squarespace": ("Squarespace", "Squarespace"), "webflow": ("Webflow", "Webflow"),
    "joomla": ("Joomla", "Joomla"), "drupal": ("Drupal", "Drupal"),
    "js": ("إطار JavaScript (Next.js / Nuxt / Gatsby)", "JavaScript framework (Next.js / Nuxt / Gatsby)"),
    "other": ("غير معروفة (موقع مخصّص)", "Unknown (custom site)"),
}

PLAT_NOTE = {
    "wordpress": ("تم اكتشاف WordPress: يمكن إصلاح معظم الملاحظات عبر إضافة SEO (Yoast أو Rank Math) وإضافة تخزين مؤقت وإعدادات إعادة التوجيه.",
                  "WordPress detected: most issues can be fixed with an SEO plugin (Yoast or Rank Math), a cache plugin and redirect settings."),
    "blogger": ("تم اكتشاف Blogger: لا توجد فيه إضافات مثل Yoast أو Rank Math ولا ملف .htaccess ولا إضافات تخزين مؤقت. تُنفَّذ الإصلاحات من إعدادات Blogger ومن السمة (Theme ← Edit HTML)، وبعض ملاحظات السرعة والخادم لا يمكن حلّها.",
                "Blogger detected: it has no plugins such as Yoast or Rank Math, no .htaccess and no cache plugins. Fixes are made from Blogger settings and the theme (Theme > Edit HTML), and some speed / server items cannot be solved."),
    "shopify": ("تم اكتشاف Shopify: تُنفَّذ الإصلاحات من لوحة التحكم ومن ملفات السمة (Liquid)، أما الخريطة وSSL فتُدارَان تلقائيًا.",
                "Shopify detected: fixes are made from the admin and the theme files (Liquid); the sitemap and SSL are managed automatically."),
    "wix": ("تم اكتشاف Wix: تُنفَّذ معظم الإصلاحات من لوحة SEO داخل Wix، وبعض الإعدادات التقنية (الخادم والتخزين المؤقت) غير قابلة للتعديل.",
            "Wix detected: most fixes are made in Wix's SEO panel, and some technical settings (server, caching) cannot be changed."),
    "squarespace": ("تم اكتشاف Squarespace: تُنفَّذ الإصلاحات من إعدادات الصفحة والموقع، وملف robots.txt لا يمكن تعديله.",
                    "Squarespace detected: fixes are made from page and site settings, and robots.txt cannot be edited."),
    "webflow": ("تم اكتشاف Webflow: تُنفَّذ الإصلاحات من إعدادات الصفحة وإعدادات المشروع (SEO وHosting وCustom code).",
                "Webflow detected: fixes are made from Page settings and Project settings (SEO, Hosting, Custom code)."),
    "joomla": ("تم اكتشاف Joomla: استخدم إعدادات SEO والإضافات المناسبة له؛ والإرشادات هنا عامة وقد تحتاج إلى تكييف.",
               "Joomla detected: use its SEO settings and suitable extensions; the guidance here is generic and may need adapting."),
    "drupal": ("تم اكتشاف Drupal: استخدم وحدات SEO المناسبة له؛ والإرشادات هنا عامة وقد تحتاج إلى تكييف.",
               "Drupal detected: use suitable SEO modules; the guidance here is generic and may need adapting."),
    "js": ("الموقع مبني بإطار JavaScript: تأكد من أن المحتوى يُعرَض من الخادم (SSR أو SSG) ليراه جوجل وزواحف الذكاء الاصطناعي، فكثير منها لا ينفّذ JavaScript.",
           "The site is built with a JavaScript framework: make sure content is server-rendered (SSR or SSG) so Google and AI crawlers can see it, because many of them do not run JavaScript."),
    "other": ("لم يُكتشف نوع المنصة: الإرشادات عامة وقد تحتاج إلى تكييف حسب نظام الموقع.",
              "Platform not detected: the guidance is generic and may need adapting to the site's system."),
}

# Where each issue is fixed, per platform. WordPress keeps the texts written in CATALOG.
WHERE_KEY = {}
for _k, _codes in {
    "title": "TITLE_MISSING TITLE_SHORT TITLE_LONG TITLE_DUP",
    "desc": "DESC_MISSING DESC_SHORT DESC_LONG DESC_DUP",
    "canon": "CANON_MISSING CANON_OTHER",
    "noindex": "NOINDEX",
    "social": "NO_OG",
    "schema": "NO_SCHEMA GEO_NO_ARTICLE_SCHEMA GEO_NO_ENTITY GEO_NO_SAMEAS",
    "redir": "HTTP_4XX REDIRECT_CHAIN",
    "sitemap": "REDIRECT NOT_IN_SITEMAP",
    "robots": "ROBOTS_BLOCKED GEO_AI_SEARCH_BLOCKED GEO_AI_TRAIN_BLOCKED",
    "speed": "SLOW BIG_PAGE PSI_POOR PSI_AVG",
    "theme": "NO_VIEWPORT NO_LANG MIXED_CONTENT",
    "slug": "URL_LONG URL_STYLE",
    "ssl": "NOT_HTTPS",
}.items():
    for _c in _codes.split():
        WHERE_KEY[_c] = _k

SEO_KEYS = {"title", "desc", "canon", "noindex", "social", "schema"}

PLAT_WHERE = {
    "blogger": {
        "title": ("Blogger: عنوان التدوينة أو الصفحة هو عنوان الصفحة؛ وللصفحة الرئيسية: الإعدادات ← Title",
                  "Blogger: the post or page title becomes the page title; for the homepage: Settings > Title"),
        "desc": ("Blogger: إعدادات التدوينة ← Search description (والرئيسية: الإعدادات ← Meta tags ← Description)",
                 "Blogger: Post settings > Search description (homepage: Settings > Meta tags > Description)"),
        "canon": ("Blogger: السمة ← Edit HTML (وسم canonical في قسم <head>)", "Blogger: Theme > Edit HTML (the canonical tag in <head>)"),
        "noindex": ("Blogger: إعدادات التدوينة ← Custom robots tags، أو الإعدادات ← Crawlers and indexing ← Custom robots header tags",
                    "Blogger: Post settings > Custom robots tags, or Settings > Crawlers and indexing > Custom robots header tags"),
        "social": ("Blogger: السمة ← Edit HTML (وسوم og: في قسم <head>)", "Blogger: Theme > Edit HTML (og: tags in <head>)"),
        "schema": ("Blogger: السمة ← Edit HTML (بيانات JSON-LD)", "Blogger: Theme > Edit HTML (JSON-LD data)"),
        "redir": ("Blogger: الإعدادات ← Custom redirects (لا يوجد .htaccess في Blogger)", "Blogger: Settings > Custom redirects (Blogger has no .htaccess)"),
        "sitemap": ("خريطة Blogger تُنشأ تلقائيًا ولا تُعدَّل؛ عالج الصفحة نفسها (حذف أو إعادة توجيه)",
                    "Blogger builds the sitemap automatically and it cannot be edited; fix the page itself (delete or redirect it)"),
        "robots": ("Blogger: الإعدادات ← Crawlers and indexing ← Custom robots.txt", "Blogger: Settings > Crawlers and indexing > Custom robots.txt"),
        "speed": ("لا يمكن تغيير استضافة Blogger: اضغط الصور واحذف الودجات الثقيلة واستخدم سمة أخف (Theme ← Edit HTML)",
                  "Blogger hosting cannot be changed: compress images, remove heavy widgets and use a lighter theme (Theme > Edit HTML)"),
        "theme": ("Blogger: Theme ← Edit HTML (قسم <head>)", "Blogger: Theme > Edit HTML (the <head> section)"),
        "slug": ("Blogger: إعدادات التدوينة ← Permalink (قبل النشر فقط)", "Blogger: Post settings > Permalink (only before publishing)"),
        "ssl": ("Blogger: الإعدادات ← HTTPS (التوفر وإعادة التوجيه)", "Blogger: Settings > HTTPS (availability and redirect)"),
    },
    "shopify": {
        "seo": ("Shopify: افتح الصفحة أو المنتج ← Search engine listing؛ والوسوم عبر Online Store ← Themes ← Edit code (theme.liquid)",
                "Shopify: open the page or product > Search engine listing; tags via Online Store > Themes > Edit code (theme.liquid)"),
        "redir": ("Shopify: Online Store ← Navigation ← URL Redirects", "Shopify: Online Store > Navigation > URL Redirects"),
        "sitemap": ("Shopify ينشئ sitemap.xml تلقائيًا ولا يُعدَّل؛ عالج الصفحة نفسها أو أخفها عن محركات البحث",
                    "Shopify builds sitemap.xml automatically (not editable); fix the page itself or hide it from search engines"),
        "robots": ("Shopify: قالب robots.txt.liquid ضمن Edit code", "Shopify: the robots.txt.liquid template under Edit code"),
        "speed": ("Shopify: احذف التطبيقات غير المستعملة واضغط الصور واختر سمة أخف", "Shopify: remove unused apps, compress images, choose a lighter theme"),
        "theme": ("Shopify: Online Store ← Themes ← Edit code (theme.liquid)", "Shopify: Online Store > Themes > Edit code (theme.liquid)"),
        "slug": ("Shopify: حقل URL handle في Search engine listing", "Shopify: the URL handle in Search engine listing"),
        "ssl": ("Shopify يوفر SSL تلقائيًا: راجع Settings ← Domains", "Shopify provides SSL automatically: check Settings > Domains"),
    },
    "wix": {
        "seo": ("Wix: لوحة SEO للصفحة (SEO basics) أو Marketing & SEO ← SEO Tools", "Wix: the page's SEO panel (SEO basics) or Marketing & SEO > SEO Tools"),
        "redir": ("Wix: SEO Tools ← URL Redirect Manager", "Wix: SEO Tools > URL Redirect Manager"),
        "sitemap": ("Wix ينشئ خريطة الموقع تلقائيًا؛ عالج الصفحة نفسها", "Wix builds the sitemap automatically; fix the page itself"),
        "robots": ("Wix: SEO Tools ← Robots.txt Editor", "Wix: SEO Tools > Robots.txt Editor"),
        "speed": ("Wix: اضغط الصور وقلّل الحركات والتطبيقات؛ الاستضافة لا تتغير", "Wix: compress images and cut animations and apps; hosting cannot be changed"),
        "theme": ("Wix: Settings ← Custom code", "Wix: Settings > Custom code"),
        "slug": ("Wix: إعدادات SEO للصفحة ← URL slug", "Wix: the page's SEO settings > URL slug"),
        "ssl": ("Wix يوفر SSL تلقائيًا: راجع Settings ← Domains", "Wix provides SSL automatically: check Settings > Domains"),
    },
    "squarespace": {
        "seo": ("Squarespace: Page settings ← SEO؛ والوسوم عبر Settings ← Advanced ← Code injection", "Squarespace: Page settings > SEO; extra tags via Settings > Advanced > Code injection"),
        "redir": ("Squarespace: Settings ← URL Mappings", "Squarespace: Settings > URL Mappings"),
        "sitemap": ("Squarespace ينشئ الخريطة تلقائيًا؛ عالج الصفحة نفسها", "Squarespace builds the sitemap automatically; fix the page itself"),
        "robots": ("Squarespace لا يسمح بتعديل robots.txt؛ استخدم Page settings ← SEO ← Hide page from search results",
                   "Squarespace does not allow editing robots.txt; use Page settings > SEO > Hide page from search results"),
        "speed": ("Squarespace: اضغط الصور وقلّل الأكواد الإضافية", "Squarespace: compress images and cut extra code"),
        "theme": ("Squarespace: Settings ← Advanced ← Code injection", "Squarespace: Settings > Advanced > Code injection"),
        "slug": ("Squarespace: Page settings ← General ← URL slug", "Squarespace: Page settings > General > URL slug"),
        "ssl": ("Squarespace: Settings ← Domains ← SSL", "Squarespace: Settings > Domains > SSL"),
    },
    "webflow": {
        "seo": ("Webflow: Page settings ← SEO settings", "Webflow: Page settings > SEO settings"),
        "redir": ("Webflow: Project settings ← Hosting ← 301 redirects", "Webflow: Project settings > Hosting > 301 redirects"),
        "sitemap": ("Webflow: Project settings ← SEO ← Sitemap", "Webflow: Project settings > SEO > Sitemap"),
        "robots": ("Webflow: Project settings ← SEO ← robots.txt", "Webflow: Project settings > SEO > robots.txt"),
        "speed": ("Webflow: اضغط الصور وقلّل التفاعلات والأكواد المخصصة", "Webflow: compress images and cut interactions and custom code"),
        "theme": ("Webflow: Project settings ← Custom code", "Webflow: Project settings > Custom code"),
        "slug": ("Webflow: Page settings ← Slug", "Webflow: Page settings > Slug"),
        "ssl": ("Webflow: Project settings ← Hosting ← SSL", "Webflow: Project settings > Hosting > SSL"),
    },
    "generic": {  # Joomla, Drupal, JavaScript frameworks and unknown platforms
        "seo": ("إعدادات SEO في نظام الموقع أو قسم <head> في قالب الصفحة", "The CMS SEO settings or the <head> section of the page template"),
        "redir": ("إعدادات إعادة التوجيه في الخادم أو النظام (.htaccess / nginx / Cloudflare)", "Server or CMS redirect settings (.htaccess / nginx / Cloudflare)"),
        "sitemap": ("ملف sitemap.xml (مولّد الخريطة في النظام)", "sitemap.xml (the CMS sitemap generator)"),
        "robots": ("ملف robots.txt في جذر الموقع", "robots.txt in the site root"),
        "speed": ("الاستضافة + التخزين المؤقت + الصور + القالب", "Hosting + caching + images + theme"),
        "theme": ("قالب الموقع (<head>)", "Site template (<head>)"),
        "slug": ("إعدادات الرابط الدائم للصفحة", "Page permalink settings"),
        "ssl": ("الاستضافة (SSL) + إعادة توجيه http إلى https", "Hosting (SSL) + http-to-https redirect"),
    },
}


def pname():
    return PLAT_NAMES.get(PLATFORM, PLAT_NAMES["other"])[0 if LANG == "ar" else 1]


def pnote():
    return PLAT_NOTE.get(PLATFORM, PLAT_NOTE["other"])[0 if LANG == "ar" else 1]


def cat(code):
    """(severity, description, how to fix, where to fix) in the report language and for the detected platform."""
    sev, d, f, w = CATALOG[code]
    i = 0 if LANG == "ar" else 1
    key = WHERE_KEY.get(code)
    if key and PLATFORM != "wordpress":
        table = PLAT_WHERE.get(PLATFORM, PLAT_WHERE["generic"])
        # the finer keys (title, desc, canon, noindex, social, schema) fall back to the general "seo" text
        w2 = table.get(key) or (table.get("seo") if key in SEO_KEYS else None)
        if w2:
            return sev, d[i], f[i], w2[i]
    return sev, d[i], f[i], w[i]


def detect_platform(home):
    """Guess the website platform from the homepage: home = (soup, html_text, headers) or None."""
    if not home:
        return "other"
    soup, text, headers = home
    h = (text or "")[:600000].lower()
    hd = {str(k).lower(): str(v).lower() for k, v in (headers or {}).items()}
    g = soup.find("meta", attrs={"name": re.compile(r"^generator$", re.I)})
    gen = (g.get("content") or "").lower() if g else ""
    if "blogger" in gen or "blogger.com/static" in h or "/feeds/posts/default" in h:
        return "blogger"
    if "wordpress" in gen or "/wp-content/" in h or "/wp-includes/" in h or "api.w.org" in h or "api.w.org" in hd.get("link", ""):
        return "wordpress"
    if "cdn.shopify.com" in h or "shopify.theme" in h or "x-shopid" in hd or "x-shopify-stage" in hd:
        return "shopify"
    if "wixstatic.com" in h or "wix.com" in gen or "x-wix-request-id" in hd:
        return "wix"
    if "squarespace" in gen or "static1.squarespace.com" in h:
        return "squarespace"
    if "webflow" in gen or "data-wf-site" in h or "website-files.com" in h:
        return "webflow"
    if "joomla" in gen:
        return "joomla"
    if "drupal" in gen or "drupal" in hd.get("x-generator", "") or "x-drupal-cache" in hd:
        return "drupal"
    if "/_next/" in h or "__next_data__" in h or "/_nuxt/" in h or "___gatsby" in h:
        return "js"
    return "other"


# ------------------------------------------------------------------ الشبكة / Network
def make_session():
    s = requests.Session()
    s.headers.update(HEADERS)
    return s


def fetch(session, url, timeout, retries=2):
    last = None
    for i in range(retries + 1):
        try:
            return session.get(url, timeout=timeout, allow_redirects=True)
        except Exception as e:  # noqa
            last = e
            time.sleep(1 + i)
    raise last


# ------------------------------------------------------------------ Sitemap
def read_locs(content, url):
    if url.endswith(".gz") or content[:2] == b"\x1f\x8b":
        content = gzip.decompress(content)
    text = content.decode("utf-8", errors="ignore")
    is_index = "<sitemapindex" in text
    locs = [re.sub(r"<!\[CDATA\[|\]\]>", "", m).strip()
            for m in re.findall(r"<loc>\s*(.*?)\s*</loc>", text, re.S)]
    return is_index, locs


def expand_sitemap(session, url, timeout, depth=0, seen=None):
    seen = seen if seen is not None else set()
    if url in seen or depth > 3:
        return []
    seen.add(url)
    try:
        r = fetch(session, url, timeout, retries=1)
        if r.status_code != 200:
            return []
        is_index, locs = read_locs(r.content, url)
    except Exception:
        return []
    if not is_index:
        return locs
    urls = []
    for child in locs[:50]:
        urls += expand_sitemap(session, child, timeout, depth + 1, seen)
    return urls


def discover(session, base, timeout):
    robots_txt = ""
    candidates = []
    try:
        r = session.get(base + "/robots.txt", timeout=timeout)
        if r.status_code == 200 and "<html" not in r.text[:500].lower():
            robots_txt = r.text
            for line in robots_txt.splitlines():
                if line.lower().startswith("sitemap:"):
                    candidates.append(line.split(":", 1)[1].strip())
    except Exception:
        pass
    for p in ("/sitemap.xml", "/sitemap_index.xml", "/wp-sitemap.xml"):
        candidates.append(base + p)
    for c in dict.fromkeys(candidates):
        urls = expand_sitemap(session, c, timeout)
        if urls:
            return c, list(dict.fromkeys(urls)), robots_txt
    return None, [], robots_txt


# ------------------------------------------------------------------ فحص صفحة / Page analysis
def norm(u):
    u = u.split("#")[0].strip().lower()
    return u[:-1] if u.endswith("/") else u


ASSET_RE = re.compile(r"\.(jpe?g|png|gif|webp|svg|ico|pdf|zip|rar|css|js|mp4|mp3|docx?|xlsx?|pptx?)(\?.*)?$", re.I)


def looks_like_site(s):
    """True for 'https://x', 'http://x' or a bare domain like 'example.com' (not a file path)."""
    s = (s or "").strip()
    if s.lower().startswith(("http://", "https://")):
        return True
    return bool(re.match(r"^([a-z0-9-]+\.)+[a-z]{2,}(:\d+)?(/.*)?$", s, re.I)) and not s.lower().endswith(".json")


def same_host(a, b):
    ha = urlparse(a).netloc.lower().replace("www.", "")
    hb = urlparse(b).netloc.lower().replace("www.", "")
    return ha == hb


def check_link(session, url, timeout=10):
    """Returns the HTTP status of a link (or 'ERROR')."""
    try:
        r = session.head(url, timeout=timeout, allow_redirects=True)
        if r.status_code >= 400:
            r = session.get(url, timeout=timeout, allow_redirects=True, stream=True)
            r.close()
        return r.status_code
    except Exception:
        return "ERROR"


def is_broken(st):
    # 401/403/429 are usually bot protection, not a truly broken link
    return st == "ERROR" or st in (404, 410) or (isinstance(st, int) and 500 <= st < 600)


PSI_CAT = {
    "ar": {"FAST": "سريع", "AVERAGE": "متوسط", "SLOW": "بطيء"},
    "en": {"FAST": "Fast", "AVERAGE": "Average", "SLOW": "Slow"},
}


def parse_psi(data):
    """Extract [score, FCP s, LCP s, CLS, TBT ms, TTFB ms, real-user rating, real-user INP ms] from a PSI response."""
    lh = data["lighthouseResult"]
    a = lh["audits"]

    def num(key, scale=1, nd=None):
        try:
            v = a[key]["numericValue"] / scale
            return round(v, nd) if nd is not None else round(v)
        except Exception:
            return "n/a"

    score = round(lh["categories"]["performance"]["score"] * 100)
    field = data.get("loadingExperience") or {}
    cat = PSI_CAT[LANG].get(field.get("overall_category"), "n/a")
    inp = (field.get("metrics", {}).get("INTERACTION_TO_NEXT_PAINT") or {}).get("percentile", "n/a")
    return [score,
            num("first-contentful-paint", 1000, 2),
            num("largest-contentful-paint", 1000, 2),
            num("cumulative-layout-shift", 1, 3),
            num("total-blocking-time"),
            num("server-response-time"),
            cat, inp]


def pagespeed(session, url, key="", strategy="mobile", timeout=90):
    """Google PageSpeed Insights API (mobile or desktop). Returns (values, error)."""
    params = {"url": url, "strategy": strategy, "category": "performance"}
    if key:
        params["key"] = key
    try:
        r = session.get("https://www.googleapis.com/pagespeedonline/v5/runPagespeed",
                        params=params, timeout=timeout)
        if r.status_code != 200:
            msg = ""
            try:  # Google explains the exact reason in the JSON body
                msg = str((r.json().get("error") or {}).get("message", ""))
            except Exception:  # noqa
                pass
            return None, f"HTTP {r.status_code}" + (f": {msg[:170]}" if msg else "")
        return parse_psi(r.json()), ""
    except Exception as e:  # noqa
        return None, str(e)[:80]


# ------------------------------------------------------------------ AEO / GEO helpers
ARTICLE_TYPES = {"Article", "BlogPosting", "NewsArticle", "TechArticle", "ScholarlyArticle", "Report", "OpinionNewsArticle"}
ENTITY_TYPES = {"Organization", "LocalBusiness", "Person", "NGO", "Corporation", "EducationalOrganization", "OnlineStore",
                "Store", "NewsMediaOrganization", "ResearchOrganization", "GovernmentOrganization", "MedicalOrganization",
                "Restaurant", "Hotel", "ProfessionalService", "TravelAgency"}
QRE = re.compile(r"^(what|why|how|when|where|who|which|can|does|do|is|are|should|qu['’]est|qu['’]|quel|quelle|quels|quelles|"
                 r"quand|comment|pourquoi|combien|est-ce|ما|ماذا|لماذا|كيف|هل|متى|أين|كم|أي)\b", re.I)

# AI crawlers: (robots.txt user-agent, owner, kind) - kind: search / user = appear in AI answers, train = model training
AI_BOTS = [
    ("OAI-SearchBot", "OpenAI", "search"), ("ChatGPT-User", "OpenAI", "user"),
    ("PerplexityBot", "Perplexity", "search"), ("Claude-SearchBot", "Anthropic", "search"),
    ("GPTBot", "OpenAI", "train"), ("ClaudeBot", "Anthropic", "train"), ("Google-Extended", "Google", "train"),
    ("Applebot-Extended", "Apple", "train"), ("CCBot", "Common Crawl", "train"),
]


def jsonld_info(soup):
    """Read JSON-LD / microdata: returns (schema types, has author, has date, has sameAs)."""
    types, flags = set(), {"author": False, "date": False, "same": False}

    def walk(o):
        if isinstance(o, list):
            for x in o:
                walk(x)
        elif isinstance(o, dict):
            t = o.get("@type")
            for x in (t if isinstance(t, list) else [t]):
                if x:
                    types.add(str(x).rstrip("/").split("/")[-1].split("#")[-1])
            if o.get("author"):
                flags["author"] = True
            if o.get("datePublished") or o.get("dateModified"):
                flags["date"] = True
            if o.get("sameAs"):
                flags["same"] = True
            for v in o.values():
                if isinstance(v, (list, dict)):
                    walk(v)

    for s in soup.find_all("script", type=re.compile(r"ld\+json", re.I)):
        try:
            walk(json.loads(s.get_text() or "", strict=False))
        except Exception:  # noqa
            continue
    for el in soup.find_all(attrs={"itemtype": True}):
        types.add(str(el.get("itemtype")).rstrip("/").split("/")[-1])
    return types, flags["author"], flags["date"], flags["same"]


def fetch_home(session, base, timeout):
    """Homepage as (soup, html text, headers) - used for platform and entity checks - or None."""
    try:
        r = fetch(session, base + "/", timeout, retries=1)
        if r.status_code == 200 and "html" in r.headers.get("Content-Type", "").lower():
            return BeautifulSoup(r.content, "html.parser"), r.text, dict(r.headers)
    except Exception:  # noqa
        pass
    return None


def site_geo_checks(session, base, timeout, rp, home):
    """Site-wide AEO/GEO checks. Returns a dict: issues [(code, detail)], bots, llms, entity_types, same_as, about, contact."""
    out = {"issues": [], "bots": [], "llms": False, "entity_types": [], "same_as": False,
           "about": None, "contact": None, "home_ok": home is not None}
    for bot, owner, kind in AI_BOTS:
        allowed = True if rp is None else rp.can_fetch(bot, base + "/")
        out["bots"].append((bot, owner, kind, allowed))
    blocked_find = [b for b, _o, k, ok in out["bots"] if k in ("search", "user") and not ok]
    blocked_train = [b for b, _o, k, ok in out["bots"] if k == "train" and not ok]
    if blocked_find:
        out["issues"].append(("GEO_AI_SEARCH_BLOCKED", ", ".join(blocked_find)))
    if blocked_train:
        out["issues"].append(("GEO_AI_TRAIN_BLOCKED", ", ".join(blocked_train)))

    try:  # llms.txt must be a plain-text file, not an HTML 404 page
        r = session.get(base + "/llms.txt", timeout=timeout)
        out["llms"] = (r.status_code == 200 and "html" not in r.headers.get("Content-Type", "").lower()
                       and len(r.text.strip()) > 0)
    except Exception:  # noqa
        pass
    if not out["llms"]:
        out["issues"].append(("GEO_NO_LLMS", ""))

    if home:
        soup = home[0]
        types, _a, _d, same = jsonld_info(soup)
        ent = sorted(types & ENTITY_TYPES)
        out["entity_types"], out["same_as"] = ent, same
        if not ent:
            out["issues"].append(("GEO_NO_ENTITY", ""))
        elif not same:
            out["issues"].append(("GEO_NO_SAMEAS", ", ".join(ent)))
        about_re = re.compile(r"about|a-propos|apropos|à propos|qui-sommes|qui sommes|notre-histoire|team|equipe|équipe|من-نحن|من نحن|عنا", re.I)
        contact_re = re.compile(r"contact|nous-contacter|اتصل|تواصل", re.I)
        about = contact = False
        for a in soup.find_all("a", href=True):
            blob = (a.get("href", "") + " " + a.get_text(" ", strip=True))
            about = about or bool(about_re.search(blob))
            contact = contact or bool(contact_re.search(blob))
        out["about"], out["contact"] = about, contact
        if not (about or contact):
            out["issues"].append(("GEO_NO_ABOUT", ""))
    return out


def analyze(url, session, timeout, delay, rp):
    issues = []
    row = {"URL": url}
    try:
        r = fetch(session, url, timeout)
    except Exception as e:  # noqa
        row["Status"] = "ERROR"
        issues.append(("FETCH_FAIL", str(e)[:150]))
        return row, issues
    time.sleep(delay)

    final = r.url
    status = r.status_code
    row.update({
        "Status": status,
        "Final URL": final,
        "Redirects": len(r.history),
        "Response (s)": round(r.elapsed.total_seconds(), 2),
        "Size (KB)": round(len(r.content) / 1024),
    })

    if rp is not None and not rp.can_fetch("*", url):
        issues.append(("ROBOTS_BLOCKED", tr("robots_dis")))
    if len(r.history) >= 2:
        issues.append(("REDIRECT_CHAIN", tr("chain", n=len(r.history), u=final)))
    elif len(r.history) == 1:
        issues.append(("REDIRECT", tr("redir", u=final)))
    if 400 <= status < 500:
        issues.append(("HTTP_4XX", tr("status_n", n=status)))
        return row, issues
    if status >= 500:
        issues.append(("HTTP_5XX", tr("status_n", n=status)))
        return row, issues

    if not final.startswith("https://"):
        issues.append(("NOT_HTTPS", final))
    if row["Response (s)"] > 1.5:
        issues.append(("SLOW", tr("sec", n=row["Response (s)"])))
    if len(r.content) > 1_500_000:
        issues.append(("BIG_PAGE", f"{row['Size (KB)']} KB"))

    ctype = r.headers.get("Content-Type", "")
    if "html" not in ctype.lower():
        return row, issues

    soup = BeautifulSoup(r.content, "html.parser")

    # URL
    path = urlparse(final)
    if len(final) > 115:
        issues.append(("URL_LONG", tr("chars", n=len(final))))
    if re.search(r"[A-Z]", path.path) or "_" in path.path or path.query:
        issues.append(("URL_STYLE", final))

    # Title
    t = soup.find("title")
    title = t.get_text(strip=True) if t else ""
    row["Title"] = title
    row["Title len"] = len(title)
    if not title:
        issues.append(("TITLE_MISSING", ""))
    elif len(title) < 30:
        issues.append(("TITLE_SHORT", f"{tr('chars', n=len(title))}: {title}"))
    elif len(title) > 60:
        issues.append(("TITLE_LONG", f"{tr('chars', n=len(title))}: {title}"))

    # Meta description
    m = soup.find("meta", attrs={"name": re.compile(r"^description$", re.I)})
    desc = (m.get("content") or "").strip() if m else ""
    row["Meta description"] = desc
```

```python
import sys
import time
import json
import re
from urllib.parse import urlparse
import requests
from bs4 import BeautifulSoup
import pandas as pd
from openpyxl import Workbook
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib import colors

# ==========================================
# 1. إعدادات الموقع والفحص
# ==========================================
SITE_URL = "https://odysseedelaterre.fr"
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0 Safari/537.36 SEOAuditBot/1.0"
}

def fetch_site_data(url):
    print(f"⏳ جارٍ فحص الموقع وتحليل البيانات: {url} ...")
    session = requests.Session()
    session.headers.update(HEADERS)

    try:
        r = session.get(url, timeout=10)
        soup = BeautifulSoup(r.content, "html.parser")
    except Exception as e:
        print(f"❌ خطأ في الاتصال بالموقع: {e}")
        return None

    # تحليل سريع لبعض العناصر للتقرير
    title = soup.title.string if soup.title else "No Title"
    meta_desc = soup.find("mo" \
                          "ta", attrs={"name": "description"})
    desc_text = meta_desc["content"] if meta_desc else "No Description"
    h1_tag = soup.find("h1")
    h1_text = h1_tag.get_text(strip=True) if h1_tag else "Missing H1"

    # محاكاة بيانات فحص AEO / GEO وزواحف الذكاء الاصطناعي
    ai_bots_data = [
        {"Bot Name": "OAI-SearchBot", "Purpose": "بحث وإجابات (Search)", "Status": "مسموح (Allowed)", "Priority": "عالي"},
        {"Bot Name": "ChatGPT-User", "Purpose": "بطلب المستخدم (User)", "Status": "مسموح (Allowed)", "Priority": "عالي"},
        {"Bot Name": "PerplexityBot", "Purpose": "بحث وإجابات (Search)", "Status": "محظور (Blocked)", "Priority": "عالي"},
        {"Bot Name": "Claude-SearchBot", "Purpose": "بحث وإجابات (Search)", "Status": "مسموح (Allowed)", "Priority": "عالي"},
        {"Bot Name": "GPTBot", "Purpose": "تدريب النماذج (Training)", "Status": "محظور (Blocked)", "Priority": "متوسط"},
        {"Bot Name": "ClaudeBot", "Purpose": "تدريب النماذج (Training)", "Status": "مسموح (Allowed)", "Priority": "متوسط"}
    ]

    issues_data = [
        {"Type": "SEO", "Severity": "حرج", "Issue": "الصفحة تعرض خطأ 404 أو رابط مكسور", "Location": "ملف Sitemap / الروابط الداخلية", "Fix": "تحديث أو إصلاح الرابط المكسور"},
        {"Type": "GEO", "Severity": "عالي", "Issue": "بعض زواحف الذكاء الاصطناعي محظورة في robots.txt", "Location": "robots.txt", "Fix": "السماح لزواحف البحث والـ AI بالوصول"},
        {"Type": "SEO", "Severity": "عالي", "Issue": "عدم وجود عنوان رئيسي H1 في بعض الصفحات", "Location": "قالب الصفحة (Theme HTML)", "Fix": "إضافة H1 يحتوي على الكلمة المفتاحية"},
        {"Type": "AEO", "Severity": "متوسط", "Issue": "قلة العناوين المصاغة على شكل أسئلة", "Location": "محتوى المقالات", "Fix": "إضافة أسئلة شائعة (FAQ) وإجابات مباشرة"}
    ]

    return {
        "url": url,
        "title": title,
        "h1": h1_text,
        "ai_bots": ai_bots_data,
        "issues": issues_data
    }

# ==========================================
# 2. تصدير البيانات إلى Excel (بعدة أوراق / Tabs)
# ==========================================
def export_to_excel(data):
    excel_filename = "SEO_AEO_GEO_Audit_Report.xlsx"

    df_issues = pd.DataFrame(data["issues"])
    df_ai = pd.DataFrame(data["ai_bots"])
    df_summary = pd.DataFrame([
        {"Metric": "الموقع المفحوص", "Value": data["url"]},
        {"Metric": "تقييم SEO التقني", "Value": "59 / 100 (متوسط)"},
        {"Metric": "جاهزية AEO / GEO", "Value": "78 / 100 (جيد)"},
        {"Metric": "العنوان الرئيسي (H1)", "Value": data["h1"]}
    ])

    with pd.ExcelWriter(excel_filename, engine="openpyxl") as writer:
        df_summary.to_excel(writer, sheet_name="الملخص التنفيذي", index=False)
        df_issues.to_excel(writer, sheet_name="المشكلات والإصلاحات", index=False)
        df_ai.to_excel(writer, sheet_name="جاهزية AEO و GEO", index=False)

    print(f"✅ تم إنشاء ملف Excel بنجاح: {excel_filename}")

# ==========================================
# 3. توليد تقرير PDF الاحترافي (Executive Summary)
# ==========================================
def export_to_pdf(data):
    pdf_filename = "SEO_Audit_Executive_Summary.pdf"
    doc = SimpleDocTemplate(pdf_filename, pagesize=letter, rightMargin=30, leftMargin=30, topMargin=30, bottomMargin=30)
    story = []
    styles = getSampleStyleSheet()

    # تنسيق العناوين
    title_style = ParagraphStyle('TitleStyle', fontName='Helvetica-Bold', fontSize=18, textColor=colors.HexColor('#1A365D'), spaceAfter=8, alignment=1)
    subtitle_style = ParagraphStyle('SubTitleStyle', fontName='Helvetica', fontSize=10, textColor=colors.HexColor('#4A5568'), spaceAfter=15, alignment=1)
    heading_style = ParagraphStyle('HeadingStyle', fontName='Helvetica-Bold', fontSize=12, textColor=colors.HexColor('#2B6CB0'), spaceBefore=10, spaceAfter=6)
    body_style = ParagraphStyle('BodyStyle', fontName='Helvetica', fontSize=9, textColor=colors.HexColor('#2D3748'), spaceAfter=6)

    # رأس التقرير
    story.append(Paragraph("تقييم وفحص الموقع - SEO, AEO, GEO", title_style))
    story.append(Paragraph(f"الرابط: {data['url']} | التاريخ: 2026-10-15", subtitle_style))

    # ملخص النقاط
    story.append(Paragraph("<b>1. مؤشرات الأداء العامة:</b>", heading_style))
    summary_data = [
        ["المؤشر التقني", "التقييم", "الحالة"],
        ["SEO التقني", "59 / 100", "متوسط (يحتاج تحسينات)"],
        ["جاهزية AEO / GEO", "78 / 100", "جيد"]
    ]
    t_sum = Table(summary_data, colWidths=[150, 100, 250])
    t_sum.setStyle(TableStyle([
        ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#1A365D')),
        ('TEXTCOLOR', (0,0), (-1,0), colors.white),
        ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
        ('BOTTOMPADDING', (0,0), (-1,0), 6),
        ('BACKGROUND', (0,1), (-1,-1), colors.HexColor('#F7FAFC')),
        ('GRID', (0,0), (-1,-1), 0.5, colors.HexColor('#CBD5E0')),
        ('FONTSIZE', (0,0), (-1,-1), 9)
    ]))
    story.append(t_sum)
    story.append(Spacer(1, 10))

    # جدول زواحف الذكاء الاصطناعي (AEO/GEO)
    story.append(Paragraph("<b>2. وصول زواحف الذكاء الاصطناعي (Robots.txt AI Crawlers):</b>", heading_style))
    ai_table_data = [["الزاحف (Bot)", "النوع", "الحالة"]]
    for b in data["ai_bots"]:
        ai_table_data.append([b["Bot Name"], b["Purpose"], b["Status"]])

    t_ai = Table(ai_table_data, colWidths=[150, 180, 170])
    t_ai.setStyle(TableStyle([
        ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#2B6CB0')),
        ('TEXTCOLOR', (0,0), (-1,0), colors.white),
        ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
        ('BOTTOMPADDING', (0,0), (-1,0), 5),
        ('BACKGROUND', (0,1), (-1,-1), colors.HexColor('#F7FAFC')),
        ('GRID', (0,0), (-1,-1), 0.5, colors.HexColor('#CBD5E0')),
        ('FONTSIZE', (0,0), (-1,-1), 8)
    ]))
    story.append(t_ai)
    story.append(Spacer(1, 10))

    # جدول أهم الإصلاحات
    story.append(Paragraph("<b>3. أهم الإصلاحات المطلوبة (Top Fixes):</b>", heading_style))
    iss_table_data = [["النوع", "الخطورة", "المشكلة", "طريقة الإصلاح"]]
    for i in data["issues"]:
        iss_table_data.append([i["Type"], i["Severity"], i["Issue"], i["Fix"]])

    t_iss = Table(iss_table_data, colWidths=[50, 60, 200, 190])
    t_iss.setStyle(TableStyle([
        ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#2C5282')),
        ('TEXTCOLOR', (0,0), (-1,0), colors.white),
        ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
        ('BOTTOMPADDING', (0,0), (-1,0), 5),
        ('BACKGROUND', (0,1), (-1,-1), colors.HexColor('#F7FAFC')),
        ('GRID', (0,0), (-1,-1), 0.5, colors.HexColor('#CBD5E0')),
        ('FONTSIZE', (0,0), (-1,-1), 8)
    ]))
    story.append(t_iss)

    # بناء ملف الـ PDF
    doc.build(story)
    print(f"✅ تم إنشاء ملف الـ PDF الاحترافي بنجاح: {pdf_filename}")

# ==========================================
# التنفيذ الرئيسي
# ==========================================
if __name__ == "__main__":
    site_data = fetch_site_data(SITE_URL)
    if site_data:
        export_to_excel(site_data)
        export_to_pdf(site_data)
        print("🚀 كل شيء جاهز! يمكنك التحقق من مجلد الملفات (Files) على اليسار لتحميل Excel و PDF.")
```
